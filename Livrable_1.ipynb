### Problème avancée de niveau A (CVRPTW)

1. **Routes autorisées :** les camions circulent sur un réseau et ne peuvent emprunter que les routes existantes

13. **Livraison trop tard :** un client ne doit pas être livré après la fin de son créneau horaire.
14. **Respect de la chronologie :** l'heure d'arrivée chez un client dépend de l'heure de départ du client précédent et du temps de trajet entre les deux.
15. **Minimisation du temps :** l'objectif est de minimiser le temps total des tournées.
16. **Trafic :** les temps de trajet tiennent compte des prévisions de circulation.

# Modélisation formelle et complexité du CVRPTW


Comme énoncé dans les parties précédentes, nous commençons par poser le problème étudié et formaliser ses données. Puis, nous différencierons le problème d'optimisation et le problème de décision. Enfin, nous évaluerons leur complexité afin de guider le choix de nos méthodes de résolution


## Présentation du problème étudié

Nous étudions le **CVRPTW** (*Capacitated Vehicle Routing Problem with Time Windows*), il s'agit du problème de tournées de véhicules avec capacité et fenêtres de temps. 
Nous avons une flotte de camions identiques qui part d'un dépôt unique pour livrer un ensemble de clients sur un réseau routier. Chaque client a une demande et un créneau horaire de livraison. Nous cherchons des tournées qui desservent tous les clients en respectant les contraintes du cahier des charges (nos contraintes), tout en minimisant le temps total des tournées.

Le CVRPTW généralise le problème du voyageur de commerce (TSP) : un camion qui part du dépôt, visite chaque client une seule fois et revient au dépôt décrit un cycle hamiltonien. Le CVRPTW généralise le problème en ajoutant plusieurs camions, une capacité et des fenêtres de temps.

## Des contraintes aux paramètres : formalisation mathématique

Les contraintes du cahier des charges ainsi que les données d'entrée et de sortie présentées dans la partie précédente sont exprimées écrites en langage courant. Afin de pouvoir analyser le problème (complexité, PLNE, algorithme), nous les avons **traduites sous forme d'objets mathématiques**. 
Ainsi, chaque contrainte est modélisée selon l'un de ces trois points suivants :

- un **paramètre** : une donnée de l'instance, connue avant la résolution;
- une **condition de réalisabilité** : une propriété que les tournées proposées doivent vérifier/respecter;
- la **fonction objectif** : le critère mathématique à minimiser.

### Données d'entrée : l'instance I


Une instance du problème est le n-uplet $I = (G, t, q, [a, b], k, Q)$, où :

- $G = (V, E)$ est un graphe orienté, où nous avons : 
    - $V = \{v_0, v_1, \dots, v_n\}$ est l'ensemble des sommets (nos clients) ;
    - $v_0$ est le dépôt, point de départ et d'arrivée des camions ;
    - $N = \{v_1, \dots, v_n\}$ est l'ensemble des $n$ clients à visiter ;
    - $E = \{(v_i, v_j) : v_i, v_j \in V,\ i \ne j\}$ est l'ensemble des arcs, c'est-à-dire des routes autorisées. Toutes les routes sont à double sens : si $(v_i, v_j) \in E$, alors $(v_j, v_i) \in E$. Le graphe est néanmoins orienté, car avec le trafic, la durée d'un trajet peut dépendre du sens de circulation ($t_{ij}(h) \ne t_{ji}(h)$ en général) ;

- $t$ désigne l'ensemble des durées de trajet : pour chaque arc $(i, j) \in E$ et chaque heure de départ $h$, on note $t_{ij}(h)$ la durée du trajet de $i$ à $j$ lorsque le camion part à l'heure $h$. Cette durée est un nombre entier $\mathbb{N}$ (par exemple en minutes). Les poids du graphe sont donc des temps de trajet, et la durée d'un même trajet peut varier selon l'heure de départ, afin de tenir compte des prévisions de trafic par tranche horaire ;

- $q$ désigne les demandes : pour chaque client $i \in N$, on note $q_i$ la quantité à lui livrer, avec $0 \le q_i \le Q$ ;
- $a$ et $b$ désignent les créneaux horaires : pour chaque client $i \in N$, $[a_i, b_i]$ est son créneau de livraison, où $a_i \le b_i$ sont des nombres entiers ;

- $k \in \mathbb{N}^*$ est le nombre maximal de camions disponibles ;

- $Q \in \mathbb{N}$ est la capacité, la même pour tous les camions.

Hypothèses de modélisation, nous estimons que :

- les valeurs numériques sont entières (par exemple des minutes et des unités de marchandise) ;
- les camions partent du dépôt à l'heure $0$ ;
- la durée de livraison sur place n'est pas prise en compte : elle est considérée comme instantanée ;
- le trafic respecte la propriété FIFO : dans nos cas partir plus tard ne fait pas arriver plus tôt.

### Données de sortie : une solution

Une solution $S$ est un ensemble de tournées $R_1, \dots, R_r$. Chaque tournée est une suite $R = (0, v_1, v_2, \dots, v_m, 0)$ de sommets, dont deux sommets consécutifs sont reliés par un arc de $E$. 
<br>
Pour chaque tournée, on calcule l'emploi du temps du camion de proche en proche, c-à-d le temps de parcours client après client. On note :

- $l$ la position d'un client dans la tournée ($v_1$ est le premier client livré, $v_2$ le deuxième, etc.) et $m$ le nombre de clients de la tournée ;
- $D_l$ l'heure à laquelle le camion **quitte** le $l$-ième lieu de la tournée ($D_0$ pour le dépôt) ;
- $Arr(v_l)$ l'heure à laquelle il **arrive** chez le client $v_l$ ;
- $w_{v_l}$ l'heure à laquelle il **livre** le client $v_l$.

Les horaires se calculent alors ainsi :

- **départ du dépôt :** $D_0 = 0$. Le camion quitte le dépôt à l'heure $0$ ;
- **arrivée chez le client $v_l$ :** $Arr(v_l) = D_{l-1} + t_{v_{l-1} v_l}(D_{l-1})$. L'heure d'arrivée est l'heure de départ du lieu précédent, plus la durée du trajet. Cette durée est celle qui correspond à l'heure de départ $D_{l-1}$, car avec le trafic, un même trajet ne dure pas le même temps selon l'heure ;
- **heure de livraison :** $w_{v_l} = \max(a_{v_l},\ Arr(v_l))$. Si le camion arrive avant l'ouverture du créneau ($Arr(v_l) < a_{v_l}$), il attend et livre à l'ouverture $a_{v_l}$. Sinon, il livre dès son arrivée ;
- **départ de chez $v_l$ :** $D_l = w_{v_l}$. La livraison étant considérée comme instantanée, le camion repart dès qu'il a livré ;
- **retour au dépôt :** $T(R) = D_m + t_{v_m 0}(D_m)$. Après le dernier client $v_m$, le camion rentre au dépôt. Comme il était parti à l'heure $0$, $T(R)$ est la **durée totale de la tournée**, temps de trajet et temps d'attente compris.

> **Exemple illustratif**
>
> Considérons la tournée $R = (0, v_1, v_2, 0)$, donc $m = 2$, avec les créneaux $[30, 60]$ pour $v_1$ et $[40, 70]$ pour $v_2$, et les durées de trajet (en minutes) : dépôt $\to v_1$ = 20, $v_1 \to v_2$ = 15, $v_2 \to$ dépôt = 25.
>
> 1. départ du dépôt : $D_0 = 0$ ;
> 2. arrivée chez $v_1$ : $Arr(v_1) = 0 + 20 = 20$ ;
> 3. livraison de $v_1$ : le créneau n'ouvre qu'à 30, le camion attend 10 minutes : $w_{v_1} = \max(30, 20) = 30$ ;
> 4. départ de chez $v_1$ : $D_1 = 30$ ;
> 5. arrivée chez $v_2$ : $Arr(v_2) = 30 + 15 = 45$ ;
> 6. livraison de $v_2$ : le créneau est déjà ouvert, la livraison est immédiate : $w_{v_2} = \max(40, 45) = 45$ ;
> 7. départ de chez $v_2$ : $D_2 = 45$ ;
> 8. retour au dépôt : $T(R) = 45 + 25 = 70$.
>
> La tournée dure **70 minutes** : 60 minutes de trajet et 10 minutes d'attente. Les créneaux sont respectés, puisque $30 \le 60$ et $45 \le 70$. Pour simplifier, les durées de l'exemple ne dépendent pas de l'heure ; avec le trafic, le calcul est identique, seules les durées lues changent.

### Conditions qui doivent être respectées et la fonction objectif

Une solution $S$ est **réalisable** si :

1. chaque client de $N$ apparaît dans exactement une tournée, et une seule fois ;
2. chaque tournée commence et finit au dépôt, et deux sommets consécutifs sont reliés par un arc de $E$ ;
3. le nombre de tournées vérifie $r \le k$ ;
4. pour chaque tournée $R$, $\sum_{i \in R} q_i \le Q$ ;
5. pour chaque client $i$, $a_i \le w_i \le b_i$.

On note $\mathcal{S}(I)$ l'ensemble des solutions réalisables. Le coût d'une solution est le temps total de ses tournées, temps de trajet et temps d'attente compris :

$$f(S) = \sum_{R \in S} T(R)$$

## 4.3. Problème d'optimisation et problème de décision

Une même situation peut être posée de deux façons :

- soit on cherche la **meilleure solution** : c'est un **problème d'optimisation** ;
- soit on demande s'il existe une **solution assez bonne**, avec une réponse « oui » ou « non » : c'est un **problème de décision**.

> Pourquoi cette distinction ? Les classes P, NP et NP-complet ne contiennent que des problèmes de décision. Notre problème, tel que le pose CesiCDP, est un problème d'optimisation (on veut les meilleures tournées). Pour étudier sa complexité, il faut donc d'abord écrire le problème de décision qui lui est associé.

### 4.3.1. Problème de décision : CVRPTW-D

**Données :** une instance $I = (G, t, q, [a,b], k, Q)$ telle que définie au §4.2.1, et un entier $T_{max}$ (le seuil de temps).

**Question :** existe-t-il une solution réalisable $S \in \mathcal{S}(I)$ (§4.2.3), c'est-à-dire au plus $k$ tournées partant du dépôt et y revenant, qui livrent chaque client une seule fois, sans dépasser la capacité $Q$ ni les créneaux $[a_i, b_i]$, et dont le temps total vérifie $f(S) \le T_{max}$ ?

La réponse est « oui » ou « non ». Le seuil $T_{max}$ transforme la recherche du « plus petit temps possible » en une question fermée. Il ne faut pas le confondre avec $k$, qui est le nombre maximal de camions.

### 4.3.2. Problème d'optimisation : CVRPTW-OPT

**Données :** une instance $I = (G, t, q, [a,b], k, Q)$.

**Question :** quelles tournées réalisables $S_o \in \mathcal{S}(I)$ minimisent le temps total $f$ ? (s'il n'existe aucune solution réalisable, le problème n'a pas de solution.)

La valeur $f(S_o)$ est le **temps total optimal**. On peut aussi la voir comme la plus petite valeur de $T_{max}$ pour laquelle CVRPTW-D répond « oui ».

> **Exemple.** Deux camions, cinq clients, et une meilleure solution de 120 minutes au total.
> - CVRPTW-OPT : la réponse est un plan de tournées de 120 minutes.
> - CVRPTW-D avec $T_{max} = 150$ : « existe-t-il des tournées d'au plus 150 minutes ? » Oui.
> - CVRPTW-D avec $T_{max} = 100$ : « existe-t-il des tournées d'au plus 100 minutes ? » Non.



## 4.4. Complexité du problème de décision CVRPTW-D

L'objectif de cette partie est de **démontrer que CVRPTW-D est NP-complet**. Nous rappelons d'abord les définitions utilisées, puis nous appliquons la méthode en deux étapes.

### 4.4.1. Définitions et méthode

- **Taille d'une instance.** La taille $|x|$ d'une instance $x = (I, T_{max})$ est le nombre de symboles nécessaires pour l'écrire : le graphe $G$ (matrice d'adjacence de $(n+1)^2$ cases), la table des durées $t_{ij}(h)$ (une valeur par arc et par tranche horaire, soit au plus $(n+1)^2 P$ valeurs), les demandes, les créneaux, $k$, $Q$ et $T_{max}$, tous les entiers étant écrits en binaire. En particulier $n \le |x|$ et $P \le |x|$.
- **Classe P.** Ensemble des problèmes de décision résolus par un algorithme de temps polynomial en $|x|$.
- **Classe NP.** Un problème de décision $B$ appartient à NP s'il existe un algorithme de vérification $V$ de temps polynomial en $|x|$ et un polynôme $p$ tels que, pour toute instance $x$ : la réponse est « oui » $\iff$ il existe un certificat $c$ avec $|c| \le p(|x|)$ tel que $V(x, c)$ accepte.
- **Réduction polynomiale.** $A \le_p B$ s'il existe une fonction $\varphi$, calculable en temps polynomial, qui transforme toute instance $x$ de $A$ en une instance $\varphi(x)$ de $B$ telle que : $x$ a pour réponse « oui » $\iff$ $\varphi(x)$ a pour réponse « oui ». La relation $\le_p$ est **transitive** : la composée de deux fonctions calculables en temps polynomial l'est aussi.
- **NP-difficile, NP-complet.** $B$ est NP-difficile si tout problème $A \in$ NP vérifie $A \le_p B$. Il est NP-complet s'il est de plus dans NP. Par transitivité, pour montrer que $B$ est NP-difficile, il suffit de trouver un problème $A$ **déjà NP-complet** tel que $A \le_p B$.

> **Théorème.** CVRPTW-D est NP-complet.

La démonstration comporte deux étapes : (1) CVRPTW-D $\in$ NP ; (2) CVRPTW-D est NP-difficile.

### 4.4.2. Étape 1 : CVRPTW-D appartient à NP

**Certificat.** Le certificat $c$ est la liste des tournées non vides $R_1, \dots, R_r$, chacune donnée par la suite de ses sommets (dépôt, clients dans l'ordre de visite, dépôt). Une solution réalisable livre chaque client une seule fois (cahier des charges, point 5) : elle a donc $r \le n$ tournées et au plus $n + 2r \le 3n$ sommets. L'algorithme $V$ rejette tout certificat plus long. Chaque sommet s'écrit avec $\lceil \log_2(n+1) \rceil$ bits, donc $|c| = O(n \log n)$, ce qui est polynomial en $|x|$.

**Algorithme de vérification $V(x, c)$.** Soit $P$ le nombre de tranches horaires de la table de trafic.

| Étape | Vérification | Méthode | Coût |
|---|---|---|---|
| 1 | Chaque tournée commence et finit au dépôt, et deux sommets consécutifs sont reliés par un arc de $E$ | parcours des tournées, lecture de la matrice d'adjacence | $O(n)$ |
| 2 | Chaque client de $N$ apparaît exactement une fois | tableau de marquage de taille $n$ | $O(n)$ |
| 3 | $r \le k$ | comptage | $O(1)$ |
| 4 | Pour chaque tournée, $\sum_{i \in R} q_i \le Q$ | somme des demandes tournée par tournée | $O(n)$ |
| 5 | $w_i \le b_i$ pour chaque client | calcul des horaires $Arr$, $w$, $D$ par les formules du §4.2.2 ; $t_{ij}(h)$ se lit dans la table des $P$ tranches par recherche dichotomique | $O(n \log P)$ |
| 6 | $\sum_R T(R) \le T_{max}$ | addition des durées $T(R)$ (attentes comprises) puis comparaison | $O(n)$ |

**Lemme 1 (correction de $V$).** $V(x, c)$ accepte si et seulement si $c$ est une solution réalisable $S \in \mathcal{S}(I)$ de coût $f(S) \le T_{max}$.

*Démonstration.* Les formules du §4.2.2 définissent sans ambiguïté les horaires $Arr$, $w$, $D$ et la durée $T(R)$ de chaque tournée : les valeurs calculées par $V$ sont donc celles de la définition. Les étapes 1, 2, 3, 4 et 5 testent chacune exactement l'une des cinq conditions de réalisabilité du §4.2.3 (l'étape 5 teste la seule borne $w_i \le b_i$, car $a_i \le w_i$ est automatique puisque $w_i = \max(a_i, Arr(i))$). L'étape 6 teste $f(S) \le T_{max}$. $V$ accepte donc exactement quand toutes les conditions sont vraies.

**Lemme 2 (temps de $V$).** $V$ s'exécute en temps polynomial en $|x|$.

*Démonstration.* Soit $M$ le plus grand entier de l'instance. Comme $D_l = \max(a_{v_l}, D_{l-1} + t)$, on a $D_l \le (n+1)M$ pour tout $l$, et $\sum_R T(R) \le n(n+1)M$. Tous les nombres manipulés ont donc au plus $\log_2 M + 2\log_2(n+1) + 2$ bits : leur taille est polynomiale. Les étapes 1 à 6 effectuent $O(n \log P)$ opérations (additions, comparaisons, lectures de tables) sur de tels nombres, chacune en temps polynomial.

**Conclusion de l'étape 1.** Si la réponse à $x$ est « oui », il existe une solution réalisable de coût $\le T_{max}$ : c'est un certificat de taille polynomiale que $V$ accepte (Lemme 1). Réciproquement, si $V$ accepte un certificat, la réponse est « oui » (Lemme 1). $V$ est polynomial (Lemme 2). Donc **CVRPTW-D $\in$ NP**.

### 4.4.3. Étape 2 : CVRPTW-D est NP-difficile

Nous partons du problème du **cycle hamiltonien (HC)**, NP-complet (Karp, 1972 ; Garey et Johnson, 1979) :

> **HC.** *Données :* un graphe non orienté $G = (V, E)$ à $n \ge 3$ sommets. *Question :* $G$ possède-t-il un cycle passant exactement une fois par chaque sommet ?

Nous passons par le voyageur de commerce en version décision, qui est le cas particulier du CVRPTW avec un seul camion :

> **TSP-D.** *Données :* $m \ge 3$ villes, un graphe complet, des distances entières symétriques $d_{ij} = d_{ji} \ge 0$ et un entier $K$. *Question :* existe-t-il un cycle passant exactement une fois par chaque ville, de longueur au plus $K$ ?

Nous montrons $\text{HC} \le_p \text{TSP-D}$ (Lemme 3), puis $\text{TSP-D} \le_p \text{CVRPTW-D}$ (Lemme 4). Attention au sens : on part du problème connu pour arriver au problème étudié, jamais l'inverse. Chaque lemme établit trois points : : la transformation est polynomiale, les réponses « oui » sont équivalentes dans les deux sens, et le problème de départ est connu comme NP-complet.

**Lemme 3.** $\text{HC} \le_p \text{TSP-D}$.

*Construction.* À partir de $G = (V, E)$, on construit le graphe complet sur $V$ ($m = n$ villes) avec $d_{ij} = 1$ si $\{i,j\} \in E$ et $d_{ij} = 2$ sinon, et on pose $K = n$. Ces distances sont des entiers symétriques positifs : c'est bien une instance de TSP-D.

*Polynomial.* Il suffit d'examiner chacune des $n(n-1)/2$ paires de sommets : $O(n^2)$.

*Équivalence.*
- ($\Rightarrow$) Si $G$ a un cycle hamiltonien, il utilise $n$ arêtes de $E$, de poids $1$ dans le graphe construit. Ce cycle passe par toutes les villes et sa longueur est $n \le K$ : la réponse est « oui ».
- ($\Leftarrow$) Si le graphe construit a un cycle passant par toutes les villes, de longueur $\le n$, ce cycle utilise exactement $n$ arêtes, chacune de poids $\ge 1$. Pour que la somme ne dépasse pas $n$, toutes pèsent exactement $1$ : elles appartiennent toutes à $E$. C'est un cycle hamiltonien de $G$. 

**Lemme 4.** $\text{TSP-D} \le_p \text{CVRPTW-D}$.

*Construction.* À partir d'une instance de TSP-D ($m$ villes, distances $d_{ij}$, seuil $K$),on construit l'instance $(I, T_{max})$ de CVRPTW-D suivante, qui neutralise tout ce que le CVRPTW ajoute au TSP :

| Élément | Valeur choisie | Effet |
|---|---|---|
| Sommets | les $m$ villes ; la ville 1 devient le dépôt $v_0$ ; les $n = m - 1$ autres sont les clients | mêmes villes |
| Arcs $E$ | toutes les paires, dans les deux sens | graphe complet comme dans TSP-D |
| Durées $t_{ij}(h)$ | $d_{ij}$ pour toute heure $h$ (une seule tranche, $P = 1$) | pas de trafic |
| Demandes et capacité | $q_i = 1$ pour tout client, $Q = n$ | la capacité ne limite jamais |
| Créneaux $[a_i, b_i]$ | $[0, K]$ pour tout client | aucune attente possible |
| Nombre de camions | $k = 1$ | une seule tournée |
| Seuil | $T_{max} = K$ | même seuil |

*Validité.* C'est bien une instance de CVRPTW-D : les valeurs sont des entiers, $0 \le q_i = 1 \le Q = n$, $a_i = 0 \le b_i = K$, $k = 1 \in \mathbb{N}^*$, et l'hypothèse FIFO est vérifiée car les durées sont constantes (partir plus tard ne fait pas arriver plus tôt).

*Polynomial.* La transformation recopie les $m$ villes, les $m(m-1)/2$ distances (utilisées dans les deux sens) et $K$, puis fixe quelques constantes : $O(m^2)$. Elle n'énumère aucune tournée.

*Équivalence.*
- ($\Rightarrow$) Soit un cycle passant par les $m$ villes, de longueur $L \le K$. On le fait partir de la ville 1 : on obtient $R = (0, u_1, \dots, u_n, 0)$. Elle est réalisable : chaque client apparaît une fois ; tous les arcs existent (graphe complet) ; $r = 1 \le k$ ; la charge vaut $n \le Q$ ; comme $a_i = 0$ il n'y a aucune attente, donc $Arr(u_l)$ est la somme des $l$ premières distances, d'où $Arr(u_l) \le L \le K = b_{u_l}$. Son temps total est $T(R) = L \le K = T_{max}$ : la réponse est « oui ».
- ($\Leftarrow$) Soit une solution réalisable $S$ avec $f(S) \le K$. Il y a au moins un client à servir et $r \le k = 1$ : $S$ est une unique tournée $R = (0, u_1, \dots, u_n, 0)$. Elle livre chaque client une seule fois et revient au dépôt : c'est un cycle passant exactement une fois par chaque ville. Comme $a_i = 0$ il n'y a aucune attente, et comme les durées sont constantes, $T(R)$ est exactement la somme des distances le long de ce cycle. Sa longueur est donc $T(R) = f(S) \le K$ : la réponse est « oui ». 

**Conclusion de l'étape 2.** Par le Lemme 3 et le Lemme 4, et par transitivité de $\le_p$ :
$$\text{HC} \le_p \text{TSP-D} \le_p \text{CVRPTW-D}.$$
HC est NP-complet, donc **CVRPTW-D est NP-difficile**. Intuitivement, le TSP est un cas particulier du CVRPTW (un seul camion, une capacité qui ne limite rien, des créneaux assez larges, pas de trafic) : les contraintes supplémentaires (capacité, créneaux, plusieurs camions, trafic) ne peuvent pas rendre le problème plus facile.

### 4.4.4. Conclusion

CVRPTW-D appartient à NP (étape 1) et il est NP-difficile (étape 2). **CVRPTW-D est donc NP-complet.**

## 4.5. Complexité du problème d'optimisation CVRPTW-OPT

CVRPTW-OPT n'est pas un problème de décision (§4.3) : on le situe donc par rapport à CVRPTW-D, avec les Propositions 1 et 2 du §4.3.3.

**Définition (extension de la NP-difficulté aux problèmes d'optimisation).** Un problème $B$ est **NP-difficile** si tout problème $A \in$ NP peut être résolu par un algorithme de temps polynomial qui a le droit d'appeler un nombre polynomial de fois un algorithme résolvant $B$. On note $A \le_T B$. Pour un problème de décision, c'est exactement la définition du §4.4.1 (une réduction $\le_p$ est le cas particulier d'un seul appel dont on renvoie la réponse).

> **Théorème 1.** CVRPTW-OPT est NP-difficile.

*Démonstration.* Soit $A \in$ NP. D'après le §4.4, $A \le_p$ CVRPTW-D : toute instance $x$ de $A$ se transforme en temps polynomial en une instance $(I, T_{max})$ de CVRPTW-D de même réponse. D'après la Proposition 1, cette instance se résout avec un seul appel à un algorithme pour CVRPTW-OPT. Donc $A \le_T$ CVRPTW-OPT. 

> **Corollaire.** S'il existait un algorithme polynomial pour CVRPTW-OPT, tout problème de NP serait résolu en temps polynomial (la composition de polynômes est un polynôme) : on aurait $P = NP$. En l'absence de preuve que $P = NP$, on ne peut pas supposer qu'un tel algorithme existe.

**CVRPTW-OPT n'appartient pas à NP**, puisque ce n'est pas un problème de décision. On sait vérifier rapidement qu'une solution est réalisable et calculer son coût, mais pas qu'elle est **optimale** : il faudrait prouver que CVRPTW-D répond « non » pour $T_{max} = f(S_o) - 1$. Un certificat court pour une réponse « non » placerait CVRPTW-D dans co-NP, donc entraînerait $NP = \text{co-NP}$, une conjecture ouverte que l'on pense fausse.

**Pas plus difficile que CVRPTW-D.** D'après la Proposition 2, CVRPTW-OPT se résout avec un nombre polynomial d'appels à CVRPTW-D : il est exactement aussi difficile que lui.

### Bilan de la démarche

$$\text{HC (NP-complet)} \;\le_p\; \text{TSP-D (NP-complet)} \;\le_p\; \text{CVRPTW-D (NP-complet)} \;\le_T\; \text{CVRPTW-OPT (NP-difficile)}$$

| Problème | Question posée | Catégorie |
|---|---|---|
| Plus court chemin | Existe-t-il un chemin de durée au plus $K$ ? | P |
| Cycle hamiltonien | Existe-t-il un cycle passant exactement une fois par chaque sommet ? | NP-complet |
| Voyageur de commerce (décision) | Existe-t-il un circuit passant une fois par chaque ville, de longueur au plus $K$ ? | NP-complet |
| Voyageur de commerce (optimisation) | Quel est le circuit le plus court ? | NP-difficile |
| **CVRPTW-D** (décision) | Existe-t-il des tournées d'au plus $k$ camions de capacité $Q$, qui livrent chaque client une fois dans son créneau, avec un temps total $\le T_{max}$ ? | **NP-complet** (§4.4) |
| **CVRPTW-OPT** (optimisation) | Quelles tournées réalisables minimisent le temps total ? | **NP-difficile** (§4.5) |

### Conséquences pour la suite du projet

Si $P \ne NP$, aucun algorithme polynomial ne peut résoudre CVRPTW-OPT de façon exacte sur toutes les instances. Or le cahier des charges demande de traiter des instances pouvant atteindre plusieurs milliers de villes. Il faut donc choisir la méthode selon la taille de l'instance et le temps de calcul disponible :

| Méthode | Principe | Intérêt et limite pour notre projet |
|---|---|---|
| **Algorithme exact** (par exemple résolution du PLNE) | trouve la solution optimale | utile sur de petites instances, mais le temps de calcul explose avec le nombre de clients |
| **Heuristique / métaheuristique** (construction puis amélioration locale, recuit simulé, recherche tabou...) | cherche rapidement une bonne solution, sans garantie d'optimalité | adaptée aux grandes instances ; la qualité se mesure expérimentalement |
| **Algorithme d'approximation** | solution proche de l'optimum avec une garantie mathématique | peu de garanties connues avec des fenêtres de temps, qui rendent difficile même la recherche d'une tournée simplement réalisable |

Nous orientons donc notre travail vers une **méthode heuristique ou métaheuristique** pour les grandes instances, et nous garderons la résolution exacte (PLNE) comme référence sur de petites instances, afin de mesurer l'écart de qualité et de temps de calcul. Le choix précis de l'algorithme est étudié dans la section suivante.

*Source utilisée pour la méthode : cours de NP-complétude, https://perso.lisn.upsaclay.fr/jcohen/documents/enseignement/Cours-NP-completude.pdf*

Après avoir étudié la complexité théorique de notre problème et démontré son appartenance à la classe NP, nous allons maintenant nous intéresser à sa formulation mathématique sous la forme d’un Programme Linéaire en Nombres Entiers (PLNE), afin de représenter formellement les décisions à prendre, l’objectif à optimiser ainsi que les différentes contraintes du CVRPTW.

# Modélisation mathématique : programme linéaire en nombres entiers (PLNE)

Cette partie présente la formulation mathématique de notre problème de tournées de véhicules avec contraintes de capacité et fenêtres temporelles (CVRPTW) sous la forme d'un Programme Linéaire en Nombres Entiers Mixte (PLNE). L'objectif est de traduire les différentes conditions du problème en un ensemble d'équations et d'inéquations mathématiques.

Pour cela, nous définissons dans un premier temps les ensembles et les paramètres nécessaires à la modélisation, puis les variables de décision permettant de représenter les choix effectués par les véhicules, notamment les déplacements entre les différents sommets et les horaires de livraison. Nous établissons ensuite la fonction objectif, qui vise à minimiser la durée totale des tournées, ainsi que l'ensemble des contraintes garantissant la validité des solutions : respect des capacités des véhicules, satisfaction des demandes des clients, respect des fenêtres temporelles et cohérence des itinéraires.

Ce modèle constitue une représentation formelle et rigoureuse du problème étudié. Il servira notamment de référence pour vérifier la faisabilité des solutions obtenues et pourra être utilisé pour résoudre de manière exacte les instances de petite taille.
<br>

Avant de formuler le modèle mathématique, il est nécessaire de définir les ensembles et les paramètres qui caractérisent notre problème. Ils permettent de représenter le réseau routier, les véhicules disponibles, les clients à desservir ainsi que les différentes données nécessaires à l'organisation des tournées.

## Variables de décision

Les variables de décision permettent de représenter les choix effectués par le modèle afin de construire les tournées de livraison. Elles indiquent notamment quels trajets sont empruntés par les véhicules, quels clients leur sont affectés et à quels moments les livraisons sont effectuées. Nous les avons représentées comme ceci : <br>

$x_{ijk} \in$ {0, 1} : vaut $1$ si le camion $k$ emprunte la route $(i, j)$, c'est-à-dire que le camion va de $i$ à $j$ ; sinon sera noté 0


$y_{ik} \in$ {0, 1} : vaut $1$ si le camion est affecté au client


$w_{ik}$ : correspond à l'heure de livraison du client


$T_{max}$ : heure de retour du camion $k$ au dépôt, correspond à la durée de sa tournée puisqu'il part à l'heure $0$

**$T_k$, pour $k \in K$ : la durée de la tournée.** C'est la quantité $T(R)$ du §4.2.2 pour la tournée du camion $c$, temps d'attente compris. Elle vaut $0$ pour un camion qui ne sort pas. Elle sert uniquement à écrire la fonction objectif de façon linéaire.



Une fois les variables de décision définies, nous pouvons exprimer l'objectif principal de notre problème sous la forme d'une fonction mathématique. Celle-ci permet d'évaluer le coût associé à une solution et de déterminer quelle organisation des tournées doit être privilégiée.

## Fonction objectif

$$\min \  \sum_{k \in K} \sum_{(i,j) \in V} t_{ij}  \ x_{ijk}$$


Nous souhaitons minimiser la somme des durées des tournées de tous les camions. C'est exactement la fonction $f(S) = \sum_{R \in S} T(R)$ du §4.2.3, qui traduit la contrainte 16 (« minimiser le temps total des tournées »).


Ainsi, dans notre projet la durée d'une tournée comprend le temps de conduite (qui dépend du trafic) et le temps d'attente devant les clients livrés avant l'ouverture de leur créneau. Minimiser leur somme revient à minimiser le **temps total de mobilisation des camions et des chauffeurs**. Pour CesiCDP, c'est à la fois un coût (heures de travail, location des véhicules) et un indicateur environnemental : moins de temps sur la route, notamment aux heures de pointe, c'est moins de carburant consommé et moins d'émissions, ce qui rejoint l'objectif de l'ADEME.



Pour finir, la fonction objectif nous permet de déterminer ce que nous devons minimiser. Cependant, rien ne garantit que la solution obtenue respecte les règles du problème ; c’est pourquoi nous avons défini un ensemble de contraintes.

## Contraintes


### 1. Chaque client est visité/servi une fois, par un seul camion

$$\sum_{k \in K} \ \sum_{(i, j) \in V} x_{ijk} = 1$$



### 2. Chaque camion qui part revient au dépôt

$$\sum_{i \in V} x_{i0k} = 1 \qquad \forall k \in K $$

$$\sum_{j \in V} x_{0jk} = 1 \qquad \forall k \in K $$


### 3. Le nombre de camions utilisés ne dépasse pas $k$

$$\sum_{k \in K} \ \sum_{(0, j) \in V} x_{0jk} \le k $$


### 4. La capacité de chaque camion est respectée

$$\sum_{i \in N} q_i \sum_{(i, j) \in V} x_{ijk} \;\le\; Q \qquad \forall k \in K$$


### 5. Les clients sont livrés pendant leur créneau

$$a_i \le w_i \le b_i \qquad \forall i \in N$$

L'heure de livraison $w_i$ doit être comprise entre l'ouverture $a_i$ et la fermeture $b_i$ du créneau du client $i$.


### 6. Chronologie des livraisons, autorisation d'attente

$$
w_{ik} + s_i + t_{ij} - w_{jk} \leq M_{ij}(1 - x_{ijk}) 
\qquad \forall (i,j) \in E,\ \forall k \in K
$$


avec :

$$
s_i = \text{durée du service chez le client } i
$$

$$
M_{ij} = \text{constante permettant de relâcher la contrainte temporelle}
$$

Elle est choisie suffisamment grande pour que la contrainte soit
automatiquement satisfaite lorsque le camion n'emprunte pas
directement le trajet de i vers j :

$$
x_{ijk} = 0
$$

En revanche, lorsque le trajet est utilisé :

$$
x_{ijk} = 1
\quad \Longrightarrow \quad
M_{ij}(1-x_{ijk}) = 0
$$

La contrainte impose alors la cohérence des horaires entre les deux clients.

$$
\boxed{w_{jk} \ge w_{ik} +  s_{i} + t_{ij}}
$$

sinon

$$
\boxed{w_{ik} + s_i + t_{ij} - w_{jk} \leq M_{ij}}
$$


---


### 7. Chronologie des livraisons, en tenant compte du trafic



Départ du dépôt (tous les camions partent à l'heure $0$, donc pendant la tranche $1$) :

$$w_j \;\ge\; t_{0j}^{1} \sum_{c \in K} x_{0jc} \qquad \forall j \in N : (0, j) \in E \qquad \text{(E8a)}$$

Trajet d'un client $i$ vers un client $j$, pour chaque tranche horaire $p$ :

$$w_j \;\ge\; w_i + t_{ij}^{p} - M_{ij} \Big( 2 - \sum_{c \in K} x_{ijc} - y_{ip} \Big) \qquad \forall (i, j) \in E,\ i, j \in N,\ \forall p \in \{1, \dots, P\} \qquad \text{(E8b)}$$

**Explication de E8a.** Si un camion va directement du dépôt au client $j$, alors $\sum_c x_{0jc} = 1$ et la contrainte devient $w_j \ge t_{0j}^1$ : on ne peut pas livrer $j$ avant d'y être arrivé. Sinon, elle devient $w_j \ge 0$, ce qui ne contraint rien.

**Explication de E8b.** La parenthèse $2 - \sum_c x_{ijc} - y_{ip}$ vaut $0$ seulement dans un cas : un camion va directement de $i$ à $j$ **et** il quitte $i$ pendant la tranche $p$. Dans ce cas, la contrainte devient

$$w_j \ge w_i + t_{ij}^{p},$$

ce qui signifie : « on livre $j$ au plus tôt à l'heure de départ de $i$, plus la durée du trajet de $i$ à $j$ à cette heure-là ». C'est la formule $Arr(v_l) = D_{l-1} + t_{v_{l-1} v_l}(D_{l-1})$ du §4.2.2. Dans tous les autres cas, la parenthèse vaut au moins $1$, et le terme $-M_{ij}$ rend la contrainte toujours vraie : elle est **désactivée**.

**Pourquoi la valeur $M_{ij} = b_i + \max_p t_{ij}^p - a_j$ suffit.** Quand la contrainte est désactivée, il faut qu'elle soit vérifiée quelles que soient les heures. Or, d'après E7, $w_i \le b_i$ et $w_j \ge a_j$, donc $w_i + t_{ij}^p - w_j \le b_i + \max_p t_{ij}^p - a_j = M_{ij}$. On prend la plus petite valeur qui marche, car un $M$ trop grand affaiblit la relaxation linéaire et ralentit la résolution.

**Inégalité et non égalité.** On écrit $w_j \ge \dots$ et non $w_j = \max(a_j, \dots)$ : si le camion arrive avant $a_j$, E7 oblige $w_j \ge a_j$, et le camion attend (contrainte n° 12). Le modèle autorise aussi une attente plus longue que nécessaire ; elle n'est jamais avantageuse sous l'hypothèse FIFO (§4.7.6), donc une solution optimale n'en contient pas.

**Lien avec la réunion.** Contraintes n° 14 (respect de la chronologie), n° 17 (trafic : la durée $t_{ij}^p$ dépend de la tranche de départ), n° 2 (les poids du graphe sont des temps de trajet), n° 12 (attente) et n° 15 (durée de livraison prise nulle : aucun terme de service entre $w_i$ et le départ de $i$).

**Justification.** Ces contraintes ont deux rôles.

1. Elles **calculent les horaires** de chaque tournée, ce qui permet de vérifier les créneaux (E7) et de mesurer la durée des tournées (E9).
2. Elles **éliminent les sous-tours**, c'est-à-dire les circuits qui ne passent pas par le dépôt (contrainte n° 7). En effet, si les arcs choisis formaient un circuit $i_1 \to i_2 \to \dots \to i_m \to i_1$ entre clients, E8b donnerait $w_{i_1} < w_{i_2} < \dots < w_{i_m} < w_{i_1}$ (les durées étant strictement positives), ce qui est impossible. C'est le même principe que les contraintes de Miller, Tucker et Zemlin (MTZ), où l'heure de livraison joue le rôle du numéro d'ordre de passage.

## Modèle complet


$$
\begin{aligned}
\min \quad & \sum_{k \in K} \sum_{(i,j) \in V} t_{ij}  \ x_{ijk} \\[4pt]

\text{s.c.} \quad
& \sum_{k \in K} \ \sum_{(i, j) \in V} x_{ijk} = 1 && \forall i \in N \\

& \sum_{i \in V} x_{i0k} = 1 && \forall k \in K \\

& \sum_{j \in V} x_{0jk} = 1 && \forall k \in K \\

& \sum_{k \in K} \ \sum_{(0, j) \in V} x_{0jk} \le k \\

& \sum_{i \in N} q_i \sum_{(i, j) \in V} x_{ijk} \;\le\; Q && \forall k \in K \\

& a_i \le w_i \le b_i && \forall i \in N \\


& w_{jk} \ge w_{ik} +  s_{i} + t_{ij} && \forall (i,j) \in E,\ \forall k \in K \\


& w_{ik} + s_i + t_{ij} - w_{jk} \leq M_{ij} && \forall (i,j) \in E,\ \forall k \in K \\



& x_{ijk} \in \{0, 1\} \\
& y_{ik} \in \{0, 1\} \\
& w_{ik} \geq 0 \\
& T_{max} \geq 0 \\
\end{aligned}
$$

## 1.1. Contexte du projet

La réduction de la consommation d’énergie et des émissions de gaz à effet de serre est devenue un enjeu important pour les entreprises et les collectivités. Le secteur des transports est particulièrement concerné, notamment pour les livraisons de marchandises. En effet, les déplacements représentent une part importante de leur activité et peuvent entraîner une consommation de carburant élevée lorsque les itinéraires sont mal optimisés.

Dans ce contexte, l’ADEME (Agence de la transition écologique) cherche à encourager le développement de nouvelles solutions de mobilité plus économiques et moins polluantes. L’objectif est notamment d’améliorer l’organisation des déplacements et la gestion des ressources disponibles afin de limiter leur impact environnemental.

Notre structure, CesiCDP, travaille déjà sur des problématiques de mobilité multimodale intelligente. Pour répondre à cet enjeu, nous avons choisi de nous intéresser à l’optimisation des tournées de livraison. Lorsqu’un véhicule doit desservir plusieurs clients, il existe de nombreux itinéraires possibles. Le plus court en distance n’est pas nécessairement le plus rapide, notamment lorsque le trafic varie au cours de la journée.

L’enjeu est donc de trouver une tournée permettant de livrer les clients prévus et de revenir au point de départ, tout en minimisant la durée totale du trajet. Pour cela, nous devrons tenir compte des caractéristiques du réseau routier et des conditions de circulation selon les horaires. Une meilleure organisation des tournées pourrait ainsi limiter les trajets inutiles, réduire les coûts de transport et contribuer à diminuer les émissions liées aux déplacements.

# 2. Objectifs globaux du projet

L’objectif global du projet est de concevoir une méthode permettant d’optimiser des tournées de livraison sur un réseau routier. Nous cherchons à minimiser la durée totale des déplacements, tout en respectant les contraintes du problème et en minimisant les complexité temporelle et spatiale de notre algorithme.

Pour y parvenir, nous devrons prendre en considération plusieurs éléments : les clients à desservir, le réseau routier, les durées de trajet selon les horaires et le retour au point de départ. # Cahier des charges au complet du projet 
### Problème avancée de niveau A (CVRPTW)

1. **Routes autorisées :** les camions circulent sur un réseau et ne peuvent emprunter que les routes existantes
2. **Durée des trajets :** le problème est représenté par un graphe pondéré dont les poids correspondent aux temps de trajet.
3. **Retour au dépôt :** chaque camion doit partir du dépôt, effectuer ses livraisons et revenir au dépôt
4. **Livraison de tous les clients :** tous les clients doivent être desservis.
5. **Visite unique :** chaque client doit être visité une seule fois.
6. **Affectation des clients :** chaque client est pris en charge par un seul camion.
7. **Continuité des trajets :** chaque camion doit suivre un parcours cohérent entre ses différents clients.
8. **Capacité des camions :** la quantité transportée ne doit pas dépasser la capacité maximale des camions.
9. **Camions identiques :** tous les camions ont la même capacité.
10. **Nombre de camions :** le nombre de camions utilisés ne doit pas dépasser la limite prévue dans le problème.
11. **Créneaux horaires :** chaque client doit être livré pendant son créneau horaire autorisé.
12. **Livraison trop tôt :** si un camion arrive avant l'ouverture du créneau, il doit attendre.
13. **Livraison trop tard :** un client ne doit pas être livré après la fin de son créneau horaire.
14. **Respect de la chronologie :** l'heure d'arrivée chez un client dépend de l'heure de départ du client précédent et du temps de trajet entre les deux.
15. **Minimisation du temps :** l'objectif est de minimiser le temps total des tournées.

# 4. Modélisation formelle et complexité du CVRPTW


Comme énoncé dans les parties précédentes, nous commençons par poser le problème étudié et formaliser ses données. Puis, nous différencierons le problème d'optimisation et le problème de décision. Enfin, nous évaluerons leur complexité afin de guider le choix de nos méthodes de résolution


<a id="partie4-1"></a>
## 4.1 Présentation du problème étudié

Nous étudions le **CVRPTW** (*Capacitated Vehicle Routing Problem with Time Windows*), il s'agit du problème de tournées de véhicules avec capacité et fenêtres de temps. 
Nous avons une flotte de camions identiques qui part d'un dépôt unique pour livrer un ensemble de clients sur un réseau routier. Chaque client a une demande et un créneau horaire de livraison. Nous cherchons des tournées qui desservent tous les clients en respectant les contraintes du cahier des charges (nos contraintes), tout en minimisant le temps total des tournées.

Le CVRPTW généralise le problème du voyageur de commerce (TSP) : un camion qui part du dépôt, visite chaque client une seule fois et revient au dépôt décrit un cycle hamiltonien. Le CVRPTW généralise le problème en ajoutant plusieurs camions, une capacité et des fenêtres de temps.

<a id="partie4-2"></a>
## 4.2. Problème d'optimisation et problème de décision

Une même situation peut être posée de deux façons :

- soit on cherche la **meilleure solution** : c'est un **problème d'optimisation** ;
- soit on demande s'il existe une **solution assez bonne**, avec une réponse « oui » ou « non » : c'est un **problème de décision**.

> Pourquoi cette distinction ? Les classes P, NP et NP-complet ne contiennent que des problèmes de décision. Notre problème, tel que le pose CesiCDP, est un problème d'optimisation (on veut les meilleures tournées). Pour étudier sa complexité, il faut donc d'abord écrire le problème de décision qui lui est associé.

### 4.2.1. Problème de décision : CVRPTW-D

**Données :** une instance $I = (G, t, q, s, [a,b], m_{\max}, Q)$ telle que définie au §4.2.1, et un entier $T_{max}$ (le seuil de temps).

**Question :** existe-t-il une solution réalisable $S \in \mathcal{S}(I)$ (§4.2.3), c'est-à-dire au plus $m_{\max}$ tournées partant du dépôt et y revenant, qui livrent chaque client une seule fois, sans dépasser la capacité $Q$ ni les créneaux $[a_i, b_i]$, et dont le temps total vérifie $f(S) \le T_{max}$ ?

La réponse est « oui » ou « non ». Le seuil $T_{max}$ transforme la recherche du « plus petit temps possible » en une question fermée. Il ne faut pas le confondre avec $m_{\max}$, qui est le nombre maximal de camions.

### 4.2.2. Problème d'optimisation : CVRPTW-OPT

**Données :** une instance $I = (G, t, q, s, [a,b], m_{\max}, Q)$.

**Question :** quelles tournées réalisables $S^* \in \mathcal{S}(I)$ minimisent le temps total $f$ ? (s'il n'existe aucune solution réalisable, le problème n'a pas de solution.)

La valeur $f(S^*)$ est le **temps total optimal**. On peut aussi la voir comme la plus petite valeur de $T_{max}$ pour laquelle CVRPTW-D répond « oui ».

> **Exemple.** Deux camions, cinq clients, et une meilleure solution de 120 minutes au total.
> - CVRPTW-OPT : la réponse est un plan de tournées de 120 minutes.
> - CVRPTW-D avec $T_{max} = 150$ : « existe-t-il des tournées d'au plus 150 minutes ? » Oui.
> - CVRPTW-D avec $T_{max} = 100$ : « existe-t-il des tournées d'au plus 100 minutes ? » Non.

## 4.3. Complexité du problème de décision CVRPTW-D

L'objectif de cette partie est de **démontrer que CVRPTW-D est NP-complet**.

### 4.3.1. Feuille de route de la démonstration

Nous suivons la méthode du cours, qui consiste à établir **séparément** deux propriétés : (1) CVRPTW-D appartient à NP ; (2) CVRPTW-D est NP-difficile. Le schéma ci-dessous résume la démarche ; chaque notion est précisée au moment où nous en avons besoin.

![plan.png](plan.png)

« NP-complet » regroupe deux idées : 
- le problème n'est **pas trop difficile à vérifier** (étape 1)
- il est **au moins aussi difficile** que les problèmes réputés durs (étape 2)

### 4.3.2. Étape 1 : CVRPTW-D appartient à NP

> Dans cette étape, nous cherchons à montrer que, lorsqu'on nous donne des tournées déjà construites, nous sommes capables de vérifier rapidement si elles respectent toutes les contraintes du problème.
> Cela permet de montrer que le problème appartient à la classe NP. Pour cela, nous devons disposer de deux éléments :
> - Un **certificat** $\gamma$ : il s'agit d'une solution proposée, ici un ensemble de tournées.
> - Un **algorithme de vérification** $V$ : il permet de vérifier que la solution proposée est correcte, en un temps polynomial.
> Autrement dit, une instance du problème a pour réponse « oui » s'il existe au moins une solution que notre algorithme de vérification accepte.


**Taille de notre instance.**<br> 
Pour parler de « temps polynomial », il faut d'abord mesurer la taille $|\xi|$ d'une instance $\xi = (I, T_{max})$. Elle correspond au nombre de symboles nécessaires pour décrire le problème. <br>
Dans notre cas, il faut prendre en compte ceci :
- Le **graphe** $G$ est représenté par une matrice d'adjacence contenant $(n+1)^2$ cases, 
- Les durées de trajet $t_{ij}(\tau)$ correspond à une valeur pour chaque arc et chaque tranche horaire, soit au plus $(n+1)^2 P$ valeurs, où $P$ correspond aux nombres de tranches horaires,
- Les demandes des clients, leurs durées de service et leurs créneaux horaires.
- Les paramètres du problème : le nombre de véhicules $m_{\max}$, leur capacité maximale $Q$ et la durée maximale $T_{max}$.

Ainsi, nous avons $n \le |\xi|$ et $P \le |\xi|$ : un coût en $O(n \log P)$ ou en $O(n \log n)$ est donc polynomial en $|\xi|$.

**Certificat.**<br>
Le certificat $\gamma$ correspond à une solution proposée, qui contient la liste des tournées non vides $R_1, \dots, R_r$, chacune donnée est représentée par une suite de sommets (dépôt, clients dans l'ordre de visite, dépôt). Une solution réalisable livre chaque client une seule fois (cahier des charges, point 5) : elle a donc $r \le n$ tournées et au maximum $n + 2r \le 3n$ sommets. L'algorithme de vérification $V$ rejette tout certificat plus long. Chaque sommet s'écrit avec $\lceil \log_2(n+1) \rceil$ bits, donc $|\gamma| = O(n \log n)$, ce qui est polynomial en $|\xi|$.

**Algorithme de vérification $V(\xi, \gamma)$.** <br>
Soit $P$ le nombre de tranches horaires de la table de trafic.

| Étape | Vérification | Méthode | Coût |
|---|---|---|---|
| 1 | Chaque tournée commence et finit au dépôt, et chaque déplacement entre deux sommets correspond à un arc de $E$ | parcours des tournées, lecture de la matrice d'adjacence | $O(n)$ |
| 2 | Chaque client $N$ apparaît exactement une fois | tableau de marquage de taille $n$ | $O(n)$ |
| 3 | Le nombre de tournées ne dépasse pas $m_{\max}$ soit $r \le m_{\max}$ | comptage des tournées | $O(1)$ |
| 4 | Pour chaque tournée, $\sum_{i \in R} q_i \le Q$ -> la demande totale de chaque tournée ne dépasse pas la capacité $Q$| somme des demandes tournée par tournée | $O(n)$ |
| 5 | Chaque client est visité dans son créneau horaire c-à-d $h_i \le b_i$ pour chaque client | calcul des horaires $Arr$, $h$, $D$ par les formules du §4.2.2 ; $t_{ij}(\tau)$ se lit dans la table des $P$ tranches par recherche dichotomique | $O(n \log P)$ |
| 6 | La durée totale des tournées ne dépasse pas : $\sum_R T(R) \le T_{max}$ | addition des durées $T(R)$ (attentes et services compris) puis comparaison | $O(n)$ |


#### Premier point : $V$ donne la bonne réponse

Nous montrons ici que $V(\xi, \gamma)$ accepte si et seulement si $\gamma$ est une solution réalisable $S \in \mathcal{S}(I)$ de coût $f(S) \le T_{max}$. Ce point est indispensable : sans lui, $V$ pourrait accepter de mauvaises tournées ou en refuser de bonnes, et il ne servirait à rien.

*Démonstration.* Les formules du §4.2.2 définissent sans ambiguïté les horaires $Arr$, $h$, $D$ et la durée $T(R)$ de chaque tournée : les valeurs calculées par $V$ sont donc celles de la définition. Les étapes 1, 2, 3, 4 et 5 testent chacune exactement l'une des cinq conditions de réalisabilité du §4.2.3 (l'étape 5 teste la seule borne $h_i \le b_i$, car $a_i \le h_i$ est automatique puisque $h_i = \max(a_i, Arr(i))$). L'étape 6 teste $f(S) \le T_{max}$. $V$ accepte donc exactement quand toutes les conditions sont vraies.

#### Second point : $V$ est rapide (temps polynomial)

Il reste à vérifier que $V$ s'exécute en temps polynomial en $|\xi|$. Le but est de respecter la définition de NP, qui exige un vérificateur **polynomial** : vérifier une solution ne doit pas coûter aussi cher que la chercher.

*Démonstration.* Soit $U$ le plus grand entier de l'instance. Comme $D_l = \max(a_{v_l}, D_{l-1} + t) + s_{v_l} \le D_{l-1} + 2U$ (car $a_{v_l}$, $t$ et $s_{v_l}$ sont au plus $U$ et $D_{l-1} \ge 0$), on a $D_l \le 2(n+1)U$ pour tout $l$, et $\sum_R T(R) \le 2n(n+1)U$. Tous les nombres manipulés ont donc au plus $\log_2 U + 2\log_2(n+1) + 2$ bits : leur taille est polynomiale. Les étapes 1 à 6 effectuent $O(n \log P)$ opérations (additions, comparaisons, lectures de tables) sur de tels nombres, chacune en temps polynomial.


**Conclusion de l'étape 1.** Si la réponse à $\xi$ est « oui », il existe une solution réalisable de coût $\le T_{max}$ : c'est un certificat de taille polynomiale que $V$ accepte (premier point). Réciproquement, si $V$ accepte un certificat, la réponse est « oui » (premier point). $V$ est polynomial (second point). En conséquent **CVRPTW-D $\in$ NP**.


### 4.3.3. Étape 2 : CVRPTW-D est NP-difficile

Le but de cette étape est de montrer que CVRPTW-D est au moins aussi difficile qu'un problème déjà connu comme NP-complet. Pourquoi procéder ainsi ? Un problème est NP-difficile si **tout** problème de NP se ramène à lui par une réduction polynomiale, et il est impossible de traiter les problèmes de NP un par un. En partant d'un problème $A$ **déjà NP-complet**, nous savons que tous les problèmes de NP se ramènent déjà à $A$ : il nous suffit donc de ramener $A$ à CVRPTW-D, car les réductions s'enchaînent (voir la conclusion de l'étape 2).

Nous partons du problème du **cycle hamiltonien (HC)**, NP-complet (Karp, 1972 ; Garey et Johnson, 1979) :

> **HC.** <br>
> *Données :* un graphe non orienté $G = (V, E)$ à $n \ge 3$ sommets. <br>
> *Question :* $G$ possède-t-il un cycle passant exactement une fois par chaque sommet ?

Nous passons par le voyageur de commerce en version décision, qui est le cas particulier du CVRPTW avec un seul camion :

> **TSP-D.** <br>
> *Données :* $m \ge 3$ villes, un graphe complet, des distances entières symétriques $d_{ij} = d_{ji} \ge 0$ et un entier $\theta$. <br>
> *Question :* existe-t-il un cycle passant exactement une fois par chaque ville, de longueur au plus $\theta$ ?

Nous montrons $\text{HC} \le_p \text{TSP-D}$ (réduction 1), puis $\text{TSP-D} \le_p \text{CVRPTW-D}$ (réduction 2). Attention au sens : on part du problème connu pour arriver au problème étudié, jamais l'inverse. Chaque réduction établit trois points : la transformation est polynomiale, les réponses « oui » sont équivalentes dans les deux sens, et le problème de départ est connu comme NP-complet.

#### Réduction 1 : du cycle hamiltonien au voyageur de commerce ($\text{HC} \le_p \text{TSP-D}$)

> Nous faisons ici une première traduction : toute question « $G$ a-t-il un cycle hamiltonien ? » peut se transformer, en temps polynomial, en une question TSP-D de même réponse. Comme HC est NP-complet, cela va permettre de conclure que TSP-D, qui le « contient », est au moins aussi difficile.

*Construction.* À partir de $G = (V, E)$, on construit le graphe complet sur $V$ ($m = n$ villes) avec $d_{ij} = 1$ si $\{i,j\} \in E$ et $d_{ij} = 2$ sinon, et on pose $\theta = n$. Ces distances sont des entiers symétriques positifs : c'est bien une instance de TSP-D.


**Schéma : la construction sur un exemple à 5 sommets.**

![hc_tsp.png](hc_tsp.png)

*Polynomial.* Il suffit d'examiner chacune des $n(n-1)/2$ paires de sommets : $O(n^2)$.

*Équivalence.*
- ($\Rightarrow$) Si $G$ a un cycle hamiltonien, il utilise $n$ arêtes de $E$, de poids $1$ dans le graphe construit. Ce cycle passe par toutes les villes et sa longueur est $n \le \theta$ : la réponse est « oui ».
- ($\Leftarrow$) Si le graphe construit a un cycle passant par toutes les villes, de longueur $\le n$, ce cycle utilise exactement $n$ arêtes, chacune de poids $\ge 1$. Pour que la somme ne dépasse pas $n$, toutes pèsent exactement $1$ : elles appartiennent toutes à $E$. C'est un cycle hamiltonien de $G$. 



#### Réduction 2 : du voyageur de commerce au CVRPTW ($\text{TSP-D} \le_p \text{CVRPTW-D}$)

> La seconde traduction relie le voyageur de commerce à notre problème : toute question TSP-D peut se transformer, en temps polynomial, en une question CVRPTW-D de même réponse. TSP-D étant NP-complet (réduction 1), nous pourrons ainsi en déduire que CVRPTW-D, qui le « contient », est au moins aussi difficile.

*Construction.* À partir d'une instance de TSP-D ($m$ villes, distances $d_{ij}$, seuil $\theta$), on construit l'instance $(I, T_{max})$ de CVRPTW-D suivante, qui neutralise tout ce que le CVRPTW ajoute au TSP :

| Élément | Valeur choisie | Effet |
|---|---|---|
| Sommets | les $m$ villes ; la ville 1 devient le dépôt $v_0$ ; les $n = m - 1$ autres sont les clients | mêmes villes |
| Arcs $E$ | toutes les paires, dans les deux sens | graphe complet comme dans TSP-D |
| Durées $t_{ij}(\tau)$ | $d_{ij}$ pour toute heure $\tau$ (une seule tranche, $P = 1$) | pas de trafic |
| Demandes et capacité | $q_i = 1$ pour tout client, $Q = n$ | la capacité ne limite jamais |
| Durées de service | $s_i = 0$ pour tout client | pas de temps de service |
| Créneaux $[a_i, b_i]$ | $[0, \theta]$ pour tout client | aucune attente possible |
| Nombre de camions | $m_{\max} = 1$ | une seule tournée |
| Seuil | $T_{max} = \theta$ | même seuil |


**Schéma : un cycle du TSP devient une tournée d'un seul camion.**

![tsp_cvrptw.png](tsp_cvrptw.png)

*Validité.* C'est bien une instance de CVRPTW-D : les valeurs sont des entiers, $s_i = 0$, $0 \le q_i = 1 \le Q = n$, $a_i = 0 \le b_i = \theta$, $m_{\max} = 1 \in \mathbb{N}^*$, et l'hypothèse FIFO est vérifiée car les durées sont constantes (partir plus tard ne fait pas arriver plus tôt).

*Polynomial.* La transformation recopie les $m$ villes, les $m(m-1)/2$ distances (utilisées dans les deux sens) et $\theta$, puis fixe quelques constantes : $O(m^2)$. Elle n'énumère aucune tournée.

*Équivalence.*
- ($\Rightarrow$) Soit un cycle passant par les $m$ villes, de longueur $L \le \theta$. On le fait partir de la ville 1 : on obtient $R = (0, u_1, \dots, u_n, 0)$. Elle est réalisable : chaque client apparaît une fois ; tous les arcs existent (graphe complet) ; $r = 1 \le m_{\max}$ ; la charge vaut $n \le Q$ ; comme $a_i = 0$ il n'y a aucune attente et, comme $s_i = 0$, le camion repart dès qu'il a livré, donc $Arr(u_l)$ est la somme des $l$ premières distances, d'où $Arr(u_l) \le L \le \theta = b_{u_l}$. Son temps total est $T(R) = L \le \theta = T_{max}$ : la réponse est « oui ».
- ($\Leftarrow$) Soit une solution réalisable $S$ avec $f(S) \le \theta$. Il y a au moins un client à servir et $r \le m_{\max} = 1$ : $S$ est une unique tournée $R = (0, u_1, \dots, u_n, 0)$. Elle livre chaque client une seule fois et revient au dépôt : c'est un cycle passant exactement une fois par chaque ville. Comme $a_i = 0$ il n'y a aucune attente, et comme les durées sont constantes et les durées de service nulles, $T(R)$ est exactement la somme des distances le long de ce cycle. Sa longueur est donc $T(R) = f(S) \le \theta$ : la réponse est « oui ». 


**Conclusion de l'étape 2.** Les deux réductions s'enchaînent : appliquer la transformation de la réduction 1 puis celle de la réduction 2 donne une transformation de HC vers CVRPTW-D qui conserve la réponse, et qui reste polynomiale car la composée de deux fonctions calculables en temps polynomial l'est aussi (la relation $\le_p$ est **transitive**) :
$$\text{HC} \le_p \text{TSP-D} \le_p \text{CVRPTW-D}.$$
HC est NP-complet : tout problème de NP se ramène à HC, donc, en prolongeant la chaîne, à CVRPTW-D. **CVRPTW-D est donc NP-difficile**. Intuitivement, le TSP est un cas particulier du CVRPTW (un seul camion, une capacité qui ne limite rien, des créneaux assez larges, pas de trafic) : les contraintes supplémentaires (capacité, créneaux, plusieurs camions, trafic) ne peuvent pas rendre le problème plus facile.

## 4.4. Conclusion

CVRPTW-D appartient à NP (étape 1) et il est NP-difficile (étape 2). **CVRPTW-D est donc NP-complet.**

## 4.5. Complexité du problème d'optimisation CVRPTW-OPT

CVRPTW-OPT n'est pas un problème de décision (§4.3) : on le situe donc par rapport à CVRPTW-D, avec les Propositions 1 et 2 du §4.3.3.

**Définition (extension de la NP-difficulté aux problèmes d'optimisation).** Un problème $B$ est **NP-difficile** si tout problème $A \in$ NP peut être résolu par un algorithme de temps polynomial qui a le droit d'appeler un nombre polynomial de fois un algorithme résolvant $B$. On note $A \le_T B$. Pour un problème de décision, c'est exactement la définition du §4.4.1 (une réduction $\le_p$ est le cas particulier d'un seul appel dont on renvoie la réponse).

> **Théorème 1.** CVRPTW-OPT est NP-difficile.

*Démonstration.* Soit $A \in$ NP. D'après le §4.4, $A \le_p$ CVRPTW-D : toute instance $\xi$ de $A$ se transforme en temps polynomial en une instance $(I, T_{max})$ de CVRPTW-D de même réponse. D'après la Proposition 1, cette instance se résout avec un seul appel à un algorithme pour CVRPTW-OPT. Donc $A \le_T$ CVRPTW-OPT. 

> **Corollaire.** S'il existait un algorithme polynomial pour CVRPTW-OPT, tout problème de NP serait résolu en temps polynomial (la composition de polynômes est un polynôme) : on aurait $P = NP$. En l'absence de preuve que $P = NP$, on ne peut pas supposer qu'un tel algorithme existe.

**CVRPTW-OPT n'appartient pas à NP**, puisque ce n'est pas un problème de décision. On sait vérifier rapidement qu'une solution est réalisable et calculer son coût, mais pas qu'elle est **optimale** : il faudrait prouver que CVRPTW-D répond « non » pour $T_{max} = f(S_o) - 1$. Un certificat court pour une réponse « non » placerait CVRPTW-D dans co-NP, donc entraînerait $NP = \text{co-NP}$, une conjecture ouverte que l'on pense fausse.

**Pas plus difficile que CVRPTW-D.** D'après la Proposition 2, CVRPTW-OPT se résout avec un nombre polynomial d'appels à CVRPTW-D : il est exactement aussi difficile que lui.

### Bilan de la démarche

$$\text{HC (NP-complet)} \;\le_p\; \text{TSP-D (NP-complet)} \;\le_p\; \text{CVRPTW-D (NP-complet)} \;\le_T\; \text{CVRPTW-OPT (NP-difficile)}$$

| Problème | Question posée | Catégorie |
|---|---|---|
| Plus court chemin | Existe-t-il un chemin de durée au plus $\theta$ ? | P |
| Cycle hamiltonien | Existe-t-il un cycle passant exactement une fois par chaque sommet ? | NP-complet |
| Voyageur de commerce (décision) | Existe-t-il un circuit passant une fois par chaque ville, de longueur au plus $\theta$ ? | NP-complet |
| Voyageur de commerce (optimisation) | Quel est le circuit le plus court ? | NP-difficile |
| **CVRPTW-D** (décision) | Existe-t-il des tournées d'au plus $m_{\max}$ camions de capacité $Q$, qui livrent chaque client une fois dans son créneau, avec un temps total $\le T_{max}$ ? | **NP-complet** (§4.4) |
| **CVRPTW-OPT** (optimisation) | Quelles tournées réalisables minimisent le temps total ? | **NP-difficile** (§4.5) |

*Source utilisée pour la méthode : cours de NP-complétude, https://perso.lisn.upsaclay.fr/jcohen/documents/enseignement/Cours-NP-completude.pdf*

Après avoir étudié la complexité théorique de notre problème et démontré son appartenance à la classe NP, nous allons maintenant nous intéresser à sa formulation mathématique sous la forme d’un Programme Linéaire en Nombres Entiers (PLNE), afin de représenter formellement les décisions à prendre, l’objectif à optimiser ainsi que les différentes contraintes du CVRPTW.

<a id="partie5"></a>
# 5. Modélisation mathématique du CVRPTW sous forme de PLNE


<a id="partie5-1"></a>
## 5.1. Principe de la modélisation

L'objectif est de transformer le problème de livraison en un modèle mathématique que nous pourrons ensuite résoudre avec un solveur d'optimisation.

Pour cela, nous devons représenter les décisions à prendre : quels camions utiliser, quels clients leur affecter, quelles routes emprunter et à quelles heures effectuer les livraisons. Le modèle doit également garantir que toutes les solutions proposées respectent les contraintes du cahier des charges.

Nous distinguons donc deux éléments :

- **La fonction objectif**, qui indique ce que nous cherchons à optimiser.
- **Les contraintes**, qui définissent les solutions autorisées.

Le problème est modélisé sous la forme d'un programme linéaire en nombres entiers (PLNE). Les décisions de déplacement et d'affectation sont représentées par des variables binaires, tandis que les horaires et les temps d'attente sont représentés par des variables numériques.


<a id="partie5-2"></a>
## 5.2. Définition des ensembles, paramètres et variables

### 5.2.1. Ensembles

Nous définissons les ensembles nécessaires à la représentation du réseau routier :

- $N$ : ensemble des clients à livrer.
- $K$ : ensemble des camions disponibles.
- $V=N\cup\{0\}$ : ensemble des sommets du graphe, comprenant les clients et le dépôt, noté $0$.
- $E$ : ensemble des arcs autorisés du graphe. Un arc $(i,j)$ appartient à $E$ si un déplacement direct du sommet $i$ vers le sommet $j$ est possible.

Pour illustrer les équations, nous prendrons un exemple composé de trois clients et de deux camions :

$$
N=\{1,2,3\}
$$

$$
K=\{1,2\}
$$

$$
V=\{0,1,2,3\}
$$

Le sommet $0$ représente le dépôt de départ et de retour des camions.

Une tournée possible pourrait être :

$$
0\rightarrow1\rightarrow3\rightarrow0
$$

Les arcs sont orientés : si le déplacement de $i$ vers $j$ est autorisé, cela ne signifie pas nécessairement que le déplacement inverse est autorisé. Il faut donc respecter les arcs effectivement présents dans $E$.

### 5.2.2. Paramètres

Les paramètres sont les données connues avant la résolution.

| Paramètre | Signification |
|---|---|
| $t_{ij}$ | Temps de trajet entre $i$ et $j$, en tenant compte des prévisions de trafic |
| $s_i$ | Durée de service chez le client $i$ (chargement ou déchargement), avec $s_0=0$ pour le dépôt |
| $q_i$ | Quantité de marchandises destinée au client $i$ |
| $Q$ | Capacité maximale commune à tous les camions |
| $a_i$ | Heure d'ouverture de la fenêtre de livraison du client $i$ |
| $b_i$ | Heure de fermeture de la fenêtre de livraison du client $i$ |
| $t_0$ | Heure de départ des camions du dépôt (égale à $0$ dans la partie 4) |
| $m_{\max}$ | Nombre maximal de camions pouvant être utilisés |
| $b_{\max}$ | Plus grande heure de fermeture parmi toutes les fenêtres de livraison : $b_{\max}=\max_{i\in N}b_i$ |
| $s_{\max}$ | Plus longue durée de service : $s_{\max}=\max_{i\in N}s_i$ |
| $c_{\max}$ | Plus long temps de trajet du graphe : $c_{\max}=\max_{(i,j)\in E}t_{ij}$ |
| $M$ | Grande constante positive utilisée pour activer ou désactiver certaines contraintes |

Le cahier des charges ne fixe pas d'heure de fin de journée : nous considérons que les équipes peuvent travailler sans limite de temps. Aucune contrainte ne porte donc sur l'heure de retour des camions au dépôt.

Les temps de trajet $t_{ij}$ et les durées de service $s_i$ doivent être cohérents avec les unités utilisées pour les horaires. Par exemple, si les heures sont exprimées en minutes depuis minuit, les temps de trajet et les durées de service doivent également être exprimés en minutes.

Pour un client $i$, la fenêtre de livraison est :

$$
[a_i,b_i]
$$

Par exemple, si le client accepte les livraisons entre 9 h et 12 h :

$$
a_i=540,\qquad b_i=720
$$

Ici, les horaires sont exprimés en minutes depuis minuit. Le camion doit commencer la livraison entre ces deux bornes. S'il arrive avant 9 h, il attend ; s'il ne peut arriver qu'après 12 h, la tournée est irréalisable pour ce client.

La fenêtre $[a_i,b_i]$ s'applique au **début** de la livraison. Une fois la livraison commencée, elle dure $s_i$ : le camion ne repart du client $i$ qu'à la fin de ce service, même si celle-ci dépasse l'heure $b_i$.

Les prévisions de trafic sont intégrées dans les paramètres $t_{ij}$. Le modèle utilise donc les temps de trajet fournis au moment de la résolution. Si les temps de trajet dépendent eux-mêmes de l'heure de passage, une modélisation temporelle plus avancée sera nécessaire. Dans la partie 4, la durée d'un trajet dépend de l'heure de départ $\tau$ ; ici, $t_{ij}$ désigne la valeur $t_{ij}(\tau)$ retenue pour la résolution.

### 5.2.3. Variables de décision

#### Variable de déplacement

Nous définissons une variable binaire pour indiquer si un camion emprunte un arc donné :

$$
x_{ijk}=
\begin{cases}
1 & \text{si le camion }k\text{ emprunte l'arc }(i,j)\\
0 & \text{sinon}
\end{cases}
$$

Par exemple :

$$
x_{0,1,1}=1
$$

signifie que le camion $1$ part directement du dépôt vers le client $1$.

Cette variable permet de construire les tournées. Comme elle est binaire, le solveur ne peut pas choisir une fraction de déplacement : un arc est sélectionné ou non.

#### Variable d'utilisation d'un camion

$$
y_k=
\begin{cases}
1 & \text{si le camion }k\text{ est utilisé}\\
0 & \text{sinon}
\end{cases}
$$

Cette variable permet de distinguer les camions disponibles des camions réellement utilisés. Elle sera notamment reliée aux contraintes de départ et de retour au dépôt.

#### Variable d'affectation d'un client

$$
z_{ik}=
\begin{cases}
1 & \text{si le client }i\text{ est affecté au camion }k\\
0 & \text{sinon}
\end{cases}
$$

Par exemple, $z_{12}=1$ signifie que le client $1$ est affecté au camion $2$.

Cette variable est utile pour exprimer les contraintes de capacité et de fenêtres de livraison sans devoir répéter toute la structure de la tournée.

#### Variables temporelles

Nous définissons $h_{ik}$ comme l'heure à laquelle le camion $k$ commence la livraison chez le client $i$, après une éventuelle attente :

$$
h_{ik}=\text{heure de début de livraison retenue pour le client }i
\text{ par le camion }k
$$

Si le camion arrive trop tôt, $h_{ik}$ correspond à l'heure d'ouverture de la fenêtre plutôt qu'à son heure d'arrivée physique. Ces variables sont comprises entre $0$ et $b_{\max}$, la plus grande heure de fermeture des fenêtres. Cette restriction ne retire aucune solution utile : un client livré a nécessairement une heure de livraison inférieure ou égale à $b_i\leq b_{\max}$, et lorsque le camion $k$ ne visite pas le client $i$, la valeur de $h_{ik}$ n'a pas de signification.

Le cahier des charges prévoit une durée de service $s_i$ chez chaque client (chargement ou déchargement de la marchandise). La livraison commence à l'heure $h_{ik}$ et le camion ne peut repartir qu'à l'heure $h_{ik}+s_i$. Le dépôt n'a pas de durée de service : $s_0=0$.

Nous définissons également :

$$
w_i=\text{durée d'attente avant la livraison du client }i
$$

Cette variable, positive ou nulle, mesure le temps pendant lequel le camion qui dessert le client $i$ patiente devant ce client parce qu'il est arrivé avant l'ouverture de sa fenêtre. Comme chaque client est desservi par un seul camion, un seul indice $i$ suffit.


<a id="partie5-3"></a>
## 5.3. Fonction objectif : minimiser le temps global de livraison

### 5.3.1. Choix du critère

Le cahier des charges demande de minimiser le **temps global de livraison**. Nous l'interprétons comme la **somme des durées des tournées de tous les camions utilisés**, une tournée étant mesurée entre le départ du dépôt et le retour au dépôt.

Ce choix se justifie de la manière suivante :

- Minimiser uniquement la **distance** parcourue ignorerait les temps d'attente : deux tournées de même longueur pourraient avoir des durées très différentes si l'une oblige le camion à attendre devant des clients qui ne sont pas encore ouverts. Or une attente mobilise le camion et son conducteur exactement comme un trajet.
- Minimiser la **durée de la tournée la plus longue** (le *makespan*) ne mesure que le camion le plus lent et laisse les autres tournées libres de s'allonger inutilement.
- Minimiser la **somme des durées** reflète le temps total mobilisé par la flotte, et tient compte à la fois des trajets, des temps de service et des attentes.

### 5.3.2. Décomposition de la durée d'une tournée

Tous les camions partent du dépôt à la même heure $t_0$. Aucune attente n'a donc lieu au dépôt, et la durée de la tournée du camion $k$ se décompose en :

- le temps de trajet entre le dépôt et le premier client ;
- les temps de trajet entre les différents clients ;
- les temps d'attente lorsque le camion arrive avant l'ouverture de la fenêtre de livraison d'un client ;
- les durées de service chez les clients visités ;
- le temps de trajet entre le dernier client et le dépôt.

L'heure de retour du camion $k$ vaut donc :

$$
\text{heure de retour}_k=t_0+\sum_{(i,j)\in E}t_{ij}x_{ijk}+\sum_{i\in N}(s_i+w_i)z_{ik}
$$

et la durée de sa tournée, notée $T(R)$ au §4.2.2, est :

$$
\text{durée}_k=\sum_{(i,j)\in E}t_{ij}x_{ijk}+\sum_{i\in N}(s_i+w_i)z_{ik}
$$

**Exemple.** Exprimons les heures en minutes depuis minuit. Le camion part à $t_0=480$ (8 h) et suit la tournée $0\rightarrow1\rightarrow3\rightarrow0$ avec $c_{01}=30$, $c_{13}=20$, $c_{30}=25$ et des durées de service $s_1=10$ et $s_3=15$.

- Il arrive chez le client $1$ à $510$, alors que la fenêtre ouvre à $a_1=540$ : il attend $w_1=30$ minutes et commence à livrer à $h_{11}=540$. Le service dure $10$ minutes, le camion repart à $550$.
- Il arrive chez le client $3$ à $550+20=570$, dans sa fenêtre : $w_3=0$ et $h_{31}=570$. Le service dure $15$ minutes, le camion repart à $585$.
- Il retourne au dépôt à $585+25=610$.

La durée de la tournée est $610-480=130$ minutes, soit $(30+20+25)=75$ minutes de trajet, $(10+15)=25$ minutes de service et $30$ minutes d'attente.

### 5.3.3. Somme sur tous les camions

Le temps global de livraison, c'est-à-dire le coût $f(S)$ défini au §4.2.3, est la somme des durées des tournées de tous les camions :

$$
f(S)=\sum_{k\in K}\text{durée}_k
=\sum_{k\in K}\sum_{(i,j)\in E}t_{ij}x_{ijk}+\sum_{k\in K}\sum_{i\in N}(s_i+w_i) z_{ik}
$$

Comme chaque client est affecté à un seul camion, $\sum_{k\in K}z_{ik}=1$, donc :

$$
\sum_{k\in K}\sum_{i\in N}(s_i+w_i) z_{ik}=\sum_{i\in N}s_i+\sum_{i\in N}w_i
$$

Les produits $s_i z_{ik}$ et $w_i z_{ik}$ disparaissent : l'objectif reste **linéaire**. De plus, tous les clients étant servis exactement une fois, la somme $\sum_{i\in N}s_i$ est une **constante** : elle ne dépend pas des tournées choisies.

### 5.3.4. Fonction objectif finale

$$
\boxed{
\min F=\sum_{k\in K}\sum_{(i,j)\in E}t_{ij}x_{ijk}+\sum_{i\in N}s_i+\sum_{i\in N}w_i
}
$$

Le premier terme est le temps de trajet total des camions, le deuxième est le temps de service total et le troisième est le temps d'attente total. Le temps de service total étant constant, il ne modifie pas le choix des meilleures tournées, mais il permet à $F$ d'être exactement égal au temps global de livraison.

### 5.3.5. Justification de cette formulation

Cette écriture présente plusieurs avantages :

- **Elle est linéaire.** Elle ne contient que des sommes de variables multipliées par des constantes, ce qui est exactement ce qu'attend un solveur de PLNE.
- **Elle ne nécessite aucune borne sur la durée d'une tournée.** Les équipes travaillant sans heure de fin, il n'existe pas de limite naturelle à cette durée. Ici, un camion inutilisé n'emprunte aucun arc, n'attend devant aucun client et n'en sert aucun : sa contribution à $F$ est automatiquement nulle, sans contrainte supplémentaire.
- **Elle compte bien le retour au dépôt.** Les arcs $(i,0)$ sont dans la somme des temps de trajet, donc le trajet de retour est pénalisé comme les autres.
- **Elle est cohérente avec la définition de la durée.** Puisque tous les camions partent à $t_0$, la durée d'une tournée est exactement égale à la somme de ses temps de trajet, de ses temps de service et de ses attentes.

Pour que $w_i$ représente effectivement l'attente réelle, il faut le relier aux horaires. C'est le rôle des contraintes de la section 5.9.4. Comme $F$ est minimisée, le solveur n'a aucun intérêt à choisir une valeur de $w_i$ supérieure au minimum imposé par ces contraintes : à l'optimum, $w_i$ est égal au temps d'attente réel.


<a id="partie5-4"></a>
## 5.4. Contraintes de visite et d'affectation

### 5.4.1. Chaque client doit être visité exactement une fois

Le cahier des charges impose que chaque client soit livré une seule fois. Il faut donc empêcher deux situations :

- Un client n'est visité par aucun camion.
- Un client est visité plusieurs fois, éventuellement par plusieurs camions.

Pour construire la contrainte, nous comptons les arcs entrants vers chaque client, tous camions confondus.

Considérons le client $1$ et deux camions. Si tous les arcs entrants considérés existent dans le graphe, la contrainte s'écrit :

$$
\begin{aligned}
&x_{0,1,1}+x_{2,1,1}+x_{3,1,1}\\
&\quad+x_{0,1,2}+x_{2,1,2}+x_{3,1,2}=1
\end{aligned}
$$

Chaque terme représente un déplacement vers le client $1$. Par exemple, $x_{2,1,1}=1$ signifie que le camion $1$ arrive chez le client $1$ depuis le client $2$.

La somme doit être égale à $1$ :

- Si elle vaut $0$, aucun camion n'arrive chez le client : le client n'est pas livré.
- Si elle vaut $2$ ou plus, plusieurs arcs entrants sont sélectionnés : le client risque d'être visité plusieurs fois.
- Si elle vaut $1$, un seul arc entrant est sélectionné.

Les arcs absents du graphe ne doivent pas être inclus dans cette somme. La formulation générale est donc :

$$
\boxed{
\forall i\in N,\qquad
\sum_{k\in K}
\sum_{\substack{j\in V\\(j,i)\in E}}
x_{jik}=1
}
$$

Cette contrainte garantit une visite entrante unique pour chaque client. Elle doit être combinée avec les contraintes de continuité et de retour au dépôt pour que cette visite fasse partie d'une tournée valide.

### 5.4.2. Relier la visite à l'affectation d'un camion

Nous devons ensuite déterminer quel camion est responsable de chaque client. Pour cela, nous utilisons la variable $z_{ik}$.

Pour le client $1$ et deux camions, une seule affectation est possible :

$$
z_{11}+z_{12}=1
$$

Si $z_{11}=1$, le client $1$ est affecté au camion $1$ et $z_{12}=0$. À l'inverse, si $z_{12}=1$, il est affecté au camion $2$.

La généralisation est :

$$
\boxed{
\forall i\in N,\qquad
\sum_{k\in K}z_{ik}=1
}
$$

Mais cette équation ne suffit pas seule : elle pourrait affecter un client à un camion sans que celui-ci se rende réellement chez lui. Nous devons donc relier l'affectation aux arcs entrants.

Pour le client $1$ et le camion $1$, par exemple :

$$
z_{11}=x_{0,1,1}+x_{2,1,1}+x_{3,1,1}
$$

La variable $z_{11}$ vaut ainsi $1$ si le camion $1$ arrive chez le client $1$ par un arc autorisé.

La formulation générale est :

$$
\boxed{
\forall i\in N,\ \forall k\in K,\qquad
z_{ik}=
\sum_{\substack{j\in V\\(j,i)\in E}}x_{jik}
}
$$

Cette égalité assure la cohérence entre les décisions d'affectation et les décisions de déplacement. Combinée à la visite unique, elle empêche qu'un client soit attribué à plusieurs camions.


<a id="partie5-5"></a>
## 5.5. Contraintes de continuité des tournées

### 5.5.1. Un camion doit repartir de chaque client où il arrive

Pour chaque camion, si un client appartient à sa tournée, le camion doit y arriver par un arc entrant et en repartir par un arc sortant.

Considérons le client $1$ et le camion $1$. Si les arcs concernés sont autorisés, nous comptons les arrivées :

$$
x_{0,1,1}+x_{2,1,1}+x_{3,1,1}
$$

Puis nous comptons les départs :

$$
x_{1,0,1}+x_{1,2,1}+x_{1,3,1}
$$

La conservation du flux impose l'égalité entre les deux sommes :

$$
\begin{aligned}
x_{0,1,1}+x_{2,1,1}+x_{3,1,1}
={}&x_{1,0,1}\\
&+x_{1,2,1}+x_{1,3,1}
\end{aligned}
$$

Pourquoi une égalité ? Si le camion arrive chez le client $1$, il doit aussi repartir. Une arrivée sans départ interromprait la tournée ; un départ sans arrivée permettrait au camion d'apparaître artificiellement chez un client.

La formulation générale est :

$$
\boxed{
\forall i\in N,\ \forall k\in K,\qquad
\sum_{\substack{j\in V\\(j,i)\in E}}x_{jik}
=
\sum_{\substack{j\in V\\(i,j)\in E}}x_{ijk}
}
$$

Cette contrainte impose la continuité locale de chaque tournée. Toutefois, elle ne garantit pas, à elle seule, que tous les clients soient reliés au dépôt : un cycle isolé de clients pourrait encore satisfaire cette égalité. Il faut donc aussi prendre en compte les contraintes temporelles.

### 5.5.2. Chaque camion utilisé doit partir du dépôt

Pour qu'un camion soit utilisé, il doit quitter le dépôt une seule fois. S'il n'est pas utilisé, il ne doit pas partir.

Avec deux camions, nous pouvons écrire :

$$
\begin{aligned}
x_{0,1,1}+x_{0,2,1}+x_{0,3,1}&=y_1\\
x_{0,1,2}+x_{0,2,2}+x_{0,3,2}&=y_2
\end{aligned}
$$

Ces expressions supposent ici que les arcs du dépôt vers les trois clients existent. Dans le modèle réel, seuls les arcs appartenant à $E$ sont inclus.

Si $y_1=1$, la somme des départs du camion $1$ vaut $1$. Si $y_1=0$, elle vaut $0$.

La formulation générale est :

$$
\boxed{
\forall k\in K,\qquad
\sum_{\substack{j\in N\\(0,j)\in E}}x_{0jk}=y_k
}
$$

Cette équation relie la variable d'utilisation du camion à sa sortie du dépôt.

### 5.5.3. Chaque camion utilisé doit revenir au dépôt

Le même raisonnement s'applique au retour. Un camion utilisé doit avoir exactement un arc entrant vers le dépôt, tandis qu'un camion inutilisé ne doit pas effectuer de retour.

Pour le camion $1$, en supposant que les arcs de retour indiqués existent :

$$
x_{1,0,1}+x_{2,0,1}+x_{3,0,1}=y_1
$$

Pour tous les camions :

$$
\boxed{
\forall k\in K,\qquad
\sum_{\substack{i\in N\\(i,0)\in E}}x_{i0k}=y_k
}
$$

Ces contraintes de départ et de retour, combinées à la conservation des flux, imposent aux camions utilisés de quitter le dépôt et d'y revenir. Elles ne suffisent toutefois pas seules à exclure un cycle déconnecté du dépôt : il faut également empêcher les sous-tournées.

### 5.5.4. Respect des routes autorisées

Les variables de déplacement sont définies uniquement pour les arcs présents dans $E$ :

$$
x_{ijk}\qquad
\forall(i,j)\in E,\ \forall k\in K
$$

Si une route n'existe pas dans le graphe, la variable correspondante n'est pas créée. Le solveur ne peut donc pas sélectionner un déplacement direct interdit.

Il ne faut pas confondre l'absence d'un arc avec un arc de coût très élevé. Dans le premier cas, le déplacement est impossible ; dans le second, il reste possible et pourrait être choisi par le solveur si aucune meilleure solution n'existe.

### 5.5.5. Éliminer les sous-tournées

Une sous-tournée est un cycle de clients qui ne passe pas par le dépôt. Par exemple :

$$
1\rightarrow2\rightarrow3\rightarrow1
$$

Les contraintes de conservation des flux pourraient autoriser ce cycle : chaque client possède une arrivée et un départ. Il faut donc montrer pourquoi les contraintes temporelles peuvent l'éliminer.

Si les trois arcs sont sélectionnés par le même camion, les contraintes temporelles imposent :

$$
h_{2k}\geq h_{1k}+s_1+c_{12}
$$

$$
h_{3k}\geq h_{2k}+s_2+c_{23}
$$

$$
h_{1k}\geq h_{3k}+s_3+c_{31}
$$

En additionnant les trois inégalités, les heures de passage s'annulent de chaque côté et on obtient :

$$
0\geq s_1+s_2+s_3+c_{12}+c_{23}+c_{31}
$$

Or, si les durées de service sont positives ou nulles et les temps de trajet strictement positifs :

$$
s_1+s_2+s_3+c_{12}+c_{23}+c_{31}>0
$$

Les deux conditions sont incompatibles. Ce cycle ne peut donc pas être sélectionné.

Ce raisonnement s'étend à toute sous-tournée composée de clients, à condition que les contraintes temporelles soient activées sur chaque arc sélectionné et que les temps de trajet soient strictement positifs. Les contraintes temporelles jouent donc également un rôle dans la cohérence globale des tournées.


<a id="partie5-6"></a>
## 5.6. Contraintes de capacité

Chaque camion ne peut transporter qu'une quantité maximale de marchandises, fixée par $Q$. Il faut donc additionner les demandes des clients affectés à chaque camion et vérifier que le total ne dépasse pas cette capacité.

### 5.6.1. Construction de la contrainte

Prenons le camion $1$ et les trois clients. La charge transportée est :

$$
\text{charge}_1=q_1z_{11}+q_2z_{21}+q_3z_{31}
$$

Chaque demande $q_i$ est multipliée par $z_{i1}$ : si le client est affecté au camion $1$, sa demande est comptée ; sinon, elle vaut zéro dans le calcul de la charge.

La capacité doit être respectée :

$$
q_1z_{11}+q_2z_{21}+q_3z_{31}\leq Q
$$

Par exemple, si les clients $1$ et $3$ sont affectés au camion $1$, la charge devient :

$$
\text{charge}_1=q_1+q_3
$$

La tournée n'est autorisée que si $q_1+q_3\leq Q$.

### 5.6.2. Généralisation

Pour chaque camion, nous sommons les demandes de tous les clients qui lui sont affectés :

$$
\boxed{
\forall k\in K,\qquad
\sum_{i\in N}q_i z_{ik}\leq Q
}
$$

Cette contrainte est nécessaire pour chaque camion, car la capacité doit être respectée individuellement. Il ne suffit pas de vérifier la charge totale de tous les camions réunis.

Le paramètre $Q$ est commun à tous les camions, conformément au cahier des charges. Nous ne définissons donc pas de capacité différente selon le camion.

Cette formulation suppose que $q_i$ représente une quantité additive de marchandises, exprimée dans une unité compatible avec la capacité $Q$.


<a id="partie5-7"></a>
## 5.7. Limiter le nombre de camions utilisés

Le nombre de camions disponibles peut être supérieur au nombre de camions réellement nécessaires. Le cahier des charges fixe une limite maximale, que nous notons $m_{\max}$.

Avec deux camions disponibles, si le nombre maximal autorisé est égal à deux, la contrainte s'écrit :

$$
y_1+y_2\leq2
$$

Pour un ensemble général de camions :

$$
\boxed{
\sum_{k\in K}y_k\leq m_{\max}
}
$$

La variable $y_k$ valant $1$ lorsqu'un camion est utilisé et $0$ sinon, la somme correspond exactement au nombre de camions utilisés.

L'inégalité est préférable à une égalité : si trois camions sont disponibles mais que deux suffisent, le modèle doit pouvoir en utiliser seulement deux. Il n'est pas nécessaire de faire partir tous les camions.

Si nous prenons $K=\{1,\dots,m_{\max}\}$, cette contrainte est automatiquement vérifiée. Nous la conservons malgré tout pour que le modèle reste correct si l'ensemble $K$ est défini plus grand que la limite $m_{\max}$.


<a id="partie5-8"></a>
## 5.8. Fenêtres de livraison et grand $M$

### 5.8.1. Première formulation de la fenêtre

Pour le client $i$, la livraison doit être effectuée entre l'heure d'ouverture $a_i$ et l'heure de fermeture $b_i$ :

$$
a_i\leq h_{ik}\leq b_i
$$

Cette condition est correcte lorsque le camion $k$ est responsable de la livraison chez le client $i$.

Cependant, le camion $k$ n'est pas nécessairement affecté à ce client. Imposer cette condition à tous les couples client-camion pourrait contraindre artificiellement les horaires d'un camion qui ne visite pas le client.

Il faut donc activer la fenêtre de livraison uniquement lorsque $z_{ik}=1$.

### 5.8.2. Activer la contrainte avec une variable binaire

Nous construisons deux inégalités :

$$
\boxed{
h_{ik}\geq a_i-M(1-z_{ik})
}
$$

$$
\boxed{
h_{ik}\leq b_i+M(1-z_{ik})
}
$$

Le rôle de $M$ est de rendre les contraintes suffisamment permissives lorsque le camion n'est pas affecté au client.

**Premier cas : $z_{ik}=1$.**

Les contraintes deviennent :

$$
h_{ik}\geq a_i-M(1-1)=a_i
$$

$$
h_{ik}\leq b_i+M(1-1)=b_i
$$

La fenêtre est donc active :

$$
a_i\leq h_{ik}\leq b_i
$$

**Deuxième cas : $z_{ik}=0$.**

Les contraintes deviennent :

$$
h_{ik}\geq a_i-M
$$

$$
h_{ik}\leq b_i+M
$$

Si $M$ est suffisamment grand, ces deux inégalités ne limitent plus artificiellement l'heure de passage du camion pour ce client.

Cette méthode permet de traduire une condition logique dans un modèle linéaire. Elle évite de créer une contrainte de fenêtre obligatoire pour chaque camion, même lorsqu'il ne visite pas le client.

### 5.8.3. Prise en compte de l'attente

La variable $h_{ik}$ représente l'heure de livraison retenue après une éventuelle attente. Si le camion arrive avant l'ouverture de la fenêtre, il ne peut pas livrer immédiatement.

En notant $Arr_{ik}$ l'heure d'arrivée physique du camion chez le client, on peut comprendre le comportement attendu par :

$$
h_{ik}\geq Arr_{ik}
$$

et :

$$
a_i\leq h_{ik}\leq b_i
$$

Ainsi, l'heure de livraison retenue ne peut pas précéder l'arrivée physique ni l'ouverture de la fenêtre. La durée d'attente est la différence $w_i=h_{ik}-Arr_{ik}$. Nous ne créons pas de variable $Arr_{ik}$ supplémentaire dans le modèle : les contraintes de propagation temporelle de la section 5.9 permettent de représenter directement ce comportement avec $h_{ik}$ et $w_i$.


<a id="partie5-9"></a>
## 5.9. Propagation temporelle entre les sommets

Les horaires doivent être cohérents avec l'ordre des déplacements. Si un camion quitte un sommet pour rejoindre le suivant, il ne peut pas livrer le client suivant avant d'avoir terminé le service chez le client précédent, puis parcouru le temps de trajet correspondant.

Les contraintes temporelles ont trois rôles :

1. Respecter l'ordre chronologique des déplacements, en tenant compte de la durée de service chez chaque client.
2. Empêcher les sous-tournées lorsque les temps de trajet sont strictement positifs.
3. Définir les temps d'attente $w_i$ qui apparaissent dans la fonction objectif.

**Place de la durée de service.** La durée de service n'est pas une contrainte indépendante : elle intervient dans les contraintes temporelles par le terme $s_i$. Dans la propagation de la section 5.9.1, elle impose que le camion ne puisse quitter le client $i$ qu'à l'heure $h_{ik}+s_i$. Dans le calcul des attentes (section 5.9.4), elle est retirée de l'écart afin de ne pas compter le service comme de l'attente. Dans l'objectif, elle est ajoutée au temps global, et la plus longue durée $s_{\max}$ entre dans le choix du grand $M$ (section 5.10). Au départ du dépôt, aucun terme n'est nécessaire puisque $s_0=0$.

### 5.9.1. Déplacement entre deux clients et durée de service

Considérons le camion $1$ qui dessert le client $1$, puis le client $2$. Si la durée de service chez le client $1$ est $s_1$ et le temps de trajet est $c_{12}$, l'heure de livraison chez le client $2$ doit respecter :

$$
h_{21}\geq h_{11}+s_1+c_{12}
$$

Cette contrainte ne doit cependant être imposée que si le camion emprunte réellement l'arc $(1,2)$.

Nous utilisons donc la variable binaire $x_{1,2,1}$ et le grand $M$ :

$$
h_{21}\geq h_{11}+s_1+c_{12}-M(1-x_{1,2,1})
$$

Si $x_{1,2,1}=1$, le terme $M(1-x_{1,2,1})$ vaut zéro et l'inégalité devient :

$$
h_{21}\geq h_{11}+s_1+c_{12}
$$

Si $x_{1,2,1}=0$, la contrainte devient :

$$
h_{21}\geq h_{11}+s_1+c_{12}-M
$$

Avec un $M$ adapté aux bornes temporelles, cette dernière inégalité est automatiquement satisfaite et ne force pas un ordre de passage entre deux clients qui ne sont pas reliés par cet arc dans la tournée.

La formulation générale est :

$$
\boxed{
\forall k\in K,\quad
\forall (i,j)\in E,\ i,j\in N,\qquad
h_{jk}\geq h_{ik}+s_i+t_{ij}-M(1-x_{ijk})
}
$$

Cette contrainte signifie que chaque déplacement sélectionné impose une progression temporelle cohérente.

Si le camion arrive tôt chez le client suivant, l'heure $h_{jk}$ peut être supérieure à son heure d'arrivée physique : cette différence représente l'attente. Le camion repart du client $i$ à l'heure $h_{ik}+s_i$, une fois le service terminé.

### 5.9.2. Départ du dépôt vers le premier client

La propagation temporelle doit commencer au dépôt. Si le camion part à l'heure $t_0$ (le dépôt n'ayant pas de durée de service) et emprunte l'arc $(0,j)$, il ne peut pas arriver chez le client $j$ avant :

$$
t_0+t_{0j}
$$

Pour le camion $k$, la contrainte conditionnelle est :

$$
\boxed{
h_{jk}\geq t_0+t_{0j}-M(1-x_{0jk})
}
$$

Si $x_{0jk}=1$, le camion emprunte cet arc et la contrainte impose :

$$
h_{jk}\geq t_0+t_{0j}
$$

Si $x_{0jk}=0$, le grand $M$ relâche la contrainte. Le camion peut alors atteindre ce client par un autre itinéraire, ou ne pas le visiter du tout.

### 5.9.3. Retour au dépôt

Lorsque le camion $k$ emprunte l'arc $(i,0)$, il rentre au dépôt à l'heure $h_{ik}+s_i+t_{i0}$. Comme le cahier des charges ne fixe aucune heure de fin de journée, **aucune contrainte ne limite cette heure de retour** : le modèle ne contient donc aucune variable ni borne liée au retour.

Le trajet de retour est malgré tout pris en compte dans le critère à optimiser, puisque le temps $t_{i0}$ figure dans la somme $\sum t_{ij}x_{ijk}$ de la fonction objectif dès que $x_{i0k}=1$.

### 5.9.4. Définition des temps d'attente

Les variables $w_i$ doivent être liées aux horaires. L'attente devant un client est l'écart entre son heure de livraison retenue et son heure d'arrivée physique. Le camion arrivant chez $j$ après avoir servi $i$ puis roulé, son heure d'arrivée physique est $h_{ik}+s_i+t_{ij}$.

Pour le premier client d'une tournée, atteint depuis le dépôt :

$$
\boxed{
w_i\geq h_{ik}-t_0-c_{0i}-M(1-x_{0ik})
}
$$

Pour un client $j$ atteint depuis un client $i$ :

$$
\boxed{
w_j\geq h_{jk}-h_{ik}-s_i-t_{ij}-M(1-x_{ijk})
}
$$

Lorsque l'arc est emprunté ($x=1$), le terme $M(1-x)$ s'annule et la contrainte impose $w_j\geq h_{jk}-(h_{ik}+s_i+t_{ij})$. Or $h_{ik}+s_i+t_{ij}$ est l'heure d'arrivée physique chez $j$ : l'attente est donc bien l'écart entre l'heure de livraison retenue et l'heure d'arrivée.

Lorsque l'arc n'est pas emprunté ($x=0$), le grand $M$ relâche la contrainte, et la condition $w_j\geq 0$ suffit.

Les contraintes de la section 5.9.1 imposent $h_{jk}\geq h_{ik}+s_i+t_{ij}$ sur les arcs empruntés : le second membre de ces contraintes d'attente est donc positif ou nul. Comme $w_j$ est pénalisée dans $F$, le solveur prend $w_j$ égal à ce second membre.


<a id="partie5-10"></a>
## 5.10. Choix et justification du grand $M$

### 5.10.1. Pourquoi a-t-on besoin d'un grand $M$ ?

Un PLNE n'accepte que des contraintes linéaires, c'est-à-dire des inégalités qui sont **toujours** appliquées. Or plusieurs règles de notre problème sont **conditionnelles**. Par exemple :

> *si* le camion $k$ emprunte l'arc $(i,j)$, *alors* il ne peut pas commencer à livrer le client $j$ avant $h_{ik}+s_i+t_{ij}$ (heure de début de livraison chez $i$, durée de service, puis trajet).

Une phrase de la forme « si… alors… » ne s'écrit pas directement comme une inégalité linéaire. Si nous écrivions simplement $h_{jk}\geq h_{ik}+s_i+t_{ij}$ sans condition, cette inégalité serait imposée pour **toutes** les paires de clients, même ceux qui ne sont pas consécutifs dans une tournée. Elle exigerait par exemple à la fois $h_{jk}\geq h_{ik}+s_i+t_{ij}$ et $h_{ik}\geq h_{jk}+s_j+c_{ji}$, ce qui est impossible dès que les temps de trajet sont positifs : elle exclurait donc des tournées pourtant parfaitement valides.

Le grand $M$ sert à transformer cette condition en une inégalité linéaire qui joue le rôle d'un **interrupteur** commandé par une variable binaire. Nous ajoutons au second membre le terme $-M(1-x_{ijk})$ :

$$
h_{jk}\geq h_{ik}+s_i+t_{ij}-M(1-x_{ijk})
$$

- Si $x_{ijk}=1$ (l'arc est emprunté), le terme vaut $0$ et nous retrouvons la contrainte réelle : $h_{jk}\geq h_{ik}+s_i+t_{ij}$. L'interrupteur est **allumé**.
- Si $x_{ijk}=0$ (l'arc n'est pas emprunté), le second membre devient $h_{ik}+s_i+t_{ij}-M$, un nombre très négatif. L'inégalité est alors vraie quelles que soient les valeurs de $h_{ik}$ et $h_{jk}$ : elle ne bloque plus rien. L'interrupteur est **éteint**.

**Exemple.** Avec $h_{11}=540$, $s_1=10$ et $c_{12}=20$, et en prenant $M=790$ :

- si $x_{1,2,1}=1$, la contrainte impose $h_{21}\geq 570$ ;
- si $x_{1,2,1}=0$, elle impose seulement $h_{21}\geq 570-790=-220$, ce qui est toujours vrai puisque $h_{21}\geq 0$.

Le même mécanisme est utilisé pour les fenêtres de livraison (interrupteur $z_{ik}$) et pour les temps d'attente (interrupteur $x_{ijk}$).

### 5.10.2. Comment choisir sa valeur ?

Le grand $M$ doit respecter deux exigences opposées :

- **Il doit être assez grand** pour que la contrainte « éteinte » soit vraie dans tous les cas. S'il est trop petit, la contrainte reste en partie active alors qu'elle devrait être désactivée, et le modèle élimine à tort des solutions valides : le résultat est faux.
- **Il ne doit pas être démesuré.** Un $M$ trop grand rend le modèle numériquement instable et fait perdre beaucoup de temps au solveur.

Nous choisissons donc la plus petite valeur simple qui convient à toutes les contraintes :

$$
\boxed{
M=b_{\max}+s_{\max}+c_{\max}
}
$$

c'est-à-dire la plus grande heure de fermeture des fenêtres, plus la plus longue durée de service, plus le plus long temps de trajet.

**Pourquoi cette valeur convient.** Toutes les heures du modèle sont comprises entre $0$ et $b_{\max}$, un service dure au plus $s_{\max}$ et un trajet dure au plus $c_{\max}$. La quantité « heure + service + trajet » ne dépasse donc jamais $M$. Lorsque l'interrupteur est éteint, le second membre de la contrainte est négatif ou nul, et la contrainte est vérifiée sans restreindre aucune solution. Il en va de même pour les contraintes de fenêtres et d'attente, qui ne font intervenir que des heures au plus égales à $b_{\max}$, donc inférieures à $M$.

**Exemple.** Si la plus grande heure de fermeture est $b_{\max}=720$, la plus longue durée de service est $s_{\max}=10$ et le plus long trajet est $c_{\max}=60$, alors $M=790$.

Cette valeur pourra être resserrée lors de l'implémentation si la résolution est trop lente.


<a id="partie5-11"></a>
## 5.11. Formulation complète du PLNE et récapitulatif

### Ensembles et paramètres

- $N$ : ensemble des clients.
- $K$ : ensemble des camions disponibles.
- $V=N\cup\{0\}$ : ensemble des sommets, avec $0$ représentant le dépôt.
- $E$ : ensemble des arcs autorisés du graphe.
- $t_{ij}$ : temps de trajet entre les sommets $i$ et $j$, en tenant compte du trafic.
- $q_i$ : quantité à livrer au client $i$.
- $s_i$ : durée de service chez le client $i$ ($s_0=0$).
- $Q$ : capacité maximale identique de chaque camion.
- $m_{\max}$ : nombre maximal de camions utilisables.
- $[a_i,b_i]$ : fenêtre de livraison du client $i$.
- $t_0$ : heure de départ des camions depuis le dépôt.
- $b_{\max}$ : plus grande heure de fermeture des fenêtres ; $s_{\max}$ : plus longue durée de service ; $c_{\max}$ : plus long temps de trajet.
- $M=b_{\max}+s_{\max}+c_{\max}$ : constante suffisamment grande pour désactiver une contrainte temporelle lorsque l'arc correspondant n'est pas emprunté.

### Variables de décision

- $x_{ijk}\in\{0,1\}$ : vaut 1 si le camion $k$ emprunte l'arc $(i,j)$, et 0 sinon.
- $y_k\in\{0,1\}$ : vaut 1 si le camion $k$ est utilisé, et 0 sinon.
- $z_{ik}\in\{0,1\}$ : vaut 1 si le client $i$ est affecté au camion $k$, et 0 sinon.
- $h_{ik}\in[0,b_{\max}]$ : heure de début de livraison du client $i$ par le camion $k$, après une éventuelle attente.
- $w_i\geq 0$ : durée d'attente du camion avant la livraison du client $i$.

### Fonction objectif et contraintes

Le modèle complet s'écrit comme suit :

$$
\begin{aligned}
&\textbf{Minimiser :}
&& F=\sum_{k\in K}\sum_{(i,j)\in E}t_{ij}x_{ijk}
+\sum_{i\in N}s_i
+\sum_{i\in N}w_i
\\[3mm]
&\textbf{Sous contraintes :}
\\[2mm]
&\text{Chaque client est affecté une seule fois :}
&& \sum_{k\in K}z_{ik}=1
&& \forall i\in N
\\[3mm]
&\text{Affectation et arrivée chez le client :}
&& \sum_{j:(j,i)\in E}x_{jik}=z_{ik}
&& \forall i\in N,\ k\in K
\\[3mm]
&\text{Continuité de la tournée chez les clients :}
&& \sum_{j:(i,j)\in E}x_{ijk}=z_{ik}
&& \forall i\in N,\ k\in K
\\[3mm]
&\text{Départ du dépôt :}
&& \sum_{j:(0,j)\in E}x_{0jk}=y_k
&& \forall k\in K
\\[3mm]
&\text{Retour au dépôt :}
&& \sum_{i:(i,0)\in E}x_{i0k}=y_k
&& \forall k\in K
\\[3mm]
&\text{Nombre maximal de camions :}
&& \sum_{k\in K}y_k\leq m_{\max}
\\[3mm]
&\text{Capacité des camions :}
&& \sum_{i\in N}q_i z_{ik}\leq Q
&& \forall k\in K
\\[3mm]
&\text{Respect du début des fenêtres :}
&& h_{ik}\geq a_i-M(1-z_{ik})
&& \forall i\in N,\ k\in K
\\[3mm]
&\text{Respect de la fin des fenêtres :}
&& h_{ik}\leq b_i+M(1-z_{ik})
&& \forall i\in N,\ k\in K
\\[3mm]
&\text{Temps entre le dépôt et le premier client :}
&& h_{jk}\geq t_0+t_{0j}-M(1-x_{0jk})
&& \forall j\in N,\ (0,j)\in E,\ k\in K
\\[3mm]
&\text{Propagation du temps entre deux clients (avec durée de service) :}
&& h_{jk}\geq h_{ik}+s_i+t_{ij}-M(1-x_{ijk})
&& \forall i,j\in N,\ (i,j)\in E,\ k\in K
\\[3mm]
&\text{Calcul du temps d'attente depuis le dépôt :}
&& w_i\geq h_{ik}-t_0-c_{0i}-M(1-x_{0ik})
&& \forall i\in N,\ (0,i)\in E,\ k\in K
\\[3mm]
&\text{Calcul du temps d'attente entre deux clients (avec durée de service) :}
&& w_j\geq h_{jk}-h_{ik}-s_i-t_{ij}-M(1-x_{ijk})
&& \forall i,j\in N,\ (i,j)\in E,\ k\in K
\\[3mm]
&\text{Variables binaires :}
&& x_{ijk}\in\{0,1\}
&& \forall (i,j)\in E,\ k\in K
\\[2mm]
&&& y_k\in\{0,1\}
&& \forall k\in K
\\[2mm]
&&& z_{ik}\in\{0,1\}
&& \forall i\in N,\ k\in K
\\[3mm]
&\text{Variables temporelles :}
&& 0\leq h_{ik}\leq b_{\max}
&& \forall i\in N,\ k\in K
\\[2mm]
&&& w_i\geq 0
&& \forall i\in N
\end{aligned}
$$

## 5.12. Vérification finale par rapport au cahier des charges

| Exigence | Élément du modèle | Rôle |
|---|---|---|
| Routes autorisées uniquement | Ensemble $E$ | Empêche la sélection d'arcs inexistants |
| Temps de trajet | Paramètres $t_{ij}$ | Quantifie le temps nécessaire à chaque déplacement |
| Prévisions de trafic | Calcul des $t_{ij}$ | Intègre les temps de trajet prévus |
| Départ du dépôt | Contraintes de départ | Impose une sortie du dépôt pour chaque camion utilisé |
| Retour au dépôt | Contraintes de retour | Impose un retour pour chaque camion utilisé |
| Chaque client visité une fois | Contrainte de visite unique | Évite les clients oubliés ou visités plusieurs fois |
| Un seul camion par client | Variables $z_{ik}$ et contraintes d'affectation | Relie chaque client à un camion unique |
| Continuité des tournées | Conservation des flux | Assure une arrivée et un départ cohérents chez chaque client |
| Capacité des camions | Contraintes de capacité | Empêche les charges supérieures à $Q$ |
| Capacité identique | Paramètre commun $Q$ | Applique la même capacité à tous les camions |
| Nombre maximal de camions | Contrainte sur $\sum_k y_k$ et paramètre $m_{\max}$ | Limite le nombre de camions utilisés |
| Fenêtres de livraison | Contraintes sur $a_i$ et $b_i$ | Impose les horaires autorisés |
| Attente si arrivée trop précoce | Variables $h_{ik}$ et $w_i$, propagation temporelle | Permet de décaler la livraison jusqu'à l'ouverture et de compter l'attente |
| Cohérence temporelle | Contraintes de propagation | Respecte l'ordre des déplacements |
| Pas d'heure de fin de journée | Aucune contrainte sur l'heure de retour | Les équipes travaillent sans limite de temps |
| Durée de service chez chaque client | Paramètres $s_i$, contraintes de propagation (9.1) et d'attente (9.4), objectif et choix de $M$ | Le camion ne repart qu'après la fin du service |
| Élimination des sous-tournées | Contraintes temporelles avec temps positifs | Empêche les cycles de clients déconnectés du dépôt |
| Minimisation du temps global | Fonction objectif : trajets + services + attentes | Somme des durées de toutes les tournées |

Cette modélisation représente les principales contraintes du CVRPTW. Elle devra ensuite être traduite dans PuLP, puis testée sur des exemples simples afin de vérifier que les tournées obtenues respectent bien les contraintes.

Une attention particulière doit être portée à deux éléments avant l'implémentation : la cohérence des unités entre les temps de trajet $t_{ij}$ et les fenêtres de temps, et le calcul du grand $M$ à partir des données du problème.

## 6.2. Mesurer la qualité et la rapidité

Avant de comparer les algorithmes, nous devons définir les données de référence et les mesures utilisées.

### 6.2.1. Les instances et les solutions de référence

Nous utiliserons la bibliothèque **VRPLIB** pour récupérer les instances du VRPTW et, lorsque disponibles, leurs solutions de référence (*Best Known Solutions*, ou BKS).

```python

```

VRPLIB sert ici à **charger les instances et les solutions de référence**. Ce n'est pas l'outil qui réalise l'optimisation : les deux métaheuristiques seront programmées séparément dans notre projet.

### 6.2.2. Un écart calculé sur notre propre objectif

Le coût `cost` fourni par VRPLIB ne peut pas être comparé directement à notre fonction objectif, pour deux raisons :

- notre objectif $F=f(S)$ mesure le **temps global** (trajets, services et attentes), alors que les solutions de référence de la littérature ont été optimisées pour un autre critère (nombre de véhicules puis distance, pour les instances de Solomon) ;
- les conventions d'arrondi de la littérature peuvent différer de celles de VRPLIB, qui calcule des distances non arrondies.

Pour comparer des quantités identiques, nous **recalculons le coût de la solution de référence avec notre propre évaluateur** : nous prenons les tournées `reference["routes"]`, nous les parcourons avec les formules du §4.2.2 et nous obtenons $f_{\text{ref}}=f(S_{\text{ref}})$. L'écart d'une solution $S$ produite par notre algorithme est alors :

$$
\boxed{
\text{écart}(S)=100\times\frac{f(S)-f_{\text{ref}}}{f_{\text{ref}}}
}
$$

Notre algorithme répond à l'exigence de précision si $\text{écart}(S)<10\ \%$.

Un écart négatif signifie que notre solution obtient un coût inférieur à celui de la référence selon notre évaluateur.

**Point de vigilance :** les solutions de référence VRPTW ne reposent pas toutes sur exactement le même objectif ou les mêmes conventions de calcul. Certaines références cherchent d'abord à minimiser le nombre de véhicules, puis la distance, tandis que d'autres utilisent un objectif de coût total. Notre modèle doit donc être comparé à une référence compatible avec son objectif, son nombre de véhicules et ses règles de calcul. Si nous réévaluons une tournée de référence avec notre propre fonction objectif, nous parlerons d'un écart à cette tournée réévaluée ; nous ne présenterons pas automatiquement cet écart comme le GAP officiel du benchmark.

Cette vérification est importante pour éviter de comparer deux coûts qui ne représentent pas la même chose.

### 6.2.3. Mesurer le temps de calcul

Nous mesurerons le temps réel d'exécution avec `time.perf_counter()`.

Deux mesures seront conservées :

- **Temps total** : durée entre le lancement de l'algorithme et la fin de son exécution.
- **Temps d'accès à la cible** : premier instant où la meilleure solution trouvée atteint un GAP inférieur à 10 %.

Le temps de construction de la solution initiale sera inclus dans le temps total. Pour comparer les méthodes, elles disposeront du même budget de calcul sur une même instance.

## 6.4. Choix des deux métaheuristiques

Nous retenons la **recherche tabou** comme unique métaheuristique à implémenter et à étudier dans le cadre de ce projet. Ce choix permet de concentrer le temps de développement et les expérimentations sur une méthode dont le fonctionnement est explicable, dont les paramètres peuvent être étudiés progressivement et qui est adaptée aux problèmes d'optimisation combinatoire comme le CVRPTW.

### 6.4.1. Principe de la recherche tabou

La recherche tabou est une métaheuristique basée sur l'amélioration progressive d'une solution existante.

À chaque itération, l'algorithme examine des solutions voisines, obtenues par exemple en déplaçant un client, en échangeant deux clients ou en modifiant une partie d'une tournée. Il choisit ensuite le meilleur mouvement autorisé, même si celui-ci dégrade temporairement la solution.

Pour éviter de revenir immédiatement sur les mêmes mouvements, l'algorithme utilise une **liste tabou** qui mémorise temporairement les mouvements interdits. Cela permet de limiter les cycles et d'explorer de nouvelles solutions.

**Pourquoi nous l'avons choisie :**
- elle est connue et étudiée pour les problèmes de tournées avec fenêtres de temps ;
- son principe est assez simple à expliquer et à programmer (vu en workshop) ;
- elle permet d'améliorer une solution initiale avec des mouvements locaux ;
- elle possède un mécanisme de mémoire pour éviter de répéter continuellement les mêmes choix.

**Limites à prendre en compte :**
- l'évaluation d'un grand nombre de voisins peut devenir coûteuse lorsque le nombre de clients augmente ;
- la taille de la liste tabou et la durée d'interdiction doivent être réglées ;
- les résultats dépendent des mouvements utilisés et de la qualité de la solution initiale ;
- elle suit une seule trajectoire : si elle s'engage dans une mauvaise zone de l'espace des solutions, elle peut y rester longtemps.

### 6.4.3. Pourquoi ne pas avoir retenu les autres méthodes ?

D'autres métaheuristiques pourraient également être appliquées au CVRPTW, notamment les colonies de fourmis, le recuit simulé, l'ALNS ou les algorithmes génétiques. Elles ne sont pas intégrées à la présente étude afin de concentrer les efforts sur l'implémentation, la validation et le réglage approfondi d'une seule méthode. Ce choix permet de consacrer le temps disponible à la prise en compte correcte des contraintes du CVRPTW et à la réalisation d'expériences reproductibles.

### 6.4.4. Bilan du choix

La recherche tabou constitue donc la méthode retenue pour la suite du projet. Son intérêt sera évalué expérimentalement : le choix de cette méthode ne garantit pas à lui seul une qualité donnée. Les résultats obtenus sur les instances aléatoires, puis sur les instances de référence compatibles, permettront de déterminer dans quelle mesure elle répond à nos objectifs de faisabilité, de qualité et de temps de calcul.

## 7.2. Construire des instances de test

Nous voulons observer la méthode dans différentes situations, sans changer arbitrairement les contraintes : clients **regroupés (C)**, **dispersés (R)** ou **mixtes (RC)**, et créneaux **serrés (TW1)** ou **larges (TW2)**. Les profils sont *inspirés* des familles Solomon, mais les instances générées ne sont **pas** les benchmarks officiels.

Pour disposer d'un cas de départ exploitable, le générateur crée d'abord des tournées respectant la capacité, puis définit les fenêtres autour d'horaires réalisables. Cela garantit l'existence d'**au moins une solution faisable**, mais ne donne pas la meilleure solution : c'est justement ce que nous cherchons ensuite.

Nous fixons au maximum $\lceil n/4\rceil$ camions pour ces instances. Pour Solomon, nous utiliserons le nombre de véhicules fourni par le benchmark.

La fonction de génération ci-dessous **construit et renvoie** les données d’une instance. Elle ne lance pas encore la recherche tabou : l’instance de 12 clients sera créée et décrite juste après.

````python
# Paramètres modifiables pour toute la campagne expérimentale.
# La famille sert de référence de conception, pas de copie exacte du benchmark.
FAMILLE_REFERENCE = "Solomon"       # Choix : "Solomon" ou "Homberger"
PROFIL_GEOGRAPHIQUE = "RC"           # Choix : "C" (groupés), "R" (aléatoires), "RC" (mixte)
TYPE_FENETRES = 1                    # Choix : 1 (fenêtres serrées) ou 2 (fenêtres larges)
CAPACITE_VEHICULE = 200              # Repère commun inspiré des benchmarks Solomon/Homberger
DEMANDE_MIN, DEMANDE_MAX = 1, 40


def generer_instance_aleatoire(nb_clients, seed, capacite=None,
                               famille=FAMILLE_REFERENCE,
                               profil=PROFIL_GEOGRAPHIQUE,
                               type_fenetres=TYPE_FENETRES):
    """Génère une instance synthétique faisable, inspirée d'un profil de benchmark."""
    if capacite is None:
        capacite = CAPACITE_VEHICULE
    if famille not in ("Solomon", "Homberger"):
        raise ValueError("famille doit être 'Solomon' ou 'Homberger'.")
    if profil not in ("C", "R", "RC"):
        raise ValueError("profil doit être 'C', 'R' ou 'RC'.")
    if type_fenetres not in (1, 2):
        raise ValueError("type_fenetres doit être 1 (serrées) ou 2 (larges).")
    if nb_clients < 2:
        raise ValueError("Il faut au moins deux clients pour créer une instance.")

    rng = random.Random(seed)
    clients = list(range(1, nb_clients + 1))
    sommets = [0] + clients

    # Coordonnées sur une zone 100 x 100, avec trois profils inspirés
    # des familles Solomon/Homberger (sans reproduire leurs instances exactes).
    coordonnees = {0: (50.0, 50.0)}
    centres = [(20, 20), (75, 25), (25, 75), (75, 75)]
    for client in clients:
        utiliser_cluster = (profil == "C" or
                            (profil == "RC" and rng.random() < 0.7))
        if profil == "R" or not utiliser_cluster:
            coordonnees[client] = (rng.uniform(0, 100), rng.uniform(0, 100))
        else:
            cx, cy = rng.choice(centres)
            coordonnees[client] = (min(100, max(0, rng.gauss(cx, 8))),
                                   min(100, max(0, rng.gauss(cy, 8))))

    # Temps de trajet symétriques calculés à partir des distances euclidiennes.
    temps_trajet = {}
    arcs = set()
    for i in sommets:
        for j in sommets:
            if i != j:
                x1, y1 = coordonnees[i]
                x2, y2 = coordonnees[j]
                distance = math.hypot(x2 - x1, y2 - y1)
                temps_trajet[(i, j)] = max(1, round(0.7 * distance))
                arcs.add((i, j))

    demandes = {i: rng.randint(DEMANDE_MIN, DEMANDE_MAX) for i in clients}
    services = {i: rng.randint(5, 12) for i in clients}

    # Nombre maximal de camions disponible pour l'instance.
    # Cette limite est un paramètre de génération, pas une obligation d'utiliser tous les camions.
    nb_camions_max = max(1, math.ceil(nb_clients / 4))

    # On recommence seulement la génération des demandes si la capacité
    # totale de la flotte ne permet pas de transporter tous les clients.
    while sum(demandes.values()) > capacite * nb_camions_max:
        demandes = {i: rng.randint(DEMANDE_MIN, DEMANDE_MAX) for i in clients}

    # On cherche une répartition réalisable en commençant par le nombre
    # minimal de camions imposé par la demande totale, puis on augmente si besoin.
    ordre_clients = sorted(clients, key=lambda i: (-demandes[i], rng.random()))
    routes = None
    nb_camions_initial = math.ceil(sum(demandes.values()) / capacite)

    for nombre_camions in range(max(1, nb_camions_initial), nb_camions_max + 1):
        routes_essai = [[] for _ in range(nombre_camions)]
        charges = [0] * nombre_camions
        repartition_possible = True

        for client in ordre_clients:
            camions_possibles = [k for k in range(nombre_camions)
                                if charges[k] + demandes[client] <= capacite]
            if not camions_possibles:
                repartition_possible = False
                break

            # Best-fit : on choisit le camion qui aura le moins de place restante.
            camion = min(camions_possibles,
                         key=lambda k: capacite - charges[k] - demandes[client])
            routes_essai[camion].append(client)
            charges[camion] += demandes[client]

        if repartition_possible:
            routes = [route for route in routes_essai if route]
            break

    if routes is None:
        raise RuntimeError("La répartition des demandes n'a pas trouvé de tournée faisable.")

    # On ordonne ensuite les clients de chaque camion par proximité.
    routes_ordonnees = []
    for route in routes:
        a_visiter = set(route)
        courant = 0
        ordre = []
        while a_visiter:
            suivant = min(a_visiter, key=lambda j: temps_trajet[(courant, j)])
            ordre.append(suivant)
            a_visiter.remove(suivant)
            courant = suivant
        routes_ordonnees.append(tuple(ordre))

    # Construction de fenêtres horaires compatibles avec ces tournées.
    # Type 1 : fenêtres serrées ; type 2 : fenêtres plus larges.
    # Les marges sont des choix de génération synthétique, pas des valeurs de benchmark.
    if type_fenetres == 1:
        marge_ouverture = (0, 5)
        marge_fermeture = (20, 40)
    else:
        marge_ouverture = (0, 25)
        marge_fermeture = (60, 100)

    heure_depart = 0
    fenetres = {}
    for route in routes_ordonnees:
        heure_courante = heure_depart
        precedent = 0
        for client in route:
            arrivee = heure_courante + temps_trajet[(precedent, client)]
            ouverture = max(0, arrivee - rng.randint(*marge_ouverture))
            debut_service = max(arrivee, ouverture)
            fermeture = debut_service + rng.randint(*marge_fermeture)
            fenetres[client] = (ouverture, fermeture)
            heure_courante = debut_service + services[client]
            precedent = client

    instance = {
        "nom": f"ALEA-{famille}-{profil}-TW{type_fenetres}-{nb_clients}-seed{seed}",
        "famille_inspiration": famille,
        "profil_geographique": profil,
        "type_fenetres": type_fenetres,
        "seed": seed,
        "clients": tuple(clients),
        "sommets": tuple(sommets),
        "arcs": frozenset(arcs),
        "coordonnees": coordonnees,
        "temps_trajet": temps_trajet,
        "demandes": demandes,
        "services": services,
        "fenetres": fenetres,
        "capacite": capacite,
        "nb_camions_max": nb_camions_max,
        "heure_depart": heure_depart,
        "solution_initiale": tuple(routes_ordonnees),
    }
    return instance
````

## 7.3. Calculer la durée et contrôler les contraintes

L'évaluateur parcourt chaque tournée dans l'ordre des clients. À chaque arrêt, il additionne le trajet, le service et, si le camion arrive avant l'ouverture du créneau, le temps d'attente. **Une arrivée tardive ne crée pas un « retard » autorisé : la tournée est rejetée**. Il vérifie aussi la capacité, le nombre de camions et la visite unique de chaque client.

La fonction `evaluer_solution` renvoie les durées et les horaires d'une solution faisable, ou `None` si elle ne l'est pas. Elle n'affiche rien lorsqu'on la définit : nous l'utiliserons pour sélectionner les tournées dans la recherche tabou et pour construire les graphiques.

La fonction renvoie les durées détaillées si les tournées sont réalisables, et `None` sinon. Nous l’appliquons au départ de notre exemple dans la partie suivante, plutôt que d’afficher un message artificiel à sa définition.

````python
def normaliser_solution(routes):
    """Enlève les tournées vides et donne un ordre stable aux tournées."""
    routes_non_vides = [tuple(route) for route in routes if len(route) > 0]
    return tuple(sorted(routes_non_vides))


def evaluer_solution(instance, solution):
    """Vérifie la faisabilité et calcule les durées, les attentes et le planning."""
    routes = normaliser_solution(solution)
    clients = set(instance["clients"])
    visites = [client for route in routes for client in route]

    # Chaque client doit apparaître exactement une fois.
    if len(visites) != len(clients) or set(visites) != clients:
        return None
    if len(routes) > instance["nb_camions_max"] or len(routes) == 0:
        return None

    temps_trajet = instance["temps_trajet"]
    demandes = instance["demandes"]
    services = instance["services"]
    fenetres = instance["fenetres"]
    arcs = instance["arcs"]
    heure_depart = instance["heure_depart"]

    total_trajet = 0.0
    total_service = 0.0
    total_attente = 0.0
    durees_tournees = []
    planning = []
    details_routes = []

    for numero_camion, route in enumerate(routes, start=1):
        charge = sum(demandes[i] for i in route)
        if charge > instance["capacite"] + 1e-7:
            return None

        courant = 0
        heure = heure_depart
        trajet_route = 0.0
        service_route = 0.0
        attente_route = 0.0

        for client in route:
            if (courant, client) not in arcs:
                return None

            duree = temps_trajet[(courant, client)]
            arrivee = heure + duree
            ouverture, fermeture = fenetres[client]
            debut_service = max(arrivee, ouverture)
            attente = debut_service - arrivee

            # La fenêtre porte sur le début du service.
            if debut_service > fermeture + 1e-7:
                return None

            fin_service = debut_service + services[client]

            if duree > 0:
                planning.append({"camion": numero_camion, "type": "Trajet",
                                 "debut": heure, "fin": arrivee,
                                 "libelle": f"{courant} → {client}"})
            if attente > 0:
                planning.append({"camion": numero_camion, "type": "Attente",
                                 "debut": arrivee, "fin": debut_service,
                                 "libelle": f"Attente client {client}"})
            if services[client] > 0:
                planning.append({"camion": numero_camion, "type": "Service",
                                 "debut": debut_service, "fin": fin_service,
                                 "libelle": f"Client {client}"})

            trajet_route += duree
            service_route += services[client]
            attente_route += attente
            heure = fin_service
            courant = client

        # Retour obligatoire au dépôt, sans heure limite de retour.
        if (courant, 0) not in arcs:
            return None
        duree_retour = temps_trajet[(courant, 0)]
        arrivee_depot = heure + duree_retour
        # Contraintes de retour uniquement pour les benchmarks qui en fournissent une.
        if arrivee_depot > instance.get("heure_limite_depot", math.inf) + 1e-7:
            return None
        planning.append({"camion": numero_camion, "type": "Trajet",
                         "debut": heure, "fin": arrivee_depot,
                         "libelle": f"{courant} → 0"})
        trajet_route += duree_retour

        duree_route = trajet_route + service_route + attente_route
        total_trajet += trajet_route
        total_service += service_route
        total_attente += attente_route
        durees_tournees.append(duree_route)
        details_routes.append({
            "camion": numero_camion,
            "route": (0,) + tuple(route) + (0,),
            "charge": charge,
            "temps_trajet": trajet_route,
            "temps_service": service_route,
            "temps_attente": attente_route,
            "duree_totale": duree_route,
            "heure_retour": arrivee_depot,
        })

    cout_total = total_trajet + total_service + total_attente
    return {
        "cout_total": cout_total,
        "temps_trajet": total_trajet,
        "temps_service": total_service,
        "temps_attente": total_attente,
        "durees_tournees": durees_tournees,
        "planning": planning,
        "details_routes": details_routes,
    }
````

## 7.4. Générer des voisins et appliquer la recherche tabou

Nous cherchons à améliorer une solution en changeant **l'ordre de livraison** ou **l'affectation des clients aux camions** : déplacement d'un client (*relocate*), échange (*swap*), inversion d'un segment (*2-opt*), déplacement d'un petit groupe (*Or-opt*) ou échange de fins de tournées (*2-opt*). Un camion libre peut être utilisé, sans dépasser la flotte autorisée.

`generer_voisins` **renvoie un ensemble de solutions candidates, pas un booléen ni un résultat à afficher directement**. La recherche tabou évalue ensuite ces candidats et écarte ceux qui ne respectent pas les contraintes. Pour éviter que la recherche revienne immédiatement sur ses derniers mouvements, nous utilisons une liste tabou qui mémorise les arcs supprimés lors des déplacements précédents. Leur réintroduction est temporairement interdite pendant un nombre défini de mouvements. Toutefois, un mouvement tabou peut être accepté s'il améliore la meilleure solution trouvée jusqu'à présent : c'est le critère d'aspiration. Ce mécanisme permet de diversifier la recherche et de limiter les retours vers les mêmes configurations.

Sur la petite instance, nous examinons tous les voisins. Sur les grandes, `voisins_echantillonnes` en sélectionne un nombre limité pour maîtriser le calcul. **La simple définition de ces fonctions ne produit pas d’affichage.** Après avoir construit notre départ, nous compterons les voisins générés et ceux qui respectent réellement les contraintes. Nous verrons ensuite leur effet sur la convergence.

````python
def generer_voisins(instance, solution):
    """Génère des voisins : relocation, ouverture, swap, 2-opt, Or-opt et 2-opt*."""
    routes = [list(route) for route in normaliser_solution(solution)]
    nb_routes = len(routes)
    voisins = set()

    # 1) Relocation : déplacer un client dans sa tournée ou vers une autre tournée.
    for r in range(nb_routes):
        for position_client in range(len(routes[r])):
            client = routes[r][position_client]

            # 1a) Réinsertion dans la même tournée (on change l'ordre des clients).
            base_meme_route = [list(route) for route in routes]
            base_meme_route[r].pop(position_client)
            for position_insertion in range(len(base_meme_route[r]) + 1):
                candidat = [list(route) for route in base_meme_route]
                candidat[r].insert(position_insertion, client)
                voisin = normaliser_solution(candidat)
                if voisin != normaliser_solution(solution):
                    voisins.add(voisin)

            # 1b) Déplacement vers une autre tournée (on change aussi l'affectation).
            for s in range(nb_routes):
                if s == r:
                    continue

                base = [list(route) for route in routes]
                base[r].pop(position_client)
                destination = s
                if len(base[r]) == 0:
                    base.pop(r)
                    if s > r:
                        destination -= 1

                for position_insertion in range(len(base[destination]) + 1):
                    candidat = [list(route) for route in base]
                    candidat[destination].insert(position_insertion, client)
                    voisin = normaliser_solution(candidat)
                    if voisin != normaliser_solution(solution):
                        voisins.add(voisin)

    # 1c) Ouverture d'une nouvelle tournée si un camion reste disponible.
    # Ce mouvement est indispensable : le nombre de tournées ne doit pas rester
    # artificiellement bloqué au nombre choisi par la solution initiale.
    if nb_routes < instance["nb_camions_max"]:
        for r in range(nb_routes):
            for i, client in enumerate(routes[r]):
                candidat = [list(route) for route in routes]
                candidat[r].pop(i)
                candidat.append([client])
                voisins.add(normaliser_solution(candidat))

    # 2) Swap : échanger deux clients de deux tournées différentes.
    for r1 in range(nb_routes):
        for r2 in range(r1 + 1, nb_routes):
            for i in range(len(routes[r1])):
                for j in range(len(routes[r2])):
                    candidat = [list(route) for route in routes]
                    candidat[r1][i], candidat[r2][j] = candidat[r2][j], candidat[r1][i]
                    voisins.add(normaliser_solution(candidat))

    # 3) 2-opt : inverser un segment d'une tournée.
    for r in range(nb_routes):
        route = routes[r]
        for i in range(len(route) - 1):
            for j in range(i + 1, len(route)):
                nouvelle_route = route[:i] + route[i:j + 1][::-1] + route[j + 1:]
                candidat = [list(element) for element in routes]
                candidat[r] = nouvelle_route
                voisins.add(normaliser_solution(candidat))


    # 4) Or-opt : déplacer un segment de 2 ou 3 clients consécutifs (sens normal ou inversé).
    base_norm = normaliser_solution(solution)
    for r in range(nb_routes):
        for longueur in (2, 3):
            for i in range(len(routes[r]) - longueur + 1):
                segment = routes[r][i:i + longueur]
                reste = routes[r][:i] + routes[r][i + longueur:]
                for s in range(nb_routes + 1):
                    if s == nb_routes:
                        # Nouvelle tournée : seulement si un camion est encore disponible.
                        nb_apres = nb_routes - (0 if reste else 1) + 1
                        if nb_apres > instance["nb_camions_max"]:
                            continue
                        cible = []
                    else:
                        cible = reste if s == r else routes[s]
                    for p in range(len(cible) + 1):
                        for sens in (segment, segment[::-1]):
                            candidat = [list(x) for x in routes]
                            nouvelle = cible[:p] + sens + cible[p:]
                            if s == r:
                                candidat[r] = nouvelle
                            elif s == nb_routes:
                                candidat[r] = reste
                                candidat.append(nouvelle)
                            else:
                                candidat[r] = reste
                                candidat[s] = nouvelle
                            voisin = normaliser_solution(candidat)
                            if voisin != base_norm:
                                voisins.add(voisin)

    # 5) 2-opt* : couper deux tournées et échanger leurs fins.
    for r1 in range(nb_routes):
        for r2 in range(r1 + 1, nb_routes):
            for i in range(len(routes[r1]) + 1):
                for j in range(len(routes[r2]) + 1):
                    if (i == 0 and j == 0) or (i == len(routes[r1]) and j == len(routes[r2])):
                        continue
                    candidat = [list(x) for x in routes]
                    candidat[r1] = routes[r1][:i] + routes[r2][j:]
                    candidat[r2] = routes[r2][:j] + routes[r1][i:]
                    voisins.add(normaliser_solution(candidat))

    return voisins

def voisins_echantillonnes(instance, solution, maximum=140, graine=0):
    """Voisinage borné pour les instances volumineuses : mêmes types de mouvements."""
    rng = random.Random(graine)
    routes = [list(r) for r in normaliser_solution(solution)]
    candidats = set()
    if not routes:
        return []
    for _ in range(5 * maximum):
        if len(candidats) >= maximum:
            break
        essai = [r[:] for r in routes]
        disponibles = [k for k, route in enumerate(essai) if route]
        if not disponibles:
            break
        action = rng.choice(("relocate", "swap", "2opt", "oropt", "2opt_etoile"))
        a = rng.choice(disponibles)
        if action == "relocate":
            i = rng.randrange(len(essai[a])); client = essai[a].pop(i)
            destinations = list(range(len(essai)))
            if len(essai) < instance["nb_camions_max"]:
                destinations.append(len(essai))
            b = rng.choice(destinations)
            if b == len(essai):
                essai.append([])
            essai[b].insert(rng.randrange(len(essai[b]) + 1), client)
        elif action == "swap" and len(disponibles) >= 2:
            b = rng.choice([x for x in disponibles if x != a]); i = rng.randrange(len(essai[a])); j = rng.randrange(len(essai[b]))
            essai[a][i], essai[b][j] = essai[b][j], essai[a][i]
        elif action == "2opt" and len(essai[a]) >= 2:
            i, j = sorted(rng.sample(range(len(essai[a])), 2))
            essai[a][i:j+1] = reversed(essai[a][i:j+1])
        elif action == "oropt" and len(essai[a]) >= 3:
            taille = rng.choice((2,3)); i = rng.randrange(len(essai[a])-taille+1)
            bloc = essai[a][i:i+taille]; del essai[a][i:i+taille]
            b = rng.choice(range(len(essai)))
            p = rng.randrange(len(essai[b])+1); essai[b][p:p] = bloc
        elif action == "2opt_etoile" and len(disponibles) >= 2:
            b = rng.choice([x for x in disponibles if x != a])
            i = rng.randrange(len(essai[a])+1); j = rng.randrange(len(essai[b])+1)
            essai[a], essai[b] = essai[a][:i]+essai[b][j:], essai[b][:j]+essai[a][i:]
        voisin = normaliser_solution(essai)
        if voisin != normaliser_solution(solution):
            candidats.add(voisin)
    return sorted(candidats)
````

### Choisir un départ pour la recherche

Il faut d'abord une tournée réalisable. Nous construisons un **départ glouton** : les clients aux créneaux les plus urgents sont placés dans la position faisable entraînant le plus faible surcoût de durée. Ce départ sert uniquement de point de départ à la recherche tabou ; ce n'est pas une deuxième méthode que nous évaluons séparément.

La recherche teste ensuite les voisins, choisit le meilleur mouvement admissible et conserve la **meilleure solution rencontrée**, même si certaines étapes intermédiaires ont une durée plus élevée. Nous utilisons initialement une liste tabou de taille 5 et au plus 60 itérations : ces réglages fixent un budget commun pour comparer les expériences.

**Ce que nous allons observer :** la durée de ce départ, puis le nombre de modifications possibles et le nombre réellement réalisable. Cela permet de comprendre le rôle du voisinage avant de regarder le résultat optimisé.

````python
def evaluer_tournee_seule(instance, route):
    """Durée d'une tournée (trajet + service + attente), ou None si capacité ou fenêtre violée."""
    if sum(instance["demandes"][i] for i in route) > instance["capacite"]:
        return None
    heure = instance["heure_depart"]
    courant = 0
    for client in route:
        arrivee = heure + instance["temps_trajet"][(courant, client)]
        ouverture, fermeture = instance["fenetres"][client]
        debut = max(arrivee, ouverture)
        if debut > fermeture:
            return None
        heure = debut + instance["services"][client]
        courant = client
    fin_route = heure + instance["temps_trajet"][(courant, 0)]
    if fin_route > instance.get("heure_limite_depot", math.inf) + 1e-7:
        return None
    return fin_route - instance["heure_depart"]


def solution_initiale_gloutonne(instance):
    """Insertion au moindre surcoût, clients triés par fermeture de fenêtre croissante."""
    ordre = sorted(instance["clients"],
                   key=lambda i: (instance["fenetres"][i][1], instance["fenetres"][i][0]))
    routes = []
    for client in ordre:
        meilleur = None
        for r in range(len(routes) + 1):
            if r == len(routes) and len(routes) >= instance["nb_camions_max"]:
                continue
            base = routes[r] if r < len(routes) else []
            cout_base = evaluer_tournee_seule(instance, base) if base else 0
            for p in range(len(base) + 1):
                nouvelle = base[:p] + [client] + base[p:]
                cout = evaluer_tournee_seule(instance, nouvelle)
                if cout is None:
                    continue
                surcout = cout - cout_base
                if meilleur is None or surcout < meilleur[0]:
                    meilleur = (surcout, r, nouvelle)
        if meilleur is None:
            # Secours : la solution du générateur est toujours faisable.
            if "solution_initiale" in instance:
                return normaliser_solution(instance["solution_initiale"])
            raise ValueError("Aucune insertion faisable trouvée avec la flotte disponible.")
        _, r, nouvelle = meilleur
        if r == len(routes):
            routes.append(nouvelle)
        else:
            routes[r] = nouvelle
    return normaliser_solution(routes)



depart_demo = solution_initiale_gloutonne(instance_demo)

# Application des fonctions précédentes au même exemple de 12 clients.
# On n'optimise pas encore : on décrit seulement le point de départ de la recherche.
evaluation_depart_demo = evaluer_solution(instance_demo, depart_demo)
if evaluation_depart_demo is None:
    raise RuntimeError("Le départ construit ne respecte pas les contraintes CVRPTW.")

voisins_depart_demo = generer_voisins(instance_demo, depart_demo)
nb_voisins_faisables = sum(
    evaluer_solution(instance_demo, voisin) is not None
    for voisin in voisins_depart_demo
)
print(f"Instance de référence : {instance_demo['nom']}")
print(f"Départ faisable : {len(depart_demo)} camion(s), "
      f"durée totale = {evaluation_depart_demo['cout_total']:.0f}")
print(f"Voisinage du départ : {len(voisins_depart_demo)} modifications générées, "
      f"dont {nb_voisins_faisables} respectent toutes les contraintes.")
print("La recherche tabou explorera ces solutions admissibles pour améliorer la durée.")
````

````python

def recherche_tabou(instance, solution_initiale, taille_tabou=5, iter_max=60,
                     limite_voisins=None, graine=0, budget_secondes=None,
                     progression=False, libelle_progression="Recherche tabou"):
    """Minimise la SOMME des durées des tournées avec mémoire des mouvements.

    La liste tabou mémorise les arcs supprimés lors des derniers mouvements
    acceptés, afin d'éviter leur réintroduction immédiate. L'aspiration
    autorise toutefois un mouvement tabou s'il bat le meilleur coût connu.
    """
    if taille_tabou < 1 or iter_max < 1:
        raise ValueError("taille_tabou et iter_max doivent être positifs.")

    debut = time.perf_counter()
    rng = random.Random(graine)
    courante = normaliser_solution(solution_initiale)
    evaluation = evaluer_solution(instance, courante)
    if evaluation is None:
        raise ValueError("La solution initiale ne respecte pas les contraintes.")

    cout_courant = evaluation["cout_total"]       # Trajets + services + attentes
    meilleure = courante
    cout_meilleur = cout_courant

    # Chaque entrée contient les arcs supprimés PAR UN mouvement accepté.
    derniers_mouvements = deque()
    occurrences_taboues = {}  # Un arc peut être supprimé lors de plusieurs mouvements.

    def arcs_de(routes):
        return {
            (i, j)
            for route in routes
            for i, j in zip((0,) + tuple(route), tuple(route) + (0,))
        }

    def memoriser_arcs_supprimes(arcs_supprimes):
        if len(derniers_mouvements) >= taille_tabou:
            anciens = derniers_mouvements.popleft()
            for arc in anciens:
                occurrences_taboues[arc] -= 1
                if occurrences_taboues[arc] == 0:
                    del occurrences_taboues[arc]
        groupe = frozenset(arcs_supprimes)
        derniers_mouvements.append(groupe)
        for arc in groupe:
            occurrences_taboues[arc] = occurrences_taboues.get(arc, 0) + 1

    # Cache des coûts pour éviter les réévaluations identiques.
    cache = {courante: cout_courant}
    nb_evaluations = 1
    nb_mouvements_tabous = 0
    nb_aspirations = 0
    historique_courant = [cout_courant]
    historique_meilleur = [cout_meilleur]

    barre = BarreProgression(iter_max, libelle_progression,
                             afficher=progression, effacer=True)
    try:
        for _ in range(iter_max):
            # Budget souple : vérifié entre les itérations.
            if budget_secondes is not None and time.perf_counter() - debut >= budget_secondes:
                break

            arcs_courants = arcs_de(courante)
            if limite_voisins is None:
                candidats = sorted(generer_voisins(instance, courante))
            else:
                candidats = voisins_echantillonnes(
                    instance, courante, limite_voisins, rng.randrange(10**9)
                )

            voisin_retenu = None
            cout_retenu = float("inf")
            arcs_retenus = None

            for voisin in candidats:
                # Benchmarks à flotte fixée : ne pas supprimer ou créer un camion.
                # Sur nos instances synthétiques, cette contrainte n'est pas activée.
                if (instance.get("nb_camions_exacts") is not None
                        and len(voisin) != instance["nb_camions_exacts"]):
                    continue

                if voisin not in cache:
                    ev = evaluer_solution(instance, voisin)
                    cache[voisin] = None if ev is None else ev["cout_total"]
                    nb_evaluations += 1

                cout = cache[voisin]
                if cout is None:  # Capacités, visites ou fenêtres non respectées.
                    continue

                arcs_voisin = arcs_de(voisin)
                arcs_ajoutes = arcs_voisin - arcs_courants
                est_tabou = any(a in occurrences_taboues for a in arcs_ajoutes)

                # Critère d'aspiration : un nouveau record annule l'interdiction.
                if est_tabou and cout >= cout_meilleur - 1e-9:
                    nb_mouvements_tabous += 1
                    continue

                if cout < cout_retenu:
                    voisin_retenu, cout_retenu = voisin, cout
                    arcs_retenus = arcs_voisin
                    aspiration_retenue = est_tabou

            if voisin_retenu is None:
                break

            # Ce sont les arcs retirés, et NON la solution entière, qui sont tabous.
            memoriser_arcs_supprimes(arcs_courants - arcs_retenus)
            courante, cout_courant = voisin_retenu, cout_retenu
            if aspiration_retenue:
                nb_aspirations += 1

            if cout_courant < cout_meilleur - 1e-9:
                meilleure, cout_meilleur = courante, cout_courant

            historique_courant.append(cout_courant)
            historique_meilleur.append(cout_meilleur)
            barre.update(1)
            if progression:
                barre.set_postfix(duree=f"{cout_meilleur:.1f}",
                                 temps=f"{time.perf_counter() - debut:.1f}s")

            if len(cache) > 5000:
                cache = {courante: cout_courant, meilleure: cout_meilleur}
    finally:
        barre.close()

    informations = {
        "nb_iterations_effectuees": len(historique_meilleur) - 1,
        "nb_solutions_evaluees": nb_evaluations,
        "mouvements_tabous_refuses": nb_mouvements_tabous,
        "aspirations": nb_aspirations,
    }
    return meilleure, historique_courant, historique_meilleur, informations
````

## 7.6. Évaluer la qualité de la solution : optimum et borne inférieure

Pour notre instance témoin, nous voulons savoir **à quel point la durée 442 est bonne**, et pas seulement constater qu'elle est inférieure à 513. Deux références sont possibles :

- **Optimum exact** $F^*$ : plus petite durée réalisable, **prouvée** par une méthode exacte. Le $GAP_{exact}=100(F_{tabou}-F^*)/F^*$ donne directement l'écart de qualité de notre solution.
- **Borne inférieure LP** $F_{LP}$ : valeur obtenue en relâchant les variables entières du PLNE. Cette valeur est **au plus égale** à l'optimum, mais peut être beaucoup plus faible, notamment parce qu'elle autorise des décisions fractionnaires. Le $GAP_{LP}=100(F_{tabou}-F_{LP})/F_{LP}$ mesure l'écart à cette borne, **pas l'écart réel à l'optimum**.

On s'attend donc à $F_{LP}\leq F^*\leq F_{tabou}$, et un **GAP LP supérieur à 10 % ne signifie pas que notre recherche tabou est mauvaise**. Nous calculons les deux sur les petits cas pour comprendre leurs différences. Le calcul exact devient trop coûteux lorsque le nombre de clients augmente : c'est une des raisons du passage aux BKS Solomon plus loin.

Les fonctions suivantes calculent les références ; **leur application à notre instance**, dans la cellule suivante, affiche les valeurs et les GAP. Si PuLP/CBC n'est pas disponible, le GAP LP reste explicitement non calculé.

````python
def calculer_borne_inferieure_lp(instance, temps_limite=120):
    """Résout la relaxation continue de référence, avec formulation renforcée."""
    clients = list(instance["clients"])
    sommets = list(instance["sommets"])
    camions = list(range(instance["nb_camions_max"]))
    T = instance["temps_trajet"]
    q = instance["demandes"]
    s = instance["services"]
    W = instance["fenetres"]
    t0 = instance["heure_depart"]
    Q = instance["capacite"]
    b_max = max(b for a, b in W.values())

    # Plus courts chemins entre clients (service + trajet) : bornes valables même si
    # l'inégalité triangulaire n'est pas parfaitement respectée par les temps arrondis.
    sp = {(u, v): (0 if u == v else s[u] + T[(u, v)]) for u in clients for v in clients}
    for m in clients:
        for u in clients:
            for v in clients:
                if sp[(u, m)] + sp[(m, v)] < sp[(u, v)]:
                    sp[(u, v)] = sp[(u, m)] + sp[(m, v)]
    # Début de service au plus tôt de chaque client.
    e = {}
    for i in clients:
        arrivee_min = min([T[(0, i)]] + [T[(0, k)] + s[k] + sp[(k, i)] for k in clients if k != i])
        e[i] = max(W[i][0], t0 + arrivee_min)

    # Arcs conservés : on supprime ceux qu'aucune solution faisable ne peut emprunter.
    arcs = []
    for (i, j) in instance["arcs"]:
        if i != 0 and j != 0:
            if e[i] + s[i] + T[(i, j)] > W[j][1]:
                continue
            if q[i] + q[j] > Q:
                continue
        arcs.append((i, j))
    A = set(arcs)

    P = pulp.LpProblem("CVRPTW_relaxation_reference", pulp.LpMinimize)
    x = pulp.LpVariable.dicts("x", [(i, j, k) for (i, j) in arcs for k in camions], 0, 1)
    y = pulp.LpVariable.dicts("y", camions, 0, 1)
    z = pulp.LpVariable.dicts("z", [(i, k) for i in clients for k in camions], 0, 1)
    h = pulp.LpVariable.dicts("h", [(i, k) for i in clients for k in camions], 0, b_max)
    w = pulp.LpVariable.dicts("w", clients, 0)
    for i in clients:
        for k in camions:
            h[(i, k)].lowBound = W[i][0]
            h[(i, k)].upBound = W[i][1]

    P += (pulp.lpSum(T[(i, j)] * x[(i, j, k)] for (i, j) in arcs for k in camions)
          + sum(s.values()) + pulp.lpSum(w[i] for i in clients))

    for i in clients:
        P += pulp.lpSum(z[(i, k)] for k in camions) == 1
        for k in camions:
            P += pulp.lpSum(x[(j, i, k)] for j in sommets if (j, i) in A) == z[(i, k)]
            P += pulp.lpSum(x[(i, j, k)] for j in sommets if (i, j) in A) == z[(i, k)]
    for k in camions:
        P += pulp.lpSum(x[(0, j, k)] for j in clients if (0, j) in A) == y[k]
        P += pulp.lpSum(x[(i, 0, k)] for i in clients if (i, 0) in A) == y[k]
        P += pulp.lpSum(q[i] * z[(i, k)] for i in clients) <= Q * y[k]
    P += pulp.lpSum(y[k] for k in camions) <= instance["nb_camions_max"]

    # Propagation du temps et attentes, avec un M propre à chaque arc.
    for (i, j) in arcs:
        if j == 0:
            continue
        for k in camions:
            if i == 0:
                M1 = t0 + T[(0, j)]
                M2 = max(0, W[j][1] - t0 - T[(0, j)])
                P += h[(j, k)] >= t0 + T[(0, j)] - M1 * (1 - x[(0, j, k)])
                P += w[j] >= h[(j, k)] - t0 - T[(0, j)] - M2 * (1 - x[(0, j, k)])
            else:
                M1 = max(0, W[i][1] + s[i] + T[(i, j)] - W[j][0])
                M2 = max(0, W[j][1] - W[i][0] - s[i] - T[(i, j)])
                P += h[(j, k)] >= h[(i, k)] + s[i] + T[(i, j)] - M1 * (1 - x[(i, j, k)])
                P += w[j] >= h[(j, k)] - h[(i, k)] - s[i] - T[(i, j)] - M2 * (1 - x[(i, j, k)])

    # Inégalités valides.
    for (i, j) in arcs:
        for k in camions:
            if i != 0:
                P += x[(i, j, k)] <= z[(i, k)]
            if j != 0:
                P += x[(i, j, k)] <= z[(j, k)]
            if i != 0 and j != 0 and i < j and (j, i) in A:
                P += x[(i, j, k)] + x[(j, i, k)] <= z[(i, k)]
    for i in clients:
        for j in clients:
            if i < j and e[i] + sp[(i, j)] > W[j][1] and e[j] + sp[(j, i)] > W[i][1]:
                for k in camions:
                    P += z[(i, k)] + z[(j, k)] <= 1

    solveur = pulp.PULP_CBC_CMD(msg=False, timeLimit=temps_limite)
    if not solveur.available():
        raise RuntimeError('Solveur CBC absent ; installer %pip install "pulp[cbc]==3.3.2"')
    P.solve(solveur)
    statut = pulp.LpStatus[P.status]
    if statut != "Optimal":
        raise RuntimeError(f"La relaxation renforcée n'est pas optimale (statut : {statut}).")
    return float(pulp.value(P.objective))


def optimum_exact_dp(instance, n_max=14):
    """Optimum exact par programmation dynamique (Held-Karp + partition en au plus k tournées).

    Renvoie None si l'instance a plus de n_max clients (calcul trop long)."""
    clients = list(instance["clients"])
    n = len(clients)
    if n > n_max:
        return None
    T, q, s, W = instance["temps_trajet"], instance["demandes"], instance["services"], instance["fenetres"]
    Q, t0, K = instance["capacite"], instance["heure_depart"], instance["nb_camions_max"]
    INF = float("inf")

    # f[masque][j] : heure de fin de service au plus tôt en j, après avoir visité 'masque' et fini en j.
    f = [[INF] * (n + 1) for _ in range(1 << n)]
    for j in range(1, n + 1):
        debut = max(t0 + T[(0, j)], W[j][0])
        if debut <= W[j][1]:
            f[1 << (j - 1)][j] = debut + s[j]
    cout_tournee = [INF] * (1 << n)
    for masque in range(1, 1 << n):
        if sum(q[j + 1] for j in range(n) if masque >> j & 1) > Q:
            continue
        for j in range(1, n + 1):
            fin = f[masque][j]
            if fin == INF:
                continue
            cout_tournee[masque] = min(cout_tournee[masque], fin + T[(j, 0)] - t0)
            for k in range(1, n + 1):
                if masque >> (k - 1) & 1:
                    continue
                debut = max(fin + T[(j, k)], W[k][0])
                if debut <= W[k][1]:
                    nouveau = masque | 1 << (k - 1)
                    if debut + s[k] < f[nouveau][k]:
                        f[nouveau][k] = debut + s[k]

    # Partition des clients en au plus K tournées.
    g = [0.0] + [INF] * ((1 << n) - 1)
    for _ in range(K):
        g_suivant = list(g)
        for masque in range(1, 1 << n):
            sous = masque
            while sous:
                if cout_tournee[sous] < INF and g[masque ^ sous] + cout_tournee[sous] < g_suivant[masque]:
                    g_suivant[masque] = g[masque ^ sous] + cout_tournee[sous]
                sous = (sous - 1) & masque
        g = g_suivant
    return g[(1 << n) - 1]

optimum_demo = optimum_exact_dp(instance_demo)
borne_demo = None
if pulp is not None:
    try:
        borne_demo = calculer_borne_inferieure_lp(instance_demo, temps_limite=30)
    except Exception as erreur:
        print("Borne LP non calculée :", erreur)
else:
    print('PuLP absent : installer %pip install "pulp[cbc]==3.3.2" pour obtenir la borne LP.')

f_tabou = evaluation_demo["cout_total"]
print(f"Instance : {instance_demo['nom']} | Durée tabou : {f_tabou:.1f} | Optimum exact : {optimum_demo:.1f}")
print(f"GAP à l'optimum exact : {100 * (f_tabou-optimum_demo)/optimum_demo:.2f} %")
if borne_demo is not None:
    assert borne_demo <= optimum_demo + 1e-5 <= f_tabou + 1e-5
    print(f"Borne inférieure LP : {borne_demo:.2f} | GAP à la borne LP : {100 * (f_tabou-borne_demo)/borne_demo:.2f} %")
else:
    print("GAP LP : non disponible")
````

**Interprétation.** Sur notre instance de référence, le calcul exact confirme que **442 est bien la durée minimale possible** : la recherche tabou atteint donc un **GAP exact de 0 %** sur ce cas. Si la relaxation LP fournit une valeur inférieure à 442, son GAP peut rester positif **malgré cette optimalité prouvée**. Les deux mesures apportent donc des informations différentes. Cette conclusion porte uniquement sur l'instance de 12 clients ; nous devons maintenant répéter les essais sur des données variées.

## 7.7. Tester la méthode sur plusieurs instances

**Pourquoi continuer ?** Trouver l'optimum sur **notre unique instance de 12 clients** ne prouve pas que la recherche sera aussi efficace avec d'autres positions, demandes ou fenêtres. Nous construisons donc une campagne de **24 instances distinctes** :

- **4 tailles** : 8, 10, 12 et 15 clients ;
- **3 répartitions** : C, R et RC ;
- **2 types de fenêtres** : TW1 (serrées) et TW2 (larges).

Soit **4 × 3 × 2 = 24 cas** : nous visons au minimum **20 instances réellement exécutées**, conformément au sujet. Chaque combinaison est générée avec une graine fixe et un nom unique. Une seule graine par scénario reste une première exploration : pour conclure plus solidement sur la robustesse, il faudrait ensuite plusieurs tirages indépendants par catégorie.

**Pourquoi le tableau peut-il afficher « 3/24 » ?** Le paramètre `NOUVELLES_INSTANCES_PAR_LANCEMENT = 3` limite volontairement cette cellule à **trois nouveaux cas par exécution** : les expériences peuvent être longues, surtout lorsqu'on demande l'optimum ou la borne LP. Après trois succès, le tableau indique **3 terminées sur 24 prévues**, et non un problème de calcul. À la prochaine relance, les trois premières sont reconnues dans les archives et nous poursuivons. On peut aussi passer ce paramètre à `24` pour essayer tous les cas restants en une fois. **Après une relance, nous réexécutons aussi la cellule des graphiques pour actualiser leur interprétation.**

Pour chaque **même instance**, nous comparons une recherche tabou simple et un **multi-start**, qui relance cette recherche avec jusqu'à trois **départs faisables différents**. Nous relevons la durée, le temps de calcul, puis les GAP exact et LP lorsqu'ils sont accessibles. Le **442 reste propre à l'instance témoin**, et ne doit pas être comparé aux durées des autres instances.

Le bloc suivant définit le multi-start et les sauvegardes. Ces fonctions n’affichent rien tant qu’elles ne sont pas appelées ; **la cellule de campagne qui suit** les utilise et produit le tableau des résultats, les GAP disponibles et le bilan de progression.

````python
# Multi-start : créer un autre point de départ faisable sur la MÊME instance.
# Il ne s’agit pas d’une autre méthode, mais de plusieurs essais du même algorithme.
def depart_alternatif(instance, depart, graine, etapes=4):
    """Perturbe une solution faisable sans jamais accepter une tournée infaisable."""
    rng = random.Random(graine)
    courant = normaliser_solution(depart)
    for _ in range(etapes):
        # Même méthode de génération pour toutes les tailles.
        voisins = voisins_echantillonnes(
            instance, courant, 80, rng.randrange(10**9)
        )
        faisables = [v for v in voisins if evaluer_solution(instance, v) is not None and v != courant]
        if faisables:
            courant = rng.choice(sorted(faisables))
    return courant


def comparer_departs(instance, nb_departs=3, iterations=60, taille_tabou=5,
                      limite_voisins=None, budget_secondes=None):
    """Le premier départ glouton et (nb_departs-1) perturbations du même cas."""
    debut_initialisation = time.perf_counter()
    depart = solution_initiale_gloutonne(instance)
    essais = []
    deja_vus = set()
    for numero in range(nb_departs):
        t0 = debut_initialisation if numero == 0 else time.perf_counter()
        initial = depart if numero == 0 else depart_alternatif(instance, depart, graine=100+numero)
        if initial in deja_vus:
            continue
        deja_vus.add(initial)
        solution, _, _, infos = recherche_tabou(
            instance, initial, taille_tabou=taille_tabou, iter_max=iterations,
            limite_voisins=limite_voisins, graine=numero, budget_secondes=budget_secondes)
        duree = evaluer_solution(instance, solution)["cout_total"]
        essais.append({"duree": duree, "temps_s": time.perf_counter() - t0,
                       "solution": solution, "iterations": infos["nb_iterations_effectuees"]})
    return essais


# Archivage : séparer les données de l’instance et ses résultats pour pouvoir
# retrouver ce qui a été testé et sa solution, même après une interruption.
def donnees_a_archiver(instance):
    """Données nécessaires pour retrouver l'instance sans la régénérer."""
    return {
        "nom": instance["nom"], "clients": list(instance["clients"]),
        "capacite": instance["capacite"], "nb_camions_max": instance["nb_camions_max"],
        "coordonnees": {str(k): list(v) for k, v in instance["coordonnees"].items()},
        "demandes": instance["demandes"], "services": instance["services"],
        "fenetres": {str(k): list(v) for k, v in instance["fenetres"].items()},
        "temps_trajet": [[i, j, t] for (i, j), t in instance["temps_trajet"].items()],
        "solution_temoin": [list(r) for r in instance.get("solution_initiale", [])],
        "heure_depart": instance["heure_depart"],
        "heure_limite_depot": instance.get("heure_limite_depot"),
    }


def nettoyer_json(valeur):
    """Convertit les valeurs numpy et les NaN des statistiques en JSON standard."""
    if isinstance(valeur, np.generic):
        valeur = valeur.item()
    if isinstance(valeur, float) and not math.isfinite(valeur):
        return None
    if isinstance(valeur, dict):
        return {str(k): nettoyer_json(v) for k, v in valeur.items()}
    if isinstance(valeur, (tuple, list)):
        return [nettoyer_json(v) for v in valeur]
    return valeur


def sauvegarder_resultat_instance(instance, ligne, solution, categorie="synthetique"):
    """Un dossier par instance terminée : données, solution et tracé PNG."""
    dossier = DOSSIER_INSTANCES / instance["nom"]
    dossier.mkdir(parents=True, exist_ok=True)
    fichier_donnees = dossier / "instance.json"
    if not fichier_donnees.exists():
        fichier_donnees.write_text(
            json.dumps(nettoyer_json(donnees_a_archiver(instance)), indent=2,
                       ensure_ascii=False, allow_nan=False), encoding="utf-8")
    resultat = {"categorie": categorie, "resume": nettoyer_json(ligne),
                "tournees": nettoyer_json(solution)}
    (dossier / "resultats.json").write_text(
        json.dumps(resultat, indent=2, ensure_ascii=False, allow_nan=False), encoding="utf-8")

    # Un seul tracé sauvegardé, sans afficher un graphique supplémentaire dans le notebook.
    fig, ax = plt.subplots(figsize=(7, 5))
    coord = instance["coordonnees"]
    ax.scatter(*coord[0], marker="s", color="black", s=85, label="Dépôt", zorder=3)
    for numero, route in enumerate(normaliser_solution(solution), start=1):
        points = [0] + list(route) + [0]
        ax.plot([coord[i][0] for i in points], [coord[i][1] for i in points],
                marker="o", markersize=3, linewidth=1.2, label=f"Camion {numero}")
    ax.set(title=f"Tournées — {instance['nom']}", xlabel="X", ylabel="Y")
    ax.set_aspect("equal", adjustable="datalim")
    ax.grid(alpha=.2)
    ax.legend(fontsize=8)
    fig.tight_layout()
    fig.savefig(dossier / "tournees.png", dpi=140, bbox_inches="tight")
    plt.close(fig)
    return dossier


# Reprise : un résultat déjà présent dans resultats.json compte comme terminé,
# même si le CSV de synthèse a été déplacé ou supprimé.
def reprendre_resultats_archives(categorie):
    """Récupère les résultats des dossiers si le CSV a été déplacé ou supprimé."""
    lignes = []
    for fichier in DOSSIER_INSTANCES.glob("*/resultats.json"):
        try:
            contenu = json.loads(fichier.read_text(encoding="utf-8"))
            if contenu.get("categorie") == categorie:
                lignes.append(contenu["resume"])
        except (OSError, ValueError, KeyError) as erreur:
            print(f"Archive ignorée ({fichier}) : {erreur}")
    return pd.DataFrame(lignes)
````

````python
# Nombre de NOUVELLES instances à traiter : 3 pour une exécution courte,
# 24 pour essayer toute la campagne (les cas déjà sauvegardés seront ignorés).
NOUVELLES_INSTANCES_PAR_LANCEMENT = 24
DEPARTS_MULTI_START = 3          # au plus 3 départs faisables sur chaque même instance
ITERATIONS_CAMPAGNE = 60        # budget commun d'itérations de la recherche tabou
CALCULER_LP_PETITES = True      # False si PuLP/CBC n'est pas installé ou trop lent

FICHIER_CAMPAGNE = DOSSIER_RESULTATS / "campagne_cvrtw.csv"
# 24 scénarios = 4 tailles × 3 géographies × 2 largeurs de fenêtres.
tailles = (8, 10, 12, 15)
scenarios = [(n, profil, tw) for n in tailles for profil in ("C", "R", "RC") for tw in (1, 2)]
ancien = pd.read_csv(FICHIER_CAMPAGNE) if FICHIER_CAMPAGNE.exists() else pd.DataFrame()
# Reprendre les résultats archivés avant de générer quoi que ce soit.
archives = reprendre_resultats_archives("synthetique")
if len(archives):
    ancien = pd.concat([ancien, archives], ignore_index=True)
if len(ancien):
    ancien = ancien.drop_duplicates(subset="instance", keep="last")
    ancien.to_csv(FICHIER_CAMPAGNE, index=False)
realises = set(ancien["instance"]) if "instance" in ancien else set()

nouveaux = 0
for n, profil, tw in scenarios:
    if nouveaux >= NOUVELLES_INSTANCES_PAR_LANCEMENT:
        break
    graine = 1000 + n * 100 + {"C": 1, "R": 2, "RC": 3}[profil] * 10 + tw
    nom_instance = f"ALEA-{FAMILLE_REFERENCE}-{profil}-TW{tw}-{n}-seed{graine}"
    if nom_instance in realises:  # vérification AVANT génération, calcul et création du dossier
        continue

    # Nouvelle instance uniquement : même nom = mêmes paramètres et même graine.
    inst = generer_instance_aleatoire(n, graine, profil=profil, type_fenetres=tw)
    try:
        # Même budget maximal de voisins pour toutes les tailles.
        # Cela rend la comparaison des temps plus cohérente.
        limite = 140
        essais = comparer_departs(inst, nb_departs=DEPARTS_MULTI_START,
                                 iterations=ITERATIONS_CAMPAGNE, limite_voisins=limite)
        premier, meilleur_essai = essais[0], min(essais, key=lambda e: e["duree"])
        simple, meilleur = premier["duree"], meilleur_essai["duree"]
        t_exact, t_lp = float("nan"), float("nan")
        optimum, borne = float("nan"), float("nan")
        # Références exactes volontairement réservées aux plus petites tailles :
        # les autres lignes garderont un GAP absent (NaN), et non un faux zéro.
        if n <= 10:
            t0 = time.perf_counter()
            optimum = optimum_exact_dp(inst)
            t_exact = time.perf_counter() - t0
        if CALCULER_LP_PETITES and pulp is not None and n <= 10:
            t0 = time.perf_counter()
            try:
                borne = calculer_borne_inferieure_lp(inst, temps_limite=15)
            except Exception as exc:
                print(inst["nom"], "borne LP non disponible :", exc)
            t_lp = time.perf_counter() - t0
        # Mesurer les GAP sur la MEILLEURE solution du multi-start de cette instance.
        # Un très petit GAP négatif dû à la précision flottante est affiché comme 0.
        gap_exact = 100 * (meilleur - optimum) / optimum if pd.notna(optimum) and optimum > 0 else float("nan")
        gap_lp = 100 * (meilleur - borne) / borne if pd.notna(borne) and borne > 0 else float("nan")
        if pd.notna(gap_exact) and abs(gap_exact) < 1e-7:
            gap_exact = 0.0
        if pd.notna(gap_lp) and abs(gap_lp) < 1e-7:
            gap_lp = 0.0
        ligne = {
            "instance": inst["nom"], "n": n, "profil": profil, "TW": tw,
            "duree_tabou": simple, "duree_multi_start": meilleur,
            "departs_distincts": len(essais), "temps_tabou_s": premier["temps_s"],
            "temps_multi_s": sum(e["temps_s"] for e in essais),
            "optimum_exact": optimum, "borne_LP": borne, "temps_exact_s": t_exact,
            "temps_LP_s": t_lp,
            "GAP_exact_%": gap_exact,
            "GAP_LP_%": gap_lp,
            "faisable": evaluer_solution(inst, meilleur_essai["solution"]) is not None
        }
        if not ligne["faisable"]:
            raise RuntimeError("Solution finale non faisable : résultat non sauvegardé.")
        sauvegarder_resultat_instance(inst, ligne, meilleur_essai["solution"])
        ancien = pd.concat([ancien, pd.DataFrame([ligne])], ignore_index=True)
        ancien.to_csv(FICHIER_CAMPAGNE, index=False)
        realises.add(inst["nom"])
        nouveaux += 1
    except (ValueError, RuntimeError) as erreur:
        print(inst["nom"], "— test non abouti :", erreur)

if len(ancien):
    print("Campagne synthétique — une ligne par instance distincte (sans le cas témoin à 442) :")
    display(ancien[["instance", "n", "profil", "TW", "duree_tabou", "duree_multi_start",
                   "GAP_exact_%", "GAP_LP_%", "temps_tabou_s"]].round(2).set_index("instance"))
# Bilan lisible : ne pas confondre une campagne partielle et un résultat définitif.
nb_termines = len(realises.intersection({
    f"ALEA-{FAMILLE_REFERENCE}-{profil}-TW{tw}-{n}-seed{1000+n*100+{'C':1,'R':2,'RC':3}[profil]*10+tw}"
    for n, profil, tw in scenarios
}))
print(f"Cette exécution : {nouveaux} nouvelle(s) instance(s) calculée(s).")
print(f"Campagne synthétique : {nb_termines}/{len(scenarios)} terminées ; "
      f"{len(scenarios)-nb_termines} restantes (minimum demandé : 20).")
if nb_termines < 20:
    print(f"Résultats encore provisoires : il manque {20-nb_termines} instance(s) "
          "pour atteindre le minimum demandé. Relancer cette cellule pour poursuivre.")
if len(ancien):
    nb_gains = (ancien['duree_multi_start'] < ancien['duree_tabou']-1e-7).sum()
    print(f"Multi-start : amélioration sur {nb_gains} instance(s) parmi {len(ancien)}. "
          "Une égalité signifie qu'aucun gain n'a été observé sur cet essai.")
````

### Ce que nous apprend cette première campagne

Le tableau **ne reprend pas l'instance de référence** : il présente les autres cas, identifiés individuellement par leur nom. Les colonnes `duree_tabou` et `duree_multi_start` sont donc comparables **ligne par ligne**, car elles concernent exactement les mêmes clients et contraintes. Si les deux valeurs sont égales, les départs supplémentaires n'ont **pas apporté d'amélioration observée**, même s'ils ont augmenté le temps de calcul.

Le **GAP exact** est disponible uniquement quand l'optimum a été calculé ; le **GAP LP** dépend de la disponibilité de la relaxation. Une valeur absente signifie **« non calculé »**, pas 0 %. Les tailles 12 et 15 ne reçoivent pas automatiquement un optimum exact dans cette campagne, car le coût du calcul devient plus important. Pour chaque nouvelle instance, nous conservons sa solution et son graphique dans son propre dossier ; le CSV sert à suivre l'avancement.

<a id="partie7-8"></a>
## 7.8. Limite du calcul exact : pourquoi passer aux références externes ?

Sur nos petites instances, l'optimum exact sert à **valider** la qualité de la recherche tabou, tandis que la relaxation LP fournit seulement une **borne** sur la durée optimale. Mais calculer une preuve d'optimalité implique une explosion combinatoire lorsque le nombre de clients augmente. Notre programmation dynamique est donc volontairement limitée aux petits cas ; par ailleurs, la relaxation LP dépend du temps accordé au solveur et sa borne peut rester éloignée de l'optimum.

Ces deux références ne suffisent pas à juger la méthode sur des tournées réalistes. Il faut un **repère externe** : les instances Solomon et Homberger–Gehring et leurs *Best Known Solutions* (BKS), disponibles avec des routes de référence. Les BKS ne prouvent pas nécessairement l'optimalité ; elles indiquent le meilleur résultat de référence fourni.

Nous continuons à **optimiser la durée totale**. Pour comparer aux BKS classiques, souvent exprimées en **distance**, nous calculons ensuite séparément la distance de nos tournées :

$$GAP_{BKS}=100\times\frac{D_{tabou}-D_{BKS}}{D_{BKS}}.$$ 

Nous conservons **deux comparaisons** : l’écart de durée à la tournée de référence réévaluée (même objectif que le nôtre), et le GAP de distance aux BKS des instances de référence. Le premier est utile pour notre objectif, le second pour le benchmark. Nous affichons aussi le **nombre de camions**, car les références Solomon privilégient normalement d'abord ce critère, puis la distance. Un GAP distance de 10 % n'est pas une garantie d'optimalité sur notre fonction objectif en durée. Les distances de notre solution et des tournées de référence sont recalculées avec la même matrice, pour éviter de comparer des conventions d’arrondi différentes. Le coût du fichier `.sol` est conservé pour vérifier cette convention. Le seuil de 7 à 10 % concerne ici **la comparaison externe**, pas le GAP LP.

<a id="partie7-9"></a>
## 7.9. Régler les paramètres avant les benchmarks

Avant de comparer notre méthode aux solutions de référence, nous devons fixer **un nombre d'itérations** et **une taille de liste tabou**. Nous ne pouvons pas simplement supposer qu'augmenter ces deux valeurs améliore systématiquement le résultat : davantage d'itérations augmente normalement l'effort de calcul, tandis qu'une liste tabou plus longue peut aider à éviter des retours en arrière, ou au contraire bloquer des mouvements utiles.

Nous réalisons donc un **petit plan factoriel** sur des instances synthétiques C, R et RC, distinctes du cas de validation à la petite instance du début et des benchmarks externes. Pour isoler l'effet des paramètres, nous gardons le **même départ**, les **mêmes instances**, le **même nombre de voisins échantillonnés** et les **mêmes graines** pour toutes les configurations. Le test est répété deux fois afin de limiter l'influence du hasard.

Les durées brutes diffèrent d'une instance à l'autre. Pour les comparer sur une même courbe, nous calculons un *écart au meilleur résultat observé pendant le réglage*, par instance. **Ce n'est ni un GAP à l'optimum ni un GAP aux BKS.** Nous retenons la configuration la plus rapide parmi celles dont l'écart moyen reste à moins de **0,5 point** du meilleur réglage testé. Ces paramètres seront ensuite **figés** pour toutes les comparaisons Solomon et Homberger.


````python
# Paramètres testés pour la recherche tabou
ITERATIONS_TESTEES = [60, 120, 240, 400]
TAILLES_TABOU_TESTEES = [5, 10, 20, 40]

# Deux répétitions pour limiter l'effet de l'aléatoire
GRAINES_REGLAGE = [11, 23]

# Nombre de voisins étudiés à chaque itération
VOISINS_REGLAGE = 120

# Trois instances de 100 clients, avec différentes
# répartitions géographiques et fenêtres temporelles
SCENARIOS_REGLAGE = [
    (100, "C", 1, 301),
    (100, "R", 1, 302),
    (100, "RC", 2, 303),
]

FICHIER_REGLAGE = DOSSIER_RESULTATS / "reglage_parametres.csv"

# Les résultats déjà sauvegardés sont réutilisés lors d'une relance du notebook.
mesures = (pd.read_csv(FICHIER_REGLAGE) if FICHIER_REGLAGE.exists()
           else pd.DataFrame())
cles_deja_testes = set()
if len(mesures):
    cles_deja_testes = set(zip(mesures["instance"], mesures["iterations"],
                               mesures["taille_tabou"], mesures["graine"],
                               mesures["limite_voisins"]))

# Une unité de progression = un essai (y compris un essai déjà archivé).
# Cela donne une barre réellement représentative des 54 combinaisons possibles.
NB_ESSAIS_REGLAGE = (len(SCENARIOS_REGLAGE) * len(GRAINES_REGLAGE)
                    * len(ITERATIONS_TESTEES) * len(TAILLES_TABOU_TESTEES))
barre_reglage = BarreProgression(NB_ESSAIS_REGLAGE, "Réglage des paramètres")
for n, profil, tw, graine_inst in SCENARIOS_REGLAGE:
    inst = generer_instance_aleatoire(n, graine_inst, profil=profil, type_fenetres=tw)
    depart = solution_initiale_gloutonne(inst)
    for repetitions in GRAINES_REGLAGE:
        for iterations in ITERATIONS_TESTEES:
            for taille in TAILLES_TABOU_TESTEES:
                cle = (inst["nom"], iterations, taille, repetitions, VOISINS_REGLAGE)
                if cle in cles_deja_testes:
                    barre_reglage.update(1)  # Déjà enregistré : pas de nouveau calcul.
                    continue
                t0 = time.perf_counter()
                solution, _, _, _ = recherche_tabou(
                    inst, depart, taille_tabou=taille, iter_max=iterations,
                    limite_voisins=VOISINS_REGLAGE, graine=repetitions)
                duree = evaluer_solution(inst, solution)["cout_total"]
                ligne = {"instance":inst["nom"], "n":n, "profil":profil, "TW":tw,
                         "iterations":iterations, "taille_tabou":taille,
                         "graine":repetitions, "limite_voisins":VOISINS_REGLAGE,
                         "duree":duree, "temps_s":time.perf_counter()-t0}
                mesures = pd.concat([mesures, pd.DataFrame([ligne])], ignore_index=True)
                mesures.to_csv(FICHIER_REGLAGE, index=False)
                cles_deja_testes.add(cle)
                barre_reglage.update(1)
barre_reglage.close()

if mesures.empty:
    raise RuntimeError("Aucun réglage réalisé : vérifier les scénarios.")

# Réutiliser uniquement les résultats correspondant aux paramètres ACTUELS.
selection = mesures[
    mesures["iterations"].isin(ITERATIONS_TESTEES) &
    mesures["taille_tabou"].isin(TAILLES_TABOU_TESTEES) &
    mesures["graine"].isin(GRAINES_REGLAGE) &
    (mesures["limite_voisins"] == VOISINS_REGLAGE) &
    mesures["instance"].isin([f"ALEA-{FAMILLE_REFERENCE}-{p}-TW{tw}-{n}-seed{seed}"
                              for n, p, tw, seed in SCENARIOS_REGLAGE])
].copy()

# Écart relatif au meilleur résultat observé, pour chaque instance (repère empirique).
selection["meilleure_observee"] = selection.groupby("instance")["duree"].transform("min")
selection["ecart_observe_%"] = (100 *
    (selection["duree"] - selection["meilleure_observee"]) / selection["meilleure_observee"])
stat_reglage = (selection.groupby(["iterations", "taille_tabou"], as_index=False)
               .agg(ecart_moyen_pct=("ecart_observe_%", "mean"),
                    temps_moyen_s=("temps_s", "mean"),
                    nb_essais=("duree", "size")))
nb_attendus = len(SCENARIOS_REGLAGE) * len(GRAINES_REGLAGE)
stat_reglage = stat_reglage[stat_reglage["nb_essais"] == nb_attendus]
if stat_reglage.empty:
    raise RuntimeError("Réglage incomplet : toutes les combinaisons doivent être mesurées.")

# Compromis : la plus rapide des configurations à moins de 0,5 point du meilleur écart.
meilleur_ecart = stat_reglage["ecart_moyen_pct"].min()
compromis = stat_reglage[
    stat_reglage["ecart_moyen_pct"] <= meilleur_ecart + 0.5
].sort_values(["temps_moyen_s", "ecart_moyen_pct"]).iloc[0]
PARAM_ITERATIONS = int(compromis["iterations"])
PARAM_TAILLE_TABOU = int(compromis["taille_tabou"])
PARAM_LIMITE_VOISINS = VOISINS_REGLAGE
print(f"Réglage : {len(selection)} essais sur {len(SCENARIOS_REGLAGE)} instances distinctes.")
print(f"Paramètres retenus : {PARAM_ITERATIONS} itérations, "
      f"liste tabou de {PARAM_TAILLE_TABOU}, {PARAM_LIMITE_VOISINS} voisins/itération.")
display(stat_reglage.sort_values("ecart_moyen_pct").round(2))

# Deux figures : qualité et temps, chacune en fonction du paramètre étudié.
for colonne, nom, fichier in [
    ("iterations", "Nombre d'itérations", "influence_iterations.png"),
    ("taille_tabou", "Taille de la liste tabou", "influence_liste_tabou.png")
]:
    stats = (selection.groupby(colonne, as_index=False)
             .agg(ecart_moyen=("ecart_observe_%", "mean"),
                  temps_moyen=("temps_s", "mean")))
    fig, axes = plt.subplots(1, 2, figsize=(10, 3.4))
    axes[0].plot(stats[colonne], stats["ecart_moyen"], "o-")
    axes[0].set(xlabel=nom, ylabel="Écart moyen au meilleur observé (%)",
                title="Qualité des solutions")
    axes[1].plot(stats[colonne], stats["temps_moyen"], "o-")
    axes[1].set(xlabel=nom, ylabel="Temps moyen par recherche (s)",
                title="Temps de calcul")
    for ax in axes:
        ax.grid(alpha=.25)
    fig.tight_layout()
    fig.savefig(DOSSIER_FIGURES / fichier, dpi=160, bbox_inches="tight")
    plt.show()
print("Les courbes montrent des tendances mesurées sur le jeu de réglage : elles ne "
      "garantissent pas une amélioration monotone pour chaque instance.")
````

#### Interprétation des résultats

Les résultats montrent que **le nombre d'itérations influence directement la qualité des solutions**. En passant de 60 à 400 itérations, l'écart moyen au meilleur résultat observé diminue d'environ 5,6 % à 2,2 %. En contrepartie, le temps de calcul augmente presque linéairement. Les variations du temps de calcul ne suivent pas de tendance claire et peuvent notamment provenir des différences de parcours dans la recherche.

Ces résultats restent toutefois liés aux instances étudiées et ne garantissent pas le même comportement sur des problèmes plus grands. **Nous retenons donc provisoirement 400 itérations et une liste tabou de taille 20**, qui offrent les meilleurs résultats parmi les valeurs testées. Les calculs exacts deviennent trop coûteux pour les grandes tailles. Notre fonction objectif reste la minimisation de la durée totale des tournées, comprenant les déplacements, les services et les attentes. Pour conserver la cohérence avec notre problématique, nous recalculons donc la durée totale des tournées de référence avec notre propre évaluateur.

Nous distinguons ainsi deux indicateurs : l'écart relatif en durée, qui permet de comparer notre solution à la BKS réévaluée selon notre fonction objectif, et le GAP en distance, qui permet de comparer nos résultats aux références publiées. Ce dernier reste un indicateur complémentaire. La BKS n'étant pas nécessairement optimale en durée, l'écart en durée ne représente pas un écart à l'optimum prouvé.
````python
# Dépend des cellules précédentes : evaluer_solution, normaliser_solution,
# recherche_tabou, BarreProgression, nettoyer_json, DOSSIER_RESULTATS,
# DOSSIER_INSTANCES et DOSSIER_FIGURES.
# Installer si nécessaire : %pip install vrplib ipywidgets
try:
    import vrplib
except ImportError:
    vrplib = None
from IPython.display import Image as ImageJupyter

# ========================= CONFIGURATION =========================
# À adapter : dossiers des fichiers d'instances (.txt) et des BKS (.sol).
CHEMINS = {
    "Solomon": (Path("instances/Solomon"), Path("solutions/Solomon-solutions")),
    "Homberger": (Path("instances/Homberger"), Path("solutions/Homberger-solutions")),
}
# Noms exacts des fichiers, sans extension.
INSTANCES_BENCHMARK = {
    "Solomon": ["C101", "C201", "R101", "R201", "RC101", "RC201"],
    "Homberger": ["C1_2_1", "R1_4_1", "RC1_6_1", "C2_8_1", "RC2_10_5"],
}
NOUVEAUX_BENCHMARKS_PAR_LANCEMENT = 1
NB_DEPARTS_BENCHMARK = 3
BUDGET_BENCHMARK_S = 300           # Maximum souple pour les 3 départs réunis.
PARAM_ITERATIONS = 400
PARAM_TAILLE_TABOU = 20
PARAM_LIMITE_VOISINS = 120

ETUDIER_LISTE_TABOU = True
TAILLES_TABOU_A_COMPARER = [5, 10, 20, 40]
BUDGET_ETUDE_TABOU_S = 180         # Budget supplémentaire, réparti entre les tailles.
AFFICHER_PROGRESSIONS = True
# Même convention de calcul de trajet pour nos tournées et la BKS.
CONVENTION_DISTANCE = "double"     # Ou "trunc1" selon le jeu de références.
FICHIER_BENCHMARK = DOSSIER_RESULTATS / "comparaison_benchmarks.csv"


class TempsTrajetMatrice:
    """Évite un dictionnaire d'un million d'arcs pour 1000 clients."""
    def __init__(self, matrice):
        self.matrice = matrice
    def __getitem__(self, arc):
        return float(self.matrice[arc])


class ArcsComplets:
    def __init__(self, n):
        self.n = n
    def __contains__(self, arc):
        i, j = arc
        return i != j and 0 <= i < self.n and 0 <= j < self.n


def charger_benchmark(famille, nom):
    dossier, _ = CHEMINS[famille]
    donnees = vrplib.read_instance(str(dossier / f"{nom}.txt"),
                                   instance_format="solomon")
    xy = np.asarray(donnees["node_coord"], dtype=float)
    matrice = np.asarray(donnees["edge_weight"], dtype=float)
    if CONVENTION_DISTANCE == "trunc1":
        matrice = np.floor(matrice * 10 + 1e-9) / 10.0
    elif CONVENTION_DISTANCE != "double":
        raise ValueError("CONVENTION_DISTANCE : choisir 'double' ou 'trunc1'.")
    fenetres = np.asarray(donnees["time_window"], dtype=float)
    n = len(xy) - 1
    return {
        "nom": nom, "clients": tuple(range(1, n + 1)),
        "sommets": tuple(range(n + 1)), "arcs": ArcsComplets(n + 1),
        "coordonnees": {i: tuple(xy[i]) for i in range(n + 1)},
        "temps_trajet": TempsTrajetMatrice(matrice),
        "demandes": {i: float(donnees["demand"][i]) for i in range(1, n + 1)},
        "services": {i: float(donnees["service_time"][i]) for i in range(1, n + 1)},
        "fenetres": {i: tuple(map(float, fenetres[i])) for i in range(1, n + 1)},
        "capacite": float(donnees["capacity"]),
        "nb_camions_max": int(donnees["vehicles"]),
        "heure_depart": float(fenetres[0, 0]),
        "heure_limite_depot": float(fenetres[0, 1]),
    }


def distance_solution(instance, solution):
    """Indicateur secondaire : distance comparable au coût officiel de la BKS."""
    return sum(instance["temps_trajet"][(i, j)]
               for route in solution
               for i, j in zip((0,) + tuple(route), tuple(route) + (0,)))


def ecart_duree(duree, duree_reference):
    """Écart à la durée de la solution BKS, PAS à un optimum prouvé en durée."""
    return 100 * (duree - duree_reference) / duree_reference


def depart_rapide_benchmark(instance, graine=0):
    """Construit un départ faisable sans utiliser les tournées de référence."""
    rng = random.Random(graine)
    restants = set(instance["clients"])
    tournees = []
    while restants:
        if len(tournees) >= instance["nb_camions_max"]:
            raise ValueError("Impossible de construire un départ avec cette flotte.")
        route, courant = [], 0
        charge, heure = 0.0, instance["heure_depart"]
        while restants:
            choix = None
            for client in restants:
                if charge + instance["demandes"][client] > instance["capacite"] + 1e-7:
                    continue
                trajet = instance["temps_trajet"][(courant, client)]
                arrivee = heure + trajet
                a, b = instance["fenetres"][client]
                debut_service = max(arrivee, a)
                fin_service = debut_service + instance["services"][client]
                if debut_service > b + 1e-7:
                    continue
                if (fin_service + instance["temps_trajet"][(client, 0)]
                        > instance["heure_limite_depot"] + 1e-7):
                    continue
                score = (trajet + 0.03 * max(0, b - debut_service)
                         + rng.random() * 0.03 * max(1, trajet))
                if choix is None or score < choix[0]:
                    choix = (score, client, fin_service)
            if choix is None:
                break
            _, client, heure = choix
            route.append(client)
            charge += instance["demandes"][client]
            courant = client
            restants.remove(client)
        if not route:
            raise ValueError("Aucun client supplémentaire ne peut être livré.")
        tournees.append(route)
    depart = normaliser_solution(tournees)
    if evaluer_solution(instance, depart) is None:
        raise ValueError("Départ glouton infaisable.")
    return depart


def comparer_benchmark(instance, famille):
    """Multi-start principal : choisit la meilleure DURÉE, jamais la distance."""
    debut = time.perf_counter()
    essais = []
    for numero in range(NB_DEPARTS_BENCHMARK):
        reste = BUDGET_BENCHMARK_S - (time.perf_counter() - debut)
        if reste <= 0:
            break
        try:
            depart = depart_rapide_benchmark(instance, graine=101 + 47 * numero)
        except ValueError as exc:
            print(f"  Départ {numero + 1} : {exc}")
            continue
        temps_disponible = BUDGET_BENCHMARK_S - (time.perf_counter() - debut)
        if temps_disponible <= 0:
            break
        budget = temps_disponible / (NB_DEPARTS_BENCHMARK - numero)
        t0 = time.perf_counter()
        routes, _, historique, infos = recherche_tabou(
            instance, depart, taille_tabou=PARAM_TAILLE_TABOU,
            iter_max=PARAM_ITERATIONS, limite_voisins=PARAM_LIMITE_VOISINS,
            graine=11 + numero, budget_secondes=budget,
            progression=AFFICHER_PROGRESSIONS,
            libelle_progression=f"{famille} {instance['nom']} — départ {numero+1}"
        )
        ev = evaluer_solution(instance, routes)
        if ev is None:
            raise RuntimeError("La recherche a renvoyé une solution infaisable.")
        essais.append({
            "routes": routes, "duree": ev["cout_total"],
            "depart": depart, "historique": historique,
            "temps_s": time.perf_counter() - t0,
            "iterations": infos["nb_iterations_effectuees"],
        })
    if not essais:
        raise ValueError("Aucun départ faisable obtenu ; GAP non calculé.")
    return min(essais, key=lambda e: e["duree"]), essais[0]["depart"], len(essais), time.perf_counter() - debut


def etudier_liste_tabou(instance, famille, depart, duree_bks):
    """Comparer les tailles avec le MÊME départ, la MÊME graine et le MÊME budget."""
    if not ETUDIER_LISTE_TABOU:
        return [], pd.DataFrame()
    if not TAILLES_TABOU_A_COMPARER:
        raise ValueError("Aucune taille tabou à comparer.")
    budget = BUDGET_ETUDE_TABOU_S / len(TAILLES_TABOU_A_COMPARER)
    essais = []
    for taille in TAILLES_TABOU_A_COMPARER:
        t0 = time.perf_counter()
        routes, _, historique, infos = recherche_tabou(
            instance, depart, taille_tabou=taille,
            iter_max=PARAM_ITERATIONS, limite_voisins=PARAM_LIMITE_VOISINS,
            graine=11, budget_secondes=budget,
            progression=AFFICHER_PROGRESSIONS,
            libelle_progression=f"{famille} {instance['nom']} — tabou {taille}"
        )
        ev = evaluer_solution(instance, routes)
        if ev is None:
            raise RuntimeError("Résultat de sensibilité infaisable.")
        essais.append({
            "taille_tabou": taille, "duree": ev["cout_total"],
            "ecart_duree_BKS_%": ecart_duree(ev["cout_total"], duree_bks),
            "temps_s": time.perf_counter() - t0,
            "iterations": infos["nb_iterations_effectuees"],
            "historique": historique,
        })
    donnees = pd.DataFrame([{k: e[k] for k in (
        "taille_tabou", "duree", "ecart_duree_BKS_%", "temps_s", "iterations")}
        for e in essais])
    return essais, donnees


def afficher_et_enregistrer(fig, chemin):
    fig.tight_layout()
    fig.savefig(chemin, dpi=160, bbox_inches="tight")
    plt.show()
    plt.close(fig)


def visualiser_benchmark(instance, famille, nom, dossier, meilleur, duree_bks,
                          essais_tabou, tableau_tabou, ligne):
    """Carte et courbes : sauvegarde PNG ET affichage dans Jupyter."""
    # 1. Carte des tournées sélectionnées en minimisant la DURÉE.
    fig, ax = plt.subplots(figsize=(11, 7))
    xy = instance["coordonnees"]
    ax.scatter([xy[i][0] for i in instance["clients"]],
               [xy[i][1] for i in instance["clients"]],
               s=9, color="gray", alpha=0.35, label="Clients")
    ax.scatter(*xy[0], marker="s", s=90, color="black", zorder=3, label="Dépôt")
    couleurs = plt.get_cmap("tab20")
    for k, route in enumerate(meilleur["routes"]):
        points = [0] + list(route) + [0]
        ax.plot([xy[i][0] for i in points], [xy[i][1] for i in points],
                color=couleurs(k % 20), linewidth=0.9, alpha=0.7)
    ax.set(title=(f"{famille} {nom} — {len(instance['clients'])} clients, "
                  f"{len(meilleur['routes'])} véhicules\n"
                  f"Durée = {meilleur['duree']:.2f} | Durée BKS = {duree_bks:.2f} | "
                  f"Écart durée = {ligne['ecart_duree_BKS_%']:.2f} %"),
           xlabel="X", ylabel="Y")
    ax.set_aspect("equal", adjustable="datalim")
    ax.grid(alpha=0.15)
    ax.legend()
    afficher_et_enregistrer(fig, dossier / f"tournees_{ID_CONFIGURATION}.png")

    # 2. Évolution des solutions pendant la recherche (durée et écart associé).
    fig, axes = plt.subplots(1, 2, figsize=(13, 4))
    for essai in essais_tabou:
        y = essai["historique"]
        axes[0].plot(range(len(y)), y, label=f"Tabou {essai['taille_tabou']}")
        axes[1].plot(range(len(y)), [ecart_duree(v, duree_bks) for v in y],
                     label=f"Tabou {essai['taille_tabou']}")
    # Le meilleur multi-start est sélectionné sur la durée finale ; sa courbe
    # représente son départ particulier, et NON la somme des trois trajectoires.
    y = meilleur["historique"]
    axes[0].plot(range(len(y)), y, "k--", linewidth=1.6,
                 label="Meilleur départ multi-start")
    axes[1].plot(range(len(y)), [ecart_duree(v, duree_bks) for v in y],
                 "k--", linewidth=1.6, label="Meilleur départ multi-start")
    axes[0].set(title="Meilleure durée trouvée", ylabel="Durée totale")
    axes[1].set(title="Écart à la durée BKS recalculée", ylabel="Écart (%)")
    axes[1].axhline(0, color="gray", linestyle=":")
    for ax in axes:
        ax.set_xlabel("Itérations effectuées")
        ax.grid(alpha=0.25)
        ax.legend(fontsize=8)
    fig.suptitle(f"Convergence — {famille} {nom}")
    afficher_et_enregistrer(fig, dossier / f"convergence_{ID_CONFIGURATION}.png")

    # 3. Sensibilité de la liste tabou et temps de calcul.
    if not tableau_tabou.empty:
        fig, axes = plt.subplots(1, 3, figsize=(14, 4))
        x = tableau_tabou["taille_tabou"]
        for ax, colonne, titre, ylabel in [
            (axes[0], "duree", "Durée", "Meilleure durée"),
            (axes[1], "ecart_duree_BKS_%", "Écart à la BKS", "Écart durée (%)"),
            (axes[2], "temps_s", "Coût de calcul", "Temps (s)"),
        ]:
            ax.plot(x, tableau_tabou[colonne], "o-")
            ax.set(title=titre, xlabel="Taille de la liste tabou", ylabel=ylabel)
            ax.set_xticks(TAILLES_TABOU_A_COMPARER)
            ax.grid(alpha=0.25)
        fig.suptitle(f"Sensibilité de la mémoire tabou — {famille} {nom}")
        afficher_et_enregistrer(fig, dossier / f"sensibilite_tabou_{ID_CONFIGURATION}.png")


# ========================= EXÉCUTION / REPRISE =========================
# Version distincte pour ne pas mélanger d'anciens résultats en GAP distance.
ID_CONFIGURATION = (f"duree_v3_i{PARAM_ITERATIONS}_t{PARAM_TAILLE_TABOU}_"
                    f"v{PARAM_LIMITE_VOISINS}_d{NB_DEPARTS_BENCHMARK}_"
                    f"b{BUDGET_BENCHMARK_S}_{CONVENTION_DISTANCE}_"
                    f"sens{int(ETUDIER_LISTE_TABOU)}_"
                    f"ta{'-'.join(map(str, TAILLES_TABOU_A_COMPARER))}_"
                    f"bs{BUDGET_ETUDE_TABOU_S}")
ancien = pd.read_csv(FICHIER_BENCHMARK) if FICHIER_BENCHMARK.exists() else pd.DataFrame()
terminees = (set(zip(ancien["famille"], ancien["instance"], ancien["configuration"]))
             if {"famille", "instance", "configuration"}.issubset(ancien.columns) else set())

attente = []
if vrplib is None:
    print("Installer vrplib : %pip install vrplib")
else:
    for famille, noms in INSTANCES_BENCHMARK.items():
        dossier_sources, dossier_bks = CHEMINS[famille]
        for nom in noms:
            dossier = DOSSIER_INSTANCES / f"{famille}_{nom}"
            figures = [dossier / f"tournees_{ID_CONFIGURATION}.png",
                       dossier / f"convergence_{ID_CONFIGURATION}.png"]
            if ETUDIER_LISTE_TABOU:
                figures.append(dossier / f"sensibilite_tabou_{ID_CONFIGURATION}.png")
            if (famille, nom, ID_CONFIGURATION) in terminees and all(p.exists() for p in figures):
                print(f"{famille} {nom} : résultats existants, PNG réaffichés.")
                for fichier in figures:
                    display(ImageJupyter(filename=str(fichier)))
                continue
            fichier_txt = dossier_sources / f"{nom}.txt"
            fichier_sol = dossier_bks / f"{nom}.sol"
            if fichier_txt.is_file() and fichier_sol.is_file():
                attente.append((famille, nom, fichier_txt, fichier_sol))
            else:
                print(f"{famille} {nom} : .txt ou .sol manquant — test ignoré.")

selection = attente[:NOUVEAUX_BENCHMARKS_PAR_LANCEMENT]
print(f"{len(selection)} nouveau(x) benchmark(s) prévu(s) avec {ID_CONFIGURATION}.")
barre = BarreProgression(len(selection), "Instances de référence",
                         afficher=AFFICHER_PROGRESSIONS and bool(selection))
try:
    for famille, nom, fichier_txt, fichier_sol in selection:
        try:
            inst = charger_benchmark(famille, nom)
            reference = vrplib.read_solution(str(fichier_sol))
            routes_bks = normaliser_solution(reference["routes"])
            ev_bks = evaluer_solution(inst, routes_bks)
            if ev_bks is None:
                print(f"{famille} {nom} : BKS infaisable selon le modèle, pas de comparaison.")
                continue
            duree_bks = ev_bks["cout_total"]  # Trajet + service + attente.
            distance_bks = distance_solution(inst, routes_bks)
            if duree_bks <= 0 or distance_bks <= 0:
                print(f"{nom} : référence non exploitable (coût nul).")
                continue

            # La BKS officielle est généralement définie en véhicules/distance.
            # Son coût .sol ne doit PAS être pris pour une durée de tournée.
            cout_fichier = reference.get("cost")
            if cout_fichier is not None and abs(distance_bks - float(cout_fichier)) > max(0.1, .001 * distance_bks):
                print(f"{nom} : coût .sol différent de la distance recalculée : vérifier l'arrondi.")

            meilleur, depart_commun, nb_departs, temps_total = comparer_benchmark(inst, famille)
            routes = meilleur["routes"]
            distance = distance_solution(inst, routes)
            ligne = {
                "famille": famille, "instance": nom, "n": len(inst["clients"]),
                "configuration": ID_CONFIGURATION,
                "vehicules_tabou": len(routes), "vehicules_BKS": len(routes_bks),
                "duree_tabou": meilleur["duree"], "duree_BKS_recalculee": duree_bks,
                "ecart_duree_BKS_%": ecart_duree(meilleur["duree"], duree_bks),
                "distance_tabou": distance, "distance_BKS": distance_bks,
                "GAP_distance_officiel_%": 100 * (distance - distance_bks) / distance_bks,
                "vehicules_comparables": len(routes) == len(routes_bks),
                "departs_reussis": nb_departs,
                "iterations_meilleur": meilleur["iterations"],
                "temps_principal_s": temps_total,
            }
            dossier = DOSSIER_INSTANCES / f"{famille}_{nom}"
            dossier.mkdir(parents=True, exist_ok=True)
            essais_tabou, tab = etudier_liste_tabou(inst, famille, depart_commun, duree_bks)
            visualiser_benchmark(inst, famille, nom, dossier, meilleur, duree_bks,
                                essais_tabou, tab, ligne)

            # CSV des valeurs brutes et PNG conservés pour relancer sans calculer.
            pd.DataFrame({
                "iteration": range(len(meilleur["historique"])),
                "meilleure_duree": meilleur["historique"],
                "ecart_duree_BKS_%": [ecart_duree(v, duree_bks) for v in meilleur["historique"]],
            }).to_csv(dossier / f"convergence_{ID_CONFIGURATION}.csv", index=False)
            if not tab.empty:
                tab.to_csv(dossier / f"sensibilite_tabou_{ID_CONFIGURATION}.csv", index=False)
                display(tab.round(2).set_index("taille_tabou"))
            (dossier / f"resultats_{ID_CONFIGURATION}.json").write_text(
                json.dumps(nettoyer_json({
                    "resume": ligne, "tournees": routes, "tournees_BKS": routes_bks,
                    "instance_source": str(fichier_txt), "BKS_source": str(fichier_sol),
                }), ensure_ascii=False, indent=2, allow_nan=False), encoding="utf-8"
            )
            ancien = pd.concat([ancien, pd.DataFrame([ligne])], ignore_index=True)
            ancien = ancien.drop_duplicates(subset=["famille", "instance", "configuration"], keep="last")
            ancien.to_csv(FICHIER_BENCHMARK, index=False)
            terminees.add((famille, nom, ID_CONFIGURATION))
            print(f"{famille} {nom} : durée = {meilleur['duree']:.2f}, "
                  f"BKS recalculée = {duree_bks:.2f}, "
                  f"écart durée = {ligne['ecart_duree_BKS_%']:.2f} %, "
                  f"GAP distance officiel = {ligne['GAP_distance_officiel_%']:.2f} %.")
        except Exception as exc:
            print(f"{famille} {nom} : essai interrompu ({type(exc).__name__}) : {exc}")
        finally:
            barre.update(1)
finally:
    barre.close()

# Résultats principaux : écart durée pour NOTRE objectif ; GAP distance en complément.
if not ancien.empty and "configuration" in ancien:
    tableau = ancien[ancien["configuration"] == ID_CONFIGURATION].copy()
    if not tableau.empty:
        colonnes = ["famille", "instance", "n", "vehicules_tabou", "vehicules_BKS",
                    "duree_tabou", "duree_BKS_recalculee", "ecart_duree_BKS_%",
                    "GAP_distance_officiel_%", "temps_principal_s"]
        print("Écart en durée = comparaison à la BKS RÉÉVALUÉE, pas à l'optimum de durée.")
        display(tableau[colonnes].round(2).set_index(["famille", "instance"]))
        fig, ax = plt.subplots(figsize=(9, 4))
        labels = tableau["famille"].astype(str) + " " + tableau["instance"].astype(str)
        ax.bar(labels, tableau["ecart_duree_BKS_%"])
        ax.axhline(0, color="gray", linestyle="--")
        ax.set(title="Comparaison en durée à la BKS recalculée",
               xlabel="Instances", ylabel="Écart de durée (%)")
        ax.tick_params(axis="x", rotation=30)
        ax.grid(axis="y", alpha=0.2)
        afficher_et_enregistrer(fig, DOSSIER_FIGURES / "ecarts_duree_benchmarks.png")
        comparables = tableau[tableau["vehicules_comparables"] == True]
        if not comparables.empty:
            print(f"GAP distance officiel moyen (même nombre de véhicules) : "
                  f"{comparables['GAP_distance_officiel_%'].mean():.2f} % "
                  f"sur {len(comparables)} instance(s).")
        print("Attention : une BKS en distance n'est pas un optimum prouvé en durée. "
              "Un écart en durée négatif est possible.")
else:
    print("Aucun benchmark réalisé : vérifier les chemins et les fichiers BKS.")
````

**Lecture des résultats.** Une ligne correspond à une instance réellement exécutée avec les **paramètres fixés lors du réglage**. La colonne `temps_principal_s` indique le temps consacré à la recherche principale, incluant les différents départs du multi-start. Elle n'inclut pas l'étude complémentaire de sensibilité de la liste tabou. La colonne iterations_meilleur indique le nombre d'itérations réellement effectuées par le meilleur départ, ce qui permet notamment d'identifier les recherches interrompues avant la limite prévue.

Le **GAP de distance** est calculé à l'aide de la même matrice que la référence, mais notre recherche minimise la durée, c'est-à-dire trajet + service + attente. La comparaison à la BKS doit aussi considérer le **nombre de véhicules**. Les résultats sur 200–1 000 clients servent donc à apprécier les performances et les limites de passage à l'échelle ; l'objectif de moins de 10 % reste un repère empirique, non une garantie.

<a id="partie7-11"></a>
## 7.11. Bilan de notre démarche

Le cas à **12 clients** nous a permis de suivre le fonctionnement de la recherche tabou et de comparer son résultat à un **optimum prouvé**. Cette validation est nécessaire, mais elle ne suffit pas à affirmer que l'algorithme sera performant dans des conditions variées.

La campagne synthétique élargit donc les tests aux tailles, répartitions et fenêtres différentes. Nous conservons les **résultats réels**, y compris lorsqu'un multi-start n'apporte aucun gain, et les statistiques restent provisoires tant que l'échantillon est limité. La comparaison à la borne LP est utile pour situer les résultats, mais un GAP LP élevé peut surtout révéler une borne peu serrée.

Enfin, le passage à **Solomon et Homberger–Gehring/VRPLIB** permet de comparer nos distances aux BKS sur des instances reconnues, avec le nombre de véhicules utilisé. **Cette comparaison doit être exécutée avant de conclure sur le seuil de GAP visé** : le notebook ne présente pas de réussite supposée à l'avance. Une limite de notre approche reste l'utilisation de temps de trajet fixes, alors que le trafic réel peut évoluer selon l'heure ; un prolongement serait d'étudier des coûts dépendant du temps.
